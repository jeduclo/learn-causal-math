## Theory

### Layout

For a scalar function $f(x)$ of a vector $x \in \mathbb{R}^k$, the **gradient** is the column vector $\partial f/\partial x = (\partial f/\partial x_1, \dots, \partial f/\partial x_k)^{\top}$, and the **Hessian** is the $k \times k$ matrix $\partial^2 f/\partial x\,\partial x^{\top}$. For a vector function $g(x) \in \mathbb{R}^m$, the **Jacobian** $\partial g/\partial x^{\top}$ is $m \times k$.

### Four rules cover almost everything

| Function | Gradient |
|---|---|
| $a^{\top}x$ | $a$ |
| $x^{\top}Ax$ | $(A + A^{\top})x$, which is $2Ax$ if $A$ is symmetric |
| $Ax$ (vector) | Jacobian $A$ |
| $f(g(x))$ | chain rule: $\frac{\partial g}{\partial x^{\top}}^{\!\top}\nabla f(g(x))$ |

### The least-squares objective

$$
S(\beta) = (y - X\beta)^{\top}(y - X\beta) = y^{\top}y - 2\beta^{\top}X^{\top}y + \beta^{\top}X^{\top}X\beta,
$$

$$
\nabla S(\beta) = -2X^{\top}y + 2X^{\top}X\beta = -2X^{\top}(y - X\beta), \qquad \nabla^2 S = 2X^{\top}X.
$$

Setting the gradient to zero gives the **normal equations** $X^{\top}X\hat\beta = X^{\top}y$. The Hessian is PSD (Day 17), so $S$ is convex and the solution is a minimum (Day 25).

### The same template, other estimators

- **Ridge:** $S(\beta) + \lambda\beta^{\top}\beta \Rightarrow \hat\beta = (X^{\top}X + \lambda I)^{-1}X^{\top}y$.
- **Weighted least squares:** $(y - X\beta)^{\top}W(y - X\beta) \Rightarrow \hat\beta = (X^{\top}WX)^{-1}X^{\top}Wy$.
- **Logit:** $\nabla\ell = X^{\top}(y - p)$ and $\nabla^2\ell = -X^{\top}VX$ with $V = \operatorname{diag}(p_i(1 - p_i))$.

> **Causal link.** Weighted least squares with weights $1/\hat e(x)$ and $1/(1 - \hat e(x))$ is how IPW regressions are run; ridge and lasso are the nuisance learners in double machine learning.

### Exercise 9 · A unique minimum

Show that the Hessian of $S$ is PD when $X$ has full column rank, so the solution is the unique minimiser.

<details><summary>Show solution</summary>

**Step 1.** $\nabla^2S = 2X^{\top}X$ and $v^{\top}X^{\top}Xv = \|Xv\|^2 > 0$ for $v \ne 0$ under full rank (Day 18).

**Step 2.** A function with a PD Hessian everywhere is strictly convex, so its stationary point is the unique global minimum (Day 25).

### Exercise 10 · Ridge regression

Derive the minimiser of $S(\beta) + \lambda\|\beta\|^2$ and show it exists even when $X$ is rank-deficient (for $\lambda > 0$).

<details><summary>Show solution</summary>

**Step 2.** For $v \ne 0$, $v^{\top}(X^{\top}X + \lambda I)v = \|Xv\|^2 + \lambda\|v\|^2 > 0$, so the matrix is PD and invertible whatever the rank of $X$.

</details>

### Exercise 12 · Weighted least squares

Derive the WLS estimator, and explain how weights $w_i = 1/\hat e(x_i)$ for treated and $1/(1 - \hat e(x_i))$ for control units turn a regression of $y$ on $(1, T)$ into an IPW estimator.

<details><summary>Show solution</summary>

**Step 1.** Objective: $(y - X\beta)^{\top}W(y - X\beta)$ with $W$ diagonal (one weight per unit).

**Step 2.** Gradient: $-2X^{\top}W(y - X\beta)$. Setting it to zero gives $X^{\top}WX\beta = X^{\top}Wy$, so $\hat\beta = (X^{\top}WX)^{-1}X^{\top}Wy$.

**Step 3.** With $X = (\mathbf 1, T)$, the fitted values are the weighted group means, so the slope is (weighted mean of treated outcomes) − (weighted mean of control outcomes).

**Step 4.** With weights $1/\hat e(x_i)$ for treated and $1/(1 - \hat e(x_i))$ for controls, that is the Hájek IPW estimator.

**Answer:** an IPW estimate of the ATE is a weighted regression of $y$ on a treatment dummy (Days 14 and 44).

### Exercise 13 · The Hessian of a propensity-score logit

For a logit model of programme participation, $\ell(\beta) = \sum_i[y_ix_i^{\top}\beta - \ln(1 + e^{x_i^{\top}\beta})]$ ($y_i = 1$ if household $i$ participates). Derive the gradient and Hessian and show that $\ell$ is concave.

<details><summary>Show solution</summary>

**Step 1.** Derivative of the second term: $\dfrac{\partial}{\partial\beta}\ln(1 + e^{x^{\top}\beta}) = \dfrac{e^{x^{\top}\beta}}{1 + e^{x^{\top}\beta}}x = p\,x$.

**Step 2.** Gradient: $\nabla\ell = \sum_i(y_i - p_i)x_i = X^{\top}(y - p)$.

**Step 3.** Differentiate $-p_ix_i$ again using $\partial p_i/\partial\beta^{\top} = p_i(1 - p_i)x_i^{\top}$: $\nabla^2\ell = -\sum_ip_i(1 - p_i)x_ix_i^{\top} = -X^{\top}VX$ with $V = \operatorname{diag}(p_i(1 - p_i))$.

**Step 4.** For any $v$, $v^{\top}X^{\top}VXv = \sum_ip_i(1 - p_i)(x_i^{\top}v)^2 \ge 0$, so the Hessian is negative semi-definite.

**Answer:** $\ell$ is concave, so any stationary point is a global maximum and Newton's method is reliable (Day 25).

</details>

### Exercise 14 · The MLE of the error variance in a wage regression

For a log-wage regression with normal errors, $\ell = -\frac n2\ln(2\pi\sigma^2) - \frac{S(\beta)}{2\sigma^2}$, where $S$ is the sum of squared residuals. Find the MLE of $\sigma^2$ and compare it with $s^2$.

<details><summary>Show solution</summary>

**Step 1.** Differentiate with respect to $\sigma^2$ (treat it as one variable): $\dfrac{\partial\ell}{\partial\sigma^2} = -\dfrac{n}{2\sigma^2} + \dfrac{S}{2\sigma^4}$.

**Step 2.** Set to zero and multiply by $2\sigma^4$: $-n\sigma^2 + S = 0$.

**Step 3.** At $\hat\beta$: $\hat\sigma^2 = S(\hat\beta)/n = e^{\top}e/n$.

**Answer:** $e^{\top}e/n$, which is biased downward; $s^2 = e^{\top}e/(n - k)$ is unbiased (Day 27).

</details>

## Verify in Python

*Your work:*

```python
import numpy as np

def num_grad(f, x, h=1e-6):
    return np.array([(f(x + h * e) - f(x - h * e)) / (2 * h) for e in np.eye(len(x))])

A = np.array([[2, 1], [1, 3.]])
print("grad x'Ax at (1,1):", num_grad(lambda x: x @ A @ x, np.array([1, 1.])).round(4))

x = np.array([1, 2, 3, 4.]); y = np.array([2, 3, 5, 6.]); X = np.column_stack([np.ones(4), x])
S = lambda b: (y - X @ b) @ (y - X @ b)
for b in (np.array([0, 1.]), np.array([0.5, 1.4])):
    print("grad S at", b, ":", num_grad(S, b).round(4), " formula:", (-2 * X.T @ (y - X @ b)).round(4))

print("dbeta/dy4:", (np.linalg.inv(X.T @ X) @ X.T)[:, 3])
lam = 0.5
print("ridge:", np.linalg.solve(X.T @ X + lam * np.eye(2), X.T @ y).round(4))
```

## Key takeaways

- $\partial(a^{\top}x)/\partial x = a$ and $\partial(x^{\top}Ax)/\partial x = 2Ax$ (symmetric $A$) do most of the work.
- Setting the gradient of the SSR to zero gives the normal equations; its Hessian $2X^{\top}X$ guarantees a minimum.
- Ridge, WLS and logit follow the same recipe: write the objective, differentiate, solve.
