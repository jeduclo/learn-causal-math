The bandwidth trades bias against variance. The MSE-optimal bandwidth balances them so well that the bias is no longer negligible relative to the standard error, and conventional confidence intervals under-cover. Robust bias-corrected inference fixes this by subtracting an estimate of the bias and accounting for its uncertainty.

**Learning outcomes**

- Derive the MSE-optimal bandwidth and its n^(−1/5) rate.
- Show that at the MSE-optimal bandwidth the bias is half the standard error, and compute the resulting coverage loss.
- Explain robust bias-corrected (RBC) confidence intervals and why conventional standard errors are too small after bias correction.
- Design a bandwidth robustness analysis for an RD study.

## Theory

### The trade-off

For a local linear RD estimator with bandwidth $h$ and sample size $n$,

$$
\text{Bias} \approx Bh^2, \qquad \operatorname{Var} \approx \frac{V}{nh},
$$

where $B$ depends on the curvature of $E[Y \mid X]$ near the cutoff and $V$ on the outcome variance and the density of $X$ at $c$.

### MSE-optimal bandwidth

Minimising $\text{MSE}(h) = B^2h^4 + V/(nh)$ gives

$$
h_{MSE} = \Big(\frac{V}{4B^2n}\Big)^{1/5} \propto n^{-1/5}.
$$

Data-driven selectors (Imbens–Kalyanaraman; Calonico–Cattaneo–Titiunik, CCT) estimate $B$ (using a pilot local quadratic fit) and $V$, then plug in. Different bandwidths can be used on each side.

### The inference problem

At $h_{MSE}$, squared bias equals one quarter of the variance, so

$$
\frac{\text{Bias}}{\operatorname{SE}} = \frac12.
$$

A conventional 95% interval $\hat\tau \pm 1.96\,\operatorname{SE}$ then covers only about 92% of the time, and less when bias is underestimated.

### Fixes

- **Undersmoothing:** use a smaller bandwidth than $h_{MSE}$ so bias is negligible; the point estimate becomes noisier and the right amount of undersmoothing is unclear.
- **Robust bias correction (Calonico, Cattaneo and Titiunik, 2014):** estimate the bias with a local quadratic fit (pilot bandwidth $b$), subtract it, and compute a standard error that includes the variability of the bias estimate. With $b = h$, the bias-corrected estimator coincides with the local quadratic estimator, and RBC inference is local quadratic inference at bandwidth $h$.

**Current practice:** report the local linear estimate at $h_{MSE}$ with the RBC confidence interval; show sensitivity to halving and doubling the bandwidth. Coverage-error-optimal bandwidths (smaller than $h_{MSE}$) are an alternative for inference.

### Exercise 8 · The convergence rate

Show that at $h_{MSE}$ the root mean squared error is of order $n^{-2/5}$, slower than the parametric $n^{-1/2}$.

<details><summary>Show solution</summary>

**Step 1.** $\text{MSE} = B^2h^4 + V/(nh)$ with $h \propto n^{-1/5}$.

**Step 2.** $B^2h^4 \propto n^{-4/5}$ and $V/(nh) \propto n^{-1 + 1/5} = n^{-4/5}$.

**Step 3.** $\text{RMSE} \propto n^{-2/5}$. $\blacksquare$ RD uses only data near the cutoff, so it learns more slowly than an experiment.

Let $\hat\tau_{bc} = \hat\tau - \hat B$. Show $\operatorname{Var}(\hat\tau_{bc}) = \operatorname{Var}(\hat\tau) + \operatorname{Var}(\hat B) - 2\operatorname{Cov}(\hat\tau, \hat B)$ and explain why using $\operatorname{SE}(\hat\tau)$ alone is wrong.

## Verify in Python

*Your work:*

```python
import numpy as np
from scipy.stats import norm

print("Ex 1 h_MSE:", round((4 / (4 * 4 * 1000)) ** .2, 3), " Ex 11:", round((2 / (4 * .25 * 5000)) ** .2, 3),
      " Ex 3 coverage:", round(norm.cdf(1.96 - .5) - norm.cdf(-1.96 - .5), 3), " Ex 4 factor:", round(5 ** .4, 2))

rng = np.random.default_rng(75)
mu0 = lambda x: 50 + 8 * x + 30 * x ** 2 - 20 * x ** 3
def side(x, y, h, p):
    w = np.clip(1 - np.abs(x) / h, 0, None); m = w > 0
    X = np.column_stack([x[m] ** k for k in range(p + 1)]); W = w[m]
    A = np.linalg.inv(X.T @ (W[:, None] * X)); b = A @ (X.T @ (W * y[m])); r = y[m] - X @ b
    V = A @ (X.T @ ((W ** 2 * r ** 2)[:, None] * X)) @ A                    # heteroskedasticity-robust
    return b[0], V[0, 0]
def rd(x, y, h, p):
    r = x >= 0
    a, va = side(x[r], y[r], h, p); b, vb = side(-x[~r], y[~r], h, p)
    return a - b, np.sqrt(va + vb)
print("  h   LL bias  LL sd  LL cover | LQ bias  LQ sd  LQ cover")
for h in (.15, .3, .5):
    out = {1: [], 2: []}; cov = {1: 0, 2: 0}
    for _ in range(1000):
        x = rng.uniform(-1, 1, 1000); y = mu0(x) + 5 * (x >= 0) + rng.normal(0, 3, 1000)
        for p in (1, 2):
            t, s = rd(x, y, h, p); out[p].append(t); cov[p] += abs(t - 5) < 1.96 * s
    print(f"{h:4.2f}  {np.mean(out[1]) - 5:7.2f}  {np.std(out[1]):5.2f}  {cov[1] / 1000:7.3f} | "
          f"{np.mean(out[2]) - 5:7.2f}  {np.std(out[2]):5.2f}  {cov[2] / 1000:7.3f}")
```

## Key takeaways

- The MSE-optimal bandwidth shrinks like n^(−1/5) and leaves a bias equal to half the standard error.
- Conventional intervals at that bandwidth under-cover; robust bias-corrected intervals subtract the estimated bias and account for its variance.
- Pre-specify a data-driven bandwidth, report RBC intervals, and show sensitivity to the bandwidth.
