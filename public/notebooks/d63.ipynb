When treatment effects differ across people, IV does not estimate the average effect. Under independence, exclusion and monotonicity it estimates the average effect for compliers, the people whose treatment the instrument changes. Their share, outcomes and characteristics are all identified.

**Learning outcomes**

- Define compliers, always-takers, never-takers and defiers with potential treatments.
- State and prove the LATE theorem.
- Identify the shares of each type, complier mean potential outcomes and complier characteristics.
- Explain why different instruments give different LATEs and what monotonicity failures do.

## Theory

### Potential treatments

Let $Z \in \{0, 1\}$ be the instrument and $D(z)$ the treatment a unit would take if $Z = z$. Four types:

| Type | $D(1)$ | $D(0)$ |
|---|---|---|
| Complier | 1 | 0 |
| Always-taker | 1 | 1 |
| Never-taker | 0 | 0 |
| Defier | 0 | 1 |

We observe only $D = D(Z)$, so types are not observed individually.

### Assumptions

1. **Independence:** $Z$ is independent of all potential outcomes $Y(d, z)$ and potential treatments $D(z)$.
2. **Exclusion:** $Y(d, z) = Y(d)$: the instrument matters only through treatment.
3. **Relevance:** $P(D = 1 \mid Z = 1) \ne P(D = 1 \mid Z = 0)$.
4. **Monotonicity:** $D(1) \ge D(0)$ for everyone: no defiers.

### The LATE theorem (Imbens and Angrist, 1994)

Under 1–4,

$$
\frac{E[Y \mid Z = 1] - E[Y \mid Z = 0]}{E[D \mid Z = 1] - E[D \mid Z = 0]} = E[Y(1) - Y(0) \mid \text{complier}] \equiv LATE.
$$

Always-takers and never-takers have the same treatment whatever $Z$, so they contribute nothing to the reduced form; the ratio rescales the compliers' effect by their share.

### What is identified

- **Shares:** $\pi_a = P(D = 1 \mid Z = 0)$, $\pi_n = P(D = 0 \mid Z = 1)$, $\pi_c = 1 - \pi_a - \pi_n$ (the first stage).
- **Complier outcome means:**

$$
E[Y(1) \mid c] = \frac{E[YD \mid Z = 1] - E[YD \mid Z = 0]}{\pi_c}, \qquad E[Y(0) \mid c] = \frac{E[Y(1 - D) \mid Z = 0] - E[Y(1 - D) \mid Z = 1]}{\pi_c}.
$$

- **Complier characteristics:** for any pre-treatment $X$, $E[X \mid c] = \big(E[XD \mid Z = 1] - E[XD \mid Z = 0]\big)/\pi_c$.

### Interpretation

LATE is the effect for a subpopulation defined by the instrument. Another instrument has other compliers and, with heterogeneous effects, another LATE. Whether a LATE is policy-relevant depends on whether the policy would change treatment for similar people. Profiling compliers (Exercise 12) helps judge this.

## Part 2 · Proofs

### Exercise 6 · The LATE theorem

Prove the LATE theorem.

<details><summary>Show solution</summary>

**Step 1.** By exclusion, $Y = Y(0) + D(Z)\,(Y(1) - Y(0))$.

**Step 2.** By independence, $E[Y \mid Z = 1] - E[Y \mid Z = 0] = E\big[(D(1) - D(0))(Y(1) - Y(0))\big]$.

**Step 3.** By monotonicity, $D(1) - D(0) \in \{0, 1\}$, equal to 1 exactly for compliers. So the reduced form is $\pi_c\,E[Y(1) - Y(0) \mid c]$.

**Step 4.** First stage: $E[D(1) - D(0)] = \pi_c$.

**Step 5.** Divide. $\blacksquare$

</details>

*Your work:*

### Exercise 7 · Identifying the shares

Prove $\pi_a = P(D = 1 \mid Z = 0)$ and $\pi_n = P(D = 0 \mid Z = 1)$ under independence and monotonicity.

<details><summary>Show solution</summary>

**Step 1.** With $Z = 0$, the treated are those with $D(0) = 1$: always-takers and defiers. With no defiers, only always-takers.

**Step 2.** Independence: the type distribution is the same for $Z = 0$ and $Z = 1$, so $P(D = 1 \mid Z = 0) = \pi_a$.

**Step 3.** With $Z = 1$, the untreated are never-takers (and defiers, excluded): $P(D = 0 \mid Z = 1) = \pi_n$. $\blacksquare$

</details>

*Your work:*

### Exercise 8 · Complier mean of Y(1)

Prove $E[Y(1) \mid c] = (E[YD \mid Z = 1] - E[YD \mid Z = 0])/\pi_c$.

<details><summary>Show solution</summary>

**Step 1.** With $Z = 1$, treated units are compliers and always-takers: $E[YD \mid Z = 1] = \pi_c E[Y(1) \mid c] + \pi_a E[Y(1) \mid a]$.

**Step 2.** With $Z = 0$, treated units are always-takers only: $E[YD \mid Z = 0] = \pi_a E[Y(1) \mid a]$.

**Step 3.** Subtract and divide by $\pi_c$. $\blacksquare$

</details>

*Your work:*

### Exercise 9 · Wald with defiers

Without monotonicity, show the Wald estimand is $\dfrac{\pi_c\,LATE_c - \pi_d\,LATE_d}{\pi_c - \pi_d}$.

<details><summary>Show solution</summary>

**Step 1.** From Exercise 6, Step 2: the reduced form is $E[(D(1) - D(0))(Y(1) - Y(0))]$.

**Step 2.** $D(1) - D(0) = 1$ for compliers, $-1$ for defiers, 0 otherwise: reduced form $= \pi_c LATE_c - \pi_d LATE_d$.

**Step 3.** First stage $= \pi_c - \pi_d$. $\blacksquare$

</details>

### Exercise 10 · Complier characteristics

For a pre-treatment covariate $X$, prove $E[X \mid c] = (E[XD \mid Z = 1] - E[XD \mid Z = 0])/\pi_c$.

<details><summary>Show solution</summary>

**Step 1.** $X$ is pre-treatment, so it is not affected by $Z$ and, by independence, its distribution within each type does not depend on $Z$.

**Step 2.** $E[XD \mid Z = 1] = \pi_c E[X \mid c] + \pi_a E[X \mid a]$ and $E[XD \mid Z = 0] = \pi_a E[X \mid a]$.

**Step 3.** Subtract and divide by $\pi_c$. $\blacksquare$

</details>

### Exercise 11 · A school-voucher lottery

Among lottery winners ($Z = 1$), 70% use a voucher; their mean test score is 60, and non-users score 50. Among losers, 20% attend a private school anyway, scoring 62; the rest score 48. Compute the type shares, the LATE, the compliers' mean scores with and without the voucher, and compare with always-takers and never-takers.

<details><summary>Show solution</summary>

**Step 1.** Shares: $\pi_a = 0.2$, $\pi_n = 0.3$, $\pi_c = 0.5$.

**Step 4.** LATE $= 59.2 - 46.8 = 12.4$ (matches Exercise 3).

**Step 5.** Always-takers' treated mean: 62; never-takers' untreated mean: 50.

**Answer:** vouchers raise compliers' scores by 12.4 points. Compliers would score less than never-takers without the voucher (46.8 versus 50): they are relatively disadvantaged families for whom the voucher made private school affordable.

### Exercise 14 · Defiers in a fine policy

A city sends warning letters ($Z$) about a recycling fine. The letter makes 40% of households start recycling (compliers, effect on waste of 10 kg) but makes 10% stop in protest (defiers, effect 4 kg). Compute the Wald estimate.

<details><summary>Show solution</summary>

**Step 1.** Reduced form: $0.4 \times 10 - 0.1 \times 4 = 3.6$.

**Step 2.** First stage: $0.4 - 0.1 = 0.3$.

**Step 3.** Wald: $3.6/0.3 = 12$.

**Answer:** 12, larger than every individual effect in the population (10 and 4): without monotonicity the ratio is not an average treatment effect for anyone.

### Exercise 15 · A mandate versus a lottery

A health insurance lottery gives a LATE of 0.4 extra doctor visits per year. A ministry asks what a universal mandate would do. Explain what extra information is needed.

<details><summary>Show solution</summary>

**Step 1.** A mandate changes coverage for never-takers and (if it forces nothing on them) leaves always-takers unchanged.

**Step 2.** The LATE describes compliers only; never-takers may be healthier or less inclined to use care, so their effect may be smaller.

**Step 3.** Evidence needed: complier versus never-taker characteristics (Exercise 10), effect heterogeneity by observables among compliers, or a second instrument that moves never-takers.

**Answer:** the LATE is informative but not the answer; extrapolation needs assumptions about how effects vary.

</details>

## Verify in Python

*Your work:*

```python
import numpy as np

rng = np.random.default_rng(63); n = 2_000_000
Z = rng.random(n) < .5
typ = rng.choice(["c", "a", "n"], size=n, p=[.5, .2, .3])
girl = rng.random(n) < np.select([typ == "c", typ == "a", typ == "n"], [.7, .3, .33])
Y0 = np.select([typ == "c", typ == "a", typ == "n"], [46.8, 50, 50]) + rng.normal(0, 8, n)
gain = np.select([typ == "c", typ == "a", typ == "n"], [12.4, 12.0, 5.0]) + rng.normal(0, 3, n)
D = (typ == "a") | ((typ == "c") & Z)
Y = Y0 + D * gain

fs = D[Z].mean() - D[~Z].mean(); rf = Y[Z].mean() - Y[~Z].mean()
print(f"shares: always {D[~Z].mean():.3f}, never {1 - D[Z].mean():.3f}, compliers {fs:.3f}")
print(f"Wald {rf / fs:.2f};  true LATE {gain[typ == 'c'].mean():.2f};  ATE {gain.mean():.2f}")
EY1c = ((Y * D)[Z].mean() - (Y * D)[~Z].mean()) / fs
EY0c = ((Y * ~D)[~Z].mean() - (Y * ~D)[Z].mean()) / fs
print(f"complier means: Y(1) {EY1c:.2f}, Y(0) {EY0c:.2f};  share girls among compliers {((girl * D)[Z].mean() - (girl * D)[~Z].mean()) / fs:.3f}")

# Exercise 11 from the summary statistics, and Exercise 14
print("voucher:", round((.7 * 60 - .2 * 62) / .5, 2), round((.8 * 48 - .3 * 50) / .5, 2), " defier Wald:", round((.4 * 10 - .1 * 4) / .3, 3))
```

## Key takeaways

- Under independence, exclusion and monotonicity, IV estimates the average effect for compliers (LATE), not the ATE.
- Type shares, complier mean potential outcomes and complier characteristics are identified from simple moments.
- Different instruments define different compliers; defiers break the interpretation entirely.
