# Day 70: Staggered Adoption: the Goodman-Bacon Decomposition

*Beginner to Ninja: Math for Causal Inference*

When units adopt a policy at different times, the TWFE coefficient is a weighted average of every possible 2×2 DiD, including comparisons that use already-treated units as controls. With effects that change over time those comparisons are contaminated, and TWFE can be far from, or even opposite in sign to, every true effect.

**Learning outcomes**

- List the 2×2 comparisons hidden inside a staggered TWFE regression.
- Explain the decomposition weights and why units treated in the middle of the panel get the most weight.
- Prove that TWFE is unbiased with constant effects and biased with dynamic effects through "forbidden" comparisons.
- Compute a decomposition and compare TWFE with the true average effect.

## Theory

### Staggered adoption

Units adopt treatment at different dates and stay treated. Write $k$ for an **early** group (adopts at $t_k$), $l$ for a **late** group ($t_l > t_k$), and $U$ for **never-treated** units. The common practice was to run the TWFE regression $Y_{it} = \alpha_i + \lambda_t + \beta D_{it} + \varepsilon_{it}$ and call $\hat\beta$ "the effect".

### The decomposition

Goodman-Bacon (2021) showed that, in a balanced panel,

$$
\hat\beta_{TWFE} = \sum_{\text{2×2 comparisons } j} s_j\,\hat\beta^{2\times2}_j, \qquad \sum_j s_j = 1,\; s_j \ge 0.
$$

With one early group, one late group and a never-treated group there are four comparisons:

| Comparison | Treated | Control | Window |
|---|---|---|---|
| $\hat\beta_{kU}$ | early | never | all periods |
| $\hat\beta_{lU}$ | late | never | all periods |
| $\hat\beta^{k}_{kl}$ | early | late (not yet treated) | before $t_l$ |
| $\hat\beta^{l}_{kl}$ | late | early (**already treated**) | from $t_k$ on |

### The weights

Each weight grows with the size of the two groups and with the **variance of treatment** in the comparison, $\bar D(1 - \bar D)$, where $\bar D$ is the share of the window in which the treated group is treated. Groups treated in the middle of the panel ($\bar D$ near 0.5) get the most weight; this weighting is a property of OLS, not of the policy question.

### When it goes wrong

- **Constant effects:** every 2×2 equals the effect; TWFE is fine.
- **Effects differing across groups but constant over time:** TWFE is a convex, variance-weighted average of group effects; not the ATT, but between them.
- **Effects that change over time:** the comparison $\hat\beta^{l}_{kl}$ uses the early group as a control after it was treated, so its still-evolving effect is subtracted. TWFE can be biased and even have the wrong sign. In the equivalent ATT language (de Chaisemartin and D'Haultfœuille), some group-time effects receive **negative weights**.

### Diagnostics and fixes

Plot the 2×2 estimates against their weights (a "Bacon plot"), check how much weight falls on already-treated controls, and use heterogeneity-robust estimators that never use treated units as controls (Day 71).

## Part 2 · Proofs

*Your work:*

### Exercise 6 · Constant effects mean no bias

Assume parallel trends and a constant effect $\tau$ for all units and periods. Prove each 2×2 estimand equals $\tau$, and hence $\beta_{TWFE} = \tau$.

<details><summary>Show solution</summary>

**Step 1.** In any 2×2, the treated group's change is (common trend) + $\tau$ × (change in its treatment status between pre and post windows) = trend + $\tau$.

**Step 2.** The control group's change is the common trend: for not-yet-treated controls nothing changes; for already-treated controls the effect $\tau$ is present in both windows and cancels.

**Step 3.** Each 2×2 equals $\tau$; the weights sum to 1. $\blacksquare$

</details>

*Your work:*

### Exercise 7 · The contaminated comparison

Let the early group's effect be $\tau_k(e)$ at event time $e$ and the late group's $\tau_l(e)$. Show that, under parallel trends, $\beta^{l}_{kl} = \overline{\tau_l}^{\,post} - \big(\overline{\tau_k}^{\,post} - \overline{\tau_k}^{\,pre}\big)$, where the bars average over the windows' periods.

<details><summary>Show solution</summary>

**Step 1.** Late group: its change between the window's pre part ($t_k \le t < t_l$, untreated) and post part ($t \ge t_l$) is trend + average $\tau_l$ over the post part.

**Step 2.** Early group: treated throughout; its change is trend + (average $\tau_k$ in the post part − average $\tau_k$ in the pre part).

**Step 3.** Subtract. $\blacksquare$ If $\tau_k$ grows over time, the bracket is positive and the comparison is biased downwards.

</details>

### Exercise 8 · Group-specific constant effects

Suppose effects differ across groups but not over time ($\tau_k$, $\tau_l$). Show $\beta_{TWFE} = (s_{kU} + s^{k}_{kl})\tau_k + (s_{lU} + s^{l}_{kl})\tau_l$, a convex combination that generally differs from the ATT.

<details><summary>Show solution</summary>

**Step 1.** $\beta_{kU} = \tau_k$, $\beta_{lU} = \tau_l$, $\beta^{k}_{kl} = \tau_k$ (late group untreated in that window).

**Step 2.** By Exercise 7 with constant $\tau_k$, the bracket is 0, so $\beta^{l}_{kl} = \tau_l$.

**Step 3.** Collect terms; weights are non-negative and sum to 1. $\blacksquare$ The ATT weights groups by their number of treated unit-periods instead.

</details>

Construct an example (early effect growing, late effect small) in which $\beta^{l}_{kl} < 0$ although every unit's effect is non-negative.

### Exercise 10 · Weights favour the middle

For a timing group compared with never-treated units of equal size, the weight is proportional to $\bar D(1 - \bar D)$. Show this is maximised when treatment covers half the panel.

## Verify in Python

*Your work:*

```python
import numpy as np

def panel(T, tk, tl, eff, eff_late=None):
    eff_late = eff_late or eff
    t = np.arange(1, T + 1)
    D = np.array([(t >= tk), (t >= tl), np.zeros(T, bool)]).astype(float)
    base = np.array([1.0, 2.0, 0.5])[:, None] + 0.3 * t[None, :]
    E = np.array([eff(t - tk), eff_late(t - tl), np.zeros(T)])
    return base + D * E, D, t

def twfe_weighted(Y, D, w):
    w = w / w.sum()
    dm = lambda M: M - M.mean(1, keepdims=True) - (w[:, None] * M).sum(0) + (w[:, None] * M.mean(1, keepdims=True)).sum()
    Dd = dm(D)
    return (w[:, None] * Dd * Y).sum() / (w[:, None] * Dd ** 2).sum()

def bacon(Y, t, tk, tl, w):
    did = lambda a, b, pre, post: (Y[a, post].mean() - Y[a, pre].mean()) - (Y[b, post].mean() - Y[b, pre].mean())
    pk, pl = t >= tk, t >= tl
    b = np.array([did(0, 2, ~pk, pk), did(1, 2, ~pl, pl), did(0, 1, ~pk & ~pl, pk & ~pl), did(1, 0, pk & ~pl, pl)])
    nk, nl, nu = w / w.sum(); Dk, Dl = pk.mean(), pl.mean()
    sh = lambda a, c: a / (a + c)
    s = np.array([(nk + nu) ** 2 * sh(nk, nu) * (1 - sh(nk, nu)) * Dk * (1 - Dk),
                  (nl + nu) ** 2 * sh(nl, nu) * (1 - sh(nl, nu)) * Dl * (1 - Dl),
                  ((nk + nl) * (1 - Dl)) ** 2 * sh(nk, nl) * (1 - sh(nk, nl)) * (Dk - Dl) / (1 - Dl) * (1 - Dk) / (1 - Dl),
                  ((nk + nl) * Dk) ** 2 * sh(nk, nl) * (1 - sh(nk, nl)) * Dl / Dk * (Dk - Dl) / Dk])
    return b, s / s.sum()

for label, eff in [("constant effect 2", lambda e: 2 + 0 * e), ("dynamic 1 + 0.5e", lambda e: 1 + .5 * e)]:
    for w in (np.array([1., 1, 1]), np.array([2., 1, 3])):
        Y, D, t = panel(10, 4, 7, eff)
        b, s = bacon(Y, t, 4, 7, w)
        att = (w[0] * eff(np.arange(7)).sum() + w[1] * eff(np.arange(4)).sum()) / (7 * w[0] + 4 * w[1])
        print(f"{label:18s} sizes {w.astype(int)}: 2x2s {b.round(2)} weights {s.round(3)} "
              f"sum {(s * b).sum():.3f} TWFE {twfe_weighted(Y, D, w):.3f} ATT {att:.3f}")

# Exercise 9: early effect = e (growing), late effect = 0.5
Y, D, t = panel(10, 4, 7, lambda e: e * 1.0, lambda e: 0.5 + 0 * e)
b, s = bacon(Y, t, 4, 7, np.ones(3))
print("late vs already-treated early:", round(b[3], 2))
```

## Key takeaways

- Staggered TWFE averages all 2×2 DiDs, with OLS weights that favour groups treated mid-panel.
- Comparisons that use already-treated units as controls subtract their evolving effects; with dynamic effects TWFE is biased and can flip sign.
- Report the adoption pattern and a decomposition, and use estimators that only compare with not-yet- or never-treated units.
