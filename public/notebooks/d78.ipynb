Some policies change the slope of a schedule rather than its level, such as a benefit capped at a maximum. The regression kink design divides the change in the outcome's slope by the change in the policy's slope. Also: geographic boundaries, multiple cutoffs and RD in time.

**Learning outcomes**

- Define the regression kink estimand and the smoothness assumptions it requires.
- Estimate slope changes with local polynomials and explain why they are noisier than jumps.
- Extend RD to multiple cutoffs, geographic boundaries and time, and list each extension's threats.
- Apply RKD to unemployment benefits and other capped or tiered schedules.

## Theory

### Kinked policies

A policy variable $B = b(X)$ is a **continuous** function of the running variable with a **kink** at $c$: its slope changes from $b'(c^-)$ to $b'(c^+)$. Examples: unemployment benefits equal to 50% of past earnings up to a maximum; tax credits that phase in and out; co-payment schedules.

### The RKD estimand

If $Y = g(B, X, \varepsilon)$, then

$$
\tau_{RKD} = \frac{\lim_{x \downarrow c}\frac{d}{dx}E[Y \mid X = x] - \lim_{x \uparrow c}\frac{d}{dx}E[Y \mid X = x]}{b'(c^+) - b'(c^-)}
$$

identifies the (average) marginal effect of $B$ on $Y$ at $c$, under **smoothness**: the density of $X$ and the direct dependence of $Y$ on $X$ have continuous derivatives at $c$ (Card, Lee, Pei and Weber, 2015). With imperfect implementation (actual benefits deviate from the formula), a **fuzzy RKD** divides by the estimated kink in actual $B$.

### Estimation

Fit local polynomials on each side and compare the **slopes** at $c$ (local linear or local quadratic, with a kink term $(X - c)\mathbf 1[X \ge c]$). Derivatives are harder to estimate than levels: the variance of a slope estimate scales like $1/(nh^3)$ instead of $1/(nh)$, so RKD needs large samples (administrative data).

### Extensions of RD

- **Multiple cutoffs:** different cohorts or regions have different thresholds; normalise $X - c_j$ and pool (a weighted average of cutoff-specific effects), or exploit the variation to learn how effects change with the cutoff.
- **Geographic RD:** a boundary separates treated from untreated areas (school districts, states with different laws). The running variable is distance to the boundary (two-dimensional). Threats: other policies change at the same border; people sort across borders.
- **RD in time:** treatment starts on a date. Concerns: other things change at the same time, anticipation and adjustment, serial correlation; often closer to an interrupted time series than an RD.
- **Extrapolation:** with assumptions (for example, conditional independence of $X$ and outcomes given covariates), effects away from the cutoff can be estimated.

### Exercise 6 · RKD identification

Let $Y = \beta B + m(X) + \varepsilon$ with $E[\varepsilon \mid X]$ having a continuous derivative, $m$ differentiable, and $B = b(X)$ kinked at $c$. Prove $\tau_{RKD} = \beta$.

<details><summary>Show solution</summary>

**Step 1.** $E[Y \mid X = x] = \beta b(x) + m(x) + E[\varepsilon \mid X = x]$.

**Step 2.** Derivative: $\beta b'(x) + m'(x) + \frac{d}{dx}E[\varepsilon \mid X = x]$.

**Step 3.** The last two terms are continuous at $c$; the slope change is $\beta(b'(c^+) - b'(c^-))$.

**Step 4.** Divide by the policy's slope change. $\blacksquare$

</details>

## Verify in Python

```python
import numpy as np

print("Ex 2:", -.1 / -50, "weeks per $ ->", -.1 / -50 * 100, "weeks per $100", " Ex 11 elasticity:", round((.2 * 4.5 / 100 / 18) / .01, 3),
      " Ex 13:", .5 * .10 + .25 * .06 + .25 * .04)

rng = np.random.default_rng(78)
def kink_slope(x, y, h, p):
    out = []
    for side in (x >= 0, x < 0):
        xx, yy = x[side], y[side]; w = np.clip(1 - np.abs(xx) / h, 0, None); m = w > 0
        X = np.column_stack([xx[m] ** k for k in range(p + 1)]); W = w[m]
        out.append(np.linalg.solve(X.T @ (W[:, None] * X), X.T @ (W * yy[m]))[1])
    return out[0] - out[1]
res = {1: [], 2: []}
for _ in range(300):
    x = rng.uniform(-1, 1, 50_000)
    B = .5 * np.minimum(x, 0) + 2                     # benefit: slope 0.5 below the cap, 0 above
    y = .8 * B + np.sin(1.5 * x) + .3 * x ** 2 + rng.normal(0, 1, x.size)
    for p in (1, 2):
        res[p].append(kink_slope(x, y, .4, p) / (0 - .5))
for p in (1, 2):
    print(f"local {'linear' if p == 1 else 'quadratic'}: mean {np.mean(res[p]):.3f}  sd {np.std(res[p]):.3f}  (truth 0.8)")
```

## Key takeaways

- RKD uses a change in a policy's slope: effect = (change in outcome slope) / (change in policy slope), under smoothness of everything else.
- Slopes are harder to estimate than levels (variance ∝ 1/(nh³)); RKD needs large data sets.
- Multiple cutoffs, geographic boundaries and RD in time extend the design, each with its own threats: coinciding policies, sorting, adaptation and serial correlation.
