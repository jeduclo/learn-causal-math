# Day 88: Semiparametric Efficiency Bounds

How precise can any estimator of the ATE be, without assuming a parametric model? The semiparametric efficiency bound is the variance of the efficient influence function. For the ATE under unconfoundedness it has a closed form (Hahn, 1998) that explains why poor overlap is costly, why IPW with the true propensity is inefficient, and how to allocate treatment in experiments.

**Learning outcomes**

- State the Cramér–Rao bound and the idea of a semiparametric efficiency bound.
- Compute Hahn's bound for the ATE and decompose it into noise and heterogeneity terms.
- Prove that IPW with the known propensity score exceeds the bound, and derive Neyman allocation.
- Use the bound to plan sample sizes and diagnose overlap problems.

## Theory

### Parametric bounds

In a regular parametric model with Fisher information $I(\theta)$, any unbiased estimator satisfies $\operatorname{Var}(\hat\theta) \ge I(\theta)^{-1}/n$ (Cramér–Rao), and maximum likelihood attains it asymptotically (Day 11).

### Semiparametric bounds

In a semiparametric problem (the ATE, with the outcome regressions and propensity left unrestricted), consider every smooth parametric submodel passing through the truth. The **efficiency bound** is the largest of their Cramér–Rao bounds for the target, and it equals the variance of the **efficient influence function** $\varphi^*$:

$$
V^* = E[\varphi^*(W)^2].
$$

No regular estimator has smaller asymptotic variance; estimators whose influence function is $\varphi^*$ are **efficient**.

### Hahn's bound for the ATE

Under unconfoundedness and overlap, with $\sigma_d^2(x) = \operatorname{Var}(Y(d) \mid X = x)$ and $\tau(x) = \mu_1(x) - \mu_0(x)$,

$$
V^*_{ATE} = E\Big[\frac{\sigma_1^2(X)}{e(X)} + \frac{\sigma_0^2(X)}{1 - e(X)} + \big(\tau(X) - \tau\big)^2\Big].
$$

Three lessons:

- **Overlap:** small $e(X)$ or $1 - e(X)$ inflate the bound. Poor overlap is a fundamental information problem, not a defect of a particular estimator.
- **Heterogeneity** adds $\operatorname{Var}(\tau(X))$: averaging different effects carries sampling uncertainty about the mix of $X$.
- **Attained by** AIPW with consistent nuisance estimates, by regression imputation and by IPW with an *estimated* score (Hirano, Imbens and Ridder), but not by IPW with the *true* score.

### Design

In an experiment where you choose $e(x)$, minimising $\sigma_1^2/e + \sigma_0^2/(1 - e)$ gives **Neyman allocation** $e^* = \sigma_1/(\sigma_1 + \sigma_0)$: assign more units to the noisier arm.

### Exercise 7 · Hahn's bound from the influence function

Using the ATE influence function from Day 87, show $E[\varphi^2] = V^*_{ATE}$.

<details><summary>Show solution</summary>

**Step 1.** Write $\varphi = A + B - C$ with $A = \tau(X) - \tau$, $B = D(Y - \mu_1)/e$, $C = (1 - D)(Y - \mu_0)/(1 - e)$.

**Step 2.** Cross terms vanish: given $X$ (and $D$), $B$ and $C$ have mean zero, and $BC = 0$ since $D(1 - D) = 0$.

**Step 3.** $E[B^2] = E\big[E[D(Y - \mu_1)^2 \mid X]/e^2\big] = E[e\sigma_1^2/e^2] = E[\sigma_1^2/e]$; likewise $E[C^2] = E[\sigma_0^2/(1 - e)]$.

**Step 4.** Add $E[A^2] = E[(\tau(X) - \tau)^2]$. $\blacksquare$

</details>

### Exercise 8 · Neyman allocation

Minimise $f(e) = \sigma_1^2/e + \sigma_0^2/(1 - e)$ over $e \in (0, 1)$.

<details><summary>Show solution</summary>

**Step 1.** $f'(e) = -\sigma_1^2/e^2 + \sigma_0^2/(1 - e)^2 = 0$.

**Step 2.** $(1 - e)/e = \sigma_0/\sigma_1$, so $e = \sigma_1/(\sigma_1 + \sigma_0)$.

**Step 3.** $f$ is convex (positive second derivative): a minimum, with value $(\sigma_1 + \sigma_0)^2$. $\blacksquare$

</details>

### Exercise 9 · IPW with the true score is inefficient

Let $\psi = DY/e - (1 - D)Y/(1 - e) - \tau$ (IPW with known $e$). Show $\psi - \varphi = (D - e)\big(\mu_1/e + \mu_0/(1 - e)\big)$ and that $\operatorname{Var}(\psi) = V^* + E\big[e(1 - e)\big(\mu_1/e + \mu_0/(1 - e)\big)^2\big]$.

<details><summary>Show solution</summary>

**Step 1.** $\psi - \varphi = D\mu_1/e - (1 - D)\mu_0/(1 - e) - (\mu_1 - \mu_0) = \mu_1(D - e)/e + \mu_0(D - e)/(1 - e)$.

**Step 2.** Orthogonality: $E[(\psi - \varphi)\varphi]$: given $X$, $(D - e)$ times the $A$ term has mean 0; times $B$: $E[(D - e)D(Y - \mu_1) \mid X] = 0$ because $E[Y - \mu_1 \mid D = 1, X] = 0$; likewise for $C$.

**Step 3.** So $\operatorname{Var}(\psi) = \operatorname{Var}(\varphi) + E[(\psi - \varphi)^2]$, and $E[(D - e)^2 \mid X] = e(1 - e)$. $\blacksquare$ The extra term is large when outcome levels $\mu_d$ are large: IPW ignores what $X$ predicts about $Y$.

</details>

### Exercise 10 · Covariates help in experiments

In an experiment with constant $e$, show the difference-in-means variance $\operatorname{Var}(Y(1))/e + \operatorname{Var}(Y(0))/(1 - e)$ exceeds $V^*$ by $\operatorname{Var}\big(\sqrt{(1 - e)/e}\,\mu_1(X) + \sqrt{e/(1 - e)}\,\mu_0(X)\big)$.

<details><summary>Show solution</summary>

**Step 1.** $\operatorname{Var}(Y(d)) = E[\sigma_d^2(X)] + \operatorname{Var}(\mu_d(X))$.

**Step 2.** Difference: $\frac{\operatorname{Var}(\mu_1)}{e} + \frac{\operatorname{Var}(\mu_0)}{1 - e} - \operatorname{Var}(\mu_1 - \mu_0)$.

**Step 3.** Expand $\operatorname{Var}(\mu_1 - \mu_0) = \operatorname{Var}\mu_1 + \operatorname{Var}\mu_0 - 2\operatorname{Cov}$; the difference is $\frac{1 - e}{e}\operatorname{Var}\mu_1 + \frac{e}{1 - e}\operatorname{Var}\mu_0 + 2\operatorname{Cov}(\mu_1, \mu_0)$, which is the stated variance. $\blacksquare$ Regression adjustment (Lin, Day 39) recovers this gap.

</details>

## Verify in Python

*Your work:*

```python
import numpy as np

print("Ex 2:", 1 / .5 + 1 / .5, " Ex 3:", 1 / .2 + 1 / .8, " Ex 5:", 4 / (2 / 3) + 1 / (1 / 3), "vs", 4 / .5 + 1 / .5)
low = .1 * (1 / .05 + 1 / .95); print("Ex 12 share from low-e group:", round(low / (low + .9 * 4), 3), " Ex 11 n:", round(4 / .05 ** 2))

rng = np.random.default_rng(88)
def dgp(n):
    X = rng.normal(size=n); e = 1 / (1 + np.exp(-.8 * X)); D = (rng.random(n) < e).astype(float)
    mu0 = 1 + X; mu1 = mu0 + 1 + .5 * X
    Y = np.where(D == 1, mu1, mu0) + rng.normal(size=n) * np.where(D == 1, 1.5, 1.0)
    return X, e, D, Y
X, e, D, Y = dgp(2_000_000)
print("Hahn bound:", round(np.mean(1.5 ** 2 / e + 1 / (1 - e) + (.5 * X) ** 2), 3))
def logit(Z, d):
    b = np.zeros(Z.shape[1])
    for _ in range(20):
        p = 1 / (1 + np.exp(-Z @ b)); b += np.linalg.solve(Z.T @ ((p * (1 - p))[:, None] * Z), Z.T @ (d - p))
    return 1 / (1 + np.exp(-Z @ b))
res = {"IPW, true e": [], "IPW, estimated e": [], "AIPW": []}
n = 2000
for _ in range(1000):
    X, e, D, Y = dgp(n); Z = np.column_stack([np.ones(n), X]); eh = logit(Z, D)
    res["IPW, true e"].append(np.mean(D * Y / e - (1 - D) * Y / (1 - e)))
    res["IPW, estimated e"].append(np.mean(D * Y / eh - (1 - D) * Y / (1 - eh)))
    m1 = Z @ np.linalg.lstsq(Z[D == 1], Y[D == 1], rcond=None)[0]; m0 = Z @ np.linalg.lstsq(Z[D == 0], Y[D == 0], rcond=None)[0]
    res["AIPW"].append(np.mean(m1 - m0 + D * (Y - m1) / eh - (1 - D) * (Y - m0) / (1 - eh)))
for k, v in res.items():
    print(f"{k:17s} mean {np.mean(v):.3f}   n*Var {n * np.var(v):.2f}")
```

## Key takeaways

- The semiparametric efficiency bound is the variance of the efficient influence function: no regular estimator can do better.
- For the ATE it is $E[\sigma_1^2/e + \sigma_0^2/(1 - e) + (\tau(X) - \tau)^2]$: poor overlap and noisy outcomes cost precision.
- AIPW (and IPW with an estimated score) attain it; IPW with the true score does not. In experiments, allocate more units to the noisier arm.
