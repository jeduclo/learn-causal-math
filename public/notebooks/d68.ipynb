*Beginner to Ninja: Math for Causal Inference*

With several periods, an event-study regression traces effects period by period and puts the parallel-trends assumption on display through pre-treatment coefficients. Why pre-trend tests have low power, how anticipation shifts the reference period, and how to report results that are robust to bounded violations.

**Learning outcomes**

- Specify and interpret an event-study regression with unit and time fixed effects and a reference period.
- Compute event-study coefficients from group-by-period means.
- Calculate the power of a pre-trend test and explain why passing it is weak evidence.
- Construct sensitivity intervals that allow post-treatment violations bounded by pre-treatment ones.

## Theory

### The event-study regression

With one treated group adopting at date $E$ and many periods, define event time $k = t - E$. The event-study specification is

$$
Y_{it} = \alpha_i + \lambda_t + \sum_{k \ne -1}\beta_k\,\mathbf 1[t - E = k]\,G_i + \varepsilon_{it},
$$

with unit fixed effects $\alpha_i$, period fixed effects $\lambda_t$, and the period just before treatment ($k = -1$) omitted as the **reference**. Each $\beta_k$ is a DiD comparing period $k$ with period $-1$:

$$
\beta_k = (\bar Y_{G=1, k} - \bar Y_{G=1, -1}) - (\bar Y_{G=0, k} - \bar Y_{G=0, -1}).
$$

- $k \ge 0$: dynamic treatment effects (ATT $k$ periods after adoption), under parallel trends.
- $k < 0$: **placebo** or **pre-trend** coefficients, zero if trends were parallel before treatment.

### Reading pre-trends

Flat pre-period coefficients are reassuring, but:

1. **Low power.** Pre-trend coefficients are noisy; a violation large enough to matter can easily go undetected (Exercise 12).
2. **Pre-testing distortion.** Reporting only studies that pass a pre-test selects samples where noise happened to hide the trend, which can increase bias (Roth, 2022).
3. **Parallel pre-trends do not guarantee parallel post-trends.** The assumption is about the counterfactual after treatment.

### Remedies and sensitivity

- **Group-specific linear trends:** fit a line through the pre-period differences and extrapolate. Unbiased if the violation is exactly linear; can absorb part of a dynamic effect if not.
- **Bounded violations (Rambachan and Roth):** assume the post-treatment deviation from parallel trends is at most $M$ times the largest pre-treatment deviation, and report the resulting interval for each $M$, plus the **breakdown** value of $M$ at which the effect is no longer significant.
- **Conditional parallel trends:** assume parallel trends only among units with the same pre-treatment covariates; estimate with regression, weighting or doubly robust DiD.

### Anticipation

If behaviour changes one period before adoption, the coefficient at $k = -1$ is already affected. Use $k = -2$ (or earlier) as the reference.

### Exercise 6 · Event-study coefficients identify dynamic ATTs

With one treated and one comparison group, balanced data and the saturated event-study regression, prove $\hat\beta_k$ equals the DiD of means between periods $k$ and $-1$, and that under parallel trends and no anticipation it is unbiased for $ATT_k$ ($k \ge 0$) and for 0 ($k < 0$).

<details><summary>Show solution</summary>

**Step 1.** The regression has one parameter per group-period cell (unit effects absorb group levels, period effects the comparison path, $\beta_k$ the treated deviations); it is saturated in group × period means.

**Step 2.** So fitted cell means equal sample cell means, and $\hat\beta_k$ is the treated-minus-comparison change from period $-1$ to $k$.

**Step 3.** By Day 67, Exercise 9, its expectation is $ATT_k + \Delta_k$, where $\Delta_k$ is the difference in untreated trends between $-1$ and $k$; under parallel trends $\Delta_k = 0$.

**Step 4.** For $k < 0$ the ATT is 0 (no anticipation), so $E[\hat\beta_k] = 0$. $\blacksquare$

</details>

### Exercise 7 · Linear trend adjustment

Suppose the violation is linear: $\Delta_k = s\,(k + 1)$ for all $k$. Show that $\hat\beta_k - \hat s(k + 1)$, with $\hat s$ fitted on pre-periods, is unbiased for $ATT_k$.

<details><summary>Show solution</summary>

**Step 1.** Pre-period coefficients have expectation $s(k + 1)$, so a least-squares slope through them (and the origin at $k = -1$) is unbiased for $s$.

**Step 2.** $E[\hat\beta_k] = ATT_k + s(k + 1)$ for $k \ge 0$.

**Step 3.** Subtract $E[\hat s](k + 1) = s(k + 1)$. $\blacksquare$ If the violation is not linear, the adjustment is biased.

</details>

### Exercise 8 · Power of a pre-trend test

A pre-period coefficient has expectation $\mu$ and standard error $\sigma$ (normal). Show the power of a two-sided 5% test is $\Phi(-1.96 - \mu/\sigma) + 1 - \Phi(1.96 - \mu/\sigma)$.

<details><summary>Show solution</summary>

**Step 1.** The t-statistic is $N(\mu/\sigma, 1)$.

**Step 2.** Rejection: $|t| > 1.96$.

**Step 3.** $P(t > 1.96) = 1 - \Phi(1.96 - \mu/\sigma)$ and $P(t < -1.96) = \Phi(-1.96 - \mu/\sigma)$. Add. $\blacksquare$

</details>

### Exercise 9 · Bounded violations

Assume the post-period violation satisfies $|\Delta_{post}| \le M\bar\Delta_{pre}$, where $\bar\Delta_{pre}$ is the largest absolute pre-period deviation (treated as known). Show that $ATT \in [\hat\beta - M\bar\Delta_{pre},\; \hat\beta + M\bar\Delta_{pre}]$ in population, and give a 95% version.

<details><summary>Show solution</summary>

**Step 1.** $\beta = ATT + \Delta_{post}$, so $ATT = \beta - \Delta_{post}$.

**Step 2.** With $\Delta_{post} \in [-M\bar\Delta_{pre}, M\bar\Delta_{pre}]$, $ATT \in [\beta - M\bar\Delta_{pre}, \beta + M\bar\Delta_{pre}]$.

**Step 3.** Accounting for sampling error in $\hat\beta$: $[\hat\beta - M\bar\Delta_{pre} - 1.96\,\operatorname{SE},\; \hat\beta + M\bar\Delta_{pre} + 1.96\,\operatorname{SE}]$ covers the ATT with probability at least 95%. $\blacksquare$

</details>

### Exercise 10 · Anticipation and the reference period

Suppose units react one period early, so $E[\hat\beta_{-1}]$ would be $a \ne 0$ if it were estimated. Show that normalising $\beta_{-1} = 0$ shifts every other coefficient by $-a$, and that using $k = -2$ as the reference removes the bias.

<details><summary>Show solution</summary>

**Step 1.** Each $\hat\beta_k$ is a DiD relative to period −1: $E[\hat\beta_k] = (ATT_k + \Delta_k) - (a + \Delta_{-1})$; with parallel trends $\Delta = 0$, it is $ATT_k - a$.

**Step 2.** With $k = -2$ as reference (no anticipation there), $E[\hat\beta_k] = ATT_k$, and the coefficient at $k = -1$ estimates $a$. $\blacksquare$

</details>

### Exercise 15 · Conditional parallel trends

States that expanded Medicaid had lower baseline uninsured rates, and uninsured rates predict mortality trends. How can covariates restore parallel trends, and what is assumed?

<details><summary>Show solution</summary>

**Step 1.** Assume parallel trends conditional on baseline covariates $X$ (uninsured rate, income, age structure): $E[\Delta Y(0) \mid G = 1, X] = E[\Delta Y(0) \mid G = 0, X]$.

**Step 2.** Estimate by comparing changes within $X$ cells, by outcome regression of $\Delta Y$ on $X$ in comparison states, or by doubly robust DiD (propensity weighting plus outcome regression).

**Step 3.** Use only pre-treatment covariates: post-treatment ones may be affected by the expansion (Day 53).

**Answer:** conditional parallel trends plus overlap in $X$.

</details>

## Verify in Python

*Your work:*

```python
import numpy as np
from scipy.stats import norm

def es(T, C, ref):
    """Event-study coefficients from treated and comparison means; ref = index of period -1."""
    return np.round([(T[k] - T[ref]) - (C[k] - C[ref]) for k in range(len(T))], 3)
print("Ex 3:", es([10, 11, 12, 16, 17], [8, 9, 10, 11, 12], 2))
print("Ex 11:", es([30, 30.5, 31, 30, 29.5, 29.3], [32, 32.4, 32.9, 33.2, 33.5, 33.7], 2))
m = -.9 / .5
print("pre-test power:", round(norm.cdf(-1.96 - m) + 1 - norm.cdf(1.96 - m), 3))
for M in (1, 2):
    print(f"M={M}: [{2 - .4 * M - .98:.2f}, {2 + .4 * M + .98:.2f}]", end="  ")
print("breakdown M:", round((2 - .98) / .4, 2))

# Event-study regression with unit and time fixed effects on a simulated panel (linear pre-trend 0.3)
rng = np.random.default_rng(68)
N, Tn, E = 200, 8, 5
G = (np.arange(N) < N // 2).astype(float)
t = np.arange(Tn); k = t - E
Y = (rng.normal(0, 2, N)[:, None] + 0.5 * t[None, :] + 0.3 * k[None, :] * G[:, None]
     + G[:, None] * np.where(k >= 0, 1 + 0.5 * k, 0)[None, :] + rng.normal(0, 1, (N, Tn)))
rows = []
for i in range(N):
    for s in range(Tn):
        r = [1.0] + [float(i == j) for j in range(1, N)] + [float(s == q) for q in range(1, Tn)]
        r += [G[i] * float(k[s] == kk) for kk in range(-E, Tn - E) if kk != -1]
        rows.append(r)
b = np.linalg.lstsq(np.array(rows), Y.ravel(), rcond=None)[0][-(Tn - 1):]
labels = [kk for kk in range(-E, Tn - E) if kk != -1]
print("event-study betas:", {kk: round(float(v), 2) for kk, v in zip(labels, b)})
print("truth (effect + 0.3*(k+1) trend):", {kk: round((1 + .5 * kk if kk >= 0 else 0) + .3 * (kk + 1), 2) for kk in labels})
```

## Key takeaways

- Event-study coefficients are DiDs relative to a reference period; pre-period coefficients test for diverging trends before treatment.
- Pre-trend tests have limited power, and passing them does not establish parallel trends after treatment.
- Report sensitivity to bounded violations (and the breakdown value), re-normalise if there is anticipation, and use conditional parallel trends when observable differences predict trends.
