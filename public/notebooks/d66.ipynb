## Theory

### Over-identifying restrictions

With $l$ instruments and $k$ endogenous regressors ($l > k$), the moment conditions $E[\mathbf z_i\varepsilon_i] = \mathbf 0$ are more than needed. If all instruments are valid, every just-identified subset estimates the same $\beta$ (with a constant effect), and the residual should be unrelated to every instrument.

### The Sargan–Hansen J test

1. Estimate by 2SLS and compute residuals $\hat\varepsilon = \mathbf y - \mathbf X\hat{\boldsymbol\beta}$.
2. Regress $\hat\varepsilon$ on all instruments (and controls); record $R^2$.
3. $J = nR^2 \sim \chi^2_{l - k}$ under the null that all instruments are valid (homoskedastic errors). Hansen's J, the minimised two-step GMM objective, is the robust version.

### What J can and cannot do

- If one instrument is invalid and others valid, J has power to detect the disagreement, but it cannot say which instrument is wrong.
- If all instruments are invalid **in the same way** (estimates biased by the same amount), J has no power (Exercise 9).
- With heterogeneous effects, valid instruments can identify different LATEs (Day 63), so J can reject even when every instrument is valid.

### Is the regressor endogenous?

The **Durbin–Wu–Hausman** test compares OLS and IV. A convenient regression version is the **control function**: (1) run the first stage and save the residual $\hat v$; (2) regress $Y$ on $X$, controls and $\hat v$. The coefficient on $X$ equals 2SLS, and a significant coefficient on $\hat v$ rejects exogeneity of $X$.

### Plausibly exogenous instruments

If $Z$ has a small direct effect $\gamma$ on $Y$, then $\beta_{IV} = \beta + \gamma/\pi$. Allowing $\gamma \in [-\delta, \delta]$, the union of the IV confidence intervals shifted by $\gamma/\pi$ gives a sensitivity interval (Conley, Hansen and Rossi). Weak first stages magnify small violations.

### Exercise 7 · The control function reproduces 2SLS

Scalar endogenous $x$, instrument vector $\mathbf Z$, first stage $x = \mathbf Z\hat\pi + \hat v$. Prove that in the OLS regression of $y$ on $x$ and $\hat v$, the coefficient on $x$ equals 2SLS.

<details><summary>Show solution</summary>

**Step 1.** By FWL, the coefficient on $x$ uses $x$ residualised on $\hat v$.

**Step 2.** $\hat v$ is orthogonal to $\mathbf Z\hat\pi = \hat x$, and $x = \hat x + \hat v$, so residualising $x$ on $\hat v$ gives exactly $\hat x$.

**Step 3.** The coefficient is $\hat x^\top y/\hat x^\top\hat x$, the 2SLS estimator. $\blacksquare$

</details>

### Exercise 8 · Disagreement is what J detects

With two instruments (one endogenous regressor), show that if both single-instrument estimands are equal, the population over-identifying restriction holds in the sense that some $\beta$ satisfies both moment conditions.

<details><summary>Show solution</summary>

**Step 1.** Moment conditions: $E[z_j(y - \beta x)] = 0$ for $j = 1, 2$, i.e. $\beta = E[z_j y]/E[z_j x]$.

**Step 2.** If both ratios equal the same $b$, then $\beta = b$ solves both.

**Step 3.** Conversely, if the ratios differ, no single $\beta$ solves both, and J diverges with $n$. $\blacksquare$

</details>

### Exercise 9 · No power against common invalidity

Suppose each instrument has a direct effect proportional to its first stage: $\delta_j = c\,\pi_j$. Show that every single-instrument estimand equals $\beta + c$, so J has no power.

<details><summary>Show solution</summary>

**Step 1.** From Day 62, Exercise 10: instrument $j$ gives $\beta + \delta_j/\pi_j$.

**Step 2.** $\delta_j/\pi_j = c$ for all $j$, so all estimands equal $\beta + c$.

**Step 3.** By Exercise 8, the moment conditions are jointly satisfied at $\beta + c$: J converges to $\chi^2_{l-k}$ as under validity. $\blacksquare$

### Exercise 10 · Plausibly exogenous bounds

Let $Y = \beta X + \gamma Z + \varepsilon$ with $\gamma \in [-\delta, \delta]$ unknown. Show that the IV estimand is $\beta + \gamma/\pi$ and that $\beta \in [\beta_{IV} - \delta/|\pi|,\; \beta_{IV} + \delta/|\pi|]$.

<details><summary>Show solution</summary>

**Step 1.** $\operatorname{Cov}(Z, Y) = \beta\operatorname{Cov}(Z, X) + \gamma\operatorname{Var}(Z)$.

**Step 2.** Divide by $\operatorname{Cov}(Z, X) = \pi\operatorname{Var}(Z)$: $\beta_{IV} = \beta + \gamma/\pi$.

**Step 3.** $\beta = \beta_{IV} - \gamma/\pi$, which ranges over the stated interval as $\gamma$ ranges over $[-\delta, \delta]$. $\blacksquare$

</details>

## Verify in Python

*Your work:*

```python
import numpy as np
from scipy import stats

rng = np.random.default_rng(66); n = 1000
def tsls(y, X, Z):
    Xh = Z @ np.linalg.lstsq(Z, X, rcond=None)[0]
    return np.linalg.solve(Xh.T @ Xh, Xh.T @ y)

print("direct effect  J rejection  mean 2SLS (truth 0.5)")
for d in (0, .05, .10):
    rej, est = 0, []
    for _ in range(1000):
        Z = rng.normal(size=(n, 2)); u = rng.normal(size=n); v = .5 * u + rng.normal(size=n)
        x = Z @ [.3, .3] + v; y = 1 + .5 * x + d * Z[:, 1] + u
        X = np.column_stack([np.ones(n), x]); ZZ = np.column_stack([np.ones(n), Z])
        b = tsls(y, X, ZZ); r = y - X @ b
        g = np.linalg.lstsq(ZZ, r, rcond=None)[0]
        R2 = 1 - ((r - ZZ @ g) @ (r - ZZ @ g)) / ((r - r.mean()) @ (r - r.mean()))
        rej += n * R2 > stats.chi2.ppf(.95, 1); est.append(b[1])
    print(f"{d:13.2f}  {rej / 1000:11.3f}  {np.mean(est):.3f}")

# Exercise 7: the control function reproduces 2SLS
Z = rng.normal(size=(n, 2)); u = rng.normal(size=n); x = Z @ [.3, .3] + .5 * u + rng.normal(size=n); y = .5 * x + u
ZZ = np.column_stack([np.ones(n), Z]); vhat = x - ZZ @ np.linalg.lstsq(ZZ, x, rcond=None)[0]
cf = np.linalg.lstsq(np.column_stack([np.ones(n), x, vhat]), y, rcond=None)[0]
print("control function:", cf[1].round(5), " 2SLS:", tsls(y, np.column_stack([np.ones(n), x]), ZZ)[1].round(5), " residual coef:", cf[2].round(3))

# Exercises 2, 3 and 14
print("J:", 1000 * .006, "crit", round(stats.chi2.ppf(.95, 2), 2), "| z:", round(.05 / np.sqrt(.02**2 + .04**2), 2),
      "| sensitivity interval:", np.round([.05 - .01 / .32, .21 + .01 / .32], 3))
```

## Key takeaways

- With extra instruments, J = nR² tests whether the instruments agree; it cannot identify which is wrong and has no power when all are wrong in the same way.
- The control-function regression reproduces 2SLS and tests whether OLS is biased.
- With heterogeneous effects, disagreement can reflect different LATEs; sensitivity intervals show how much a small exclusion violation matters.
