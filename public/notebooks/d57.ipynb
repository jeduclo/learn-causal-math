Who ends up in your data is itself a variable. Add a selection node S to the DAG; analysing only S = 1 is conditioning on S, and when S is a collider or a descendant of the outcome, associations, even in randomised trials, are distorted. Graphs say when the bias can be undone and how.

**Learning outcomes**

- Add a selection node to a DAG and diagnose collider-stratification, outcome-dependent and survivorship bias.
- State and prove the conditions under which P(y | x) and odds ratios are recoverable from selected data.
- Correct selection bias with external covariate data or inverse-probability-of-selection weights.
- Recognise selection bias in case-control studies, testing data, wage regressions and trials with loss to follow-up.

## Theory

### The selection node

Let $S = 1$ indicate that a unit is in the data set. Draw $S$ in the DAG with arrows from whatever determines inclusion. Every analysis of the sample conditions on $S = 1$, so the d-separation rules apply: if $S$ is a collider (or a descendant of one) on a path between $X$ and $Y$, that path is opened.

### Typical structures

| Structure | Example | Effect |
|---|---|---|
| $X \to S \leftarrow Y$ | hospital patients (Berkson), tested individuals | spurious association |
| $X \to S \leftarrow U \to Y$ | trial drop-out due to side effects and poor health | bias even with randomised $X$ |
| $Y \to S$ | wages observed only for the employed, case-control sampling | distorted $P(y \mid x)$ |
| $X \to M \to S$, $M \leftarrow U \to Y$ | studying survivors, firms that survived | survivorship bias |

### Recoverability

With only selected data we can estimate $P(\mathbf v \mid S = 1)$. A quantity is **recoverable** if it can be written in terms of this (plus any external data).

- **Conditional distributions:** $P(y \mid x)$ is recoverable from the selected sample alone if $(Y \perp\!\!\!\perp S \mid X)$ in the graph; then $P(y \mid x) = P(y \mid x, S = 1)$ (Bareinboim and Pearl).
- **With external data on covariates:** if $(Y \perp\!\!\!\perp S \mid X, \mathbf Z)$ and $P(\mathbf z \mid x)$ is known from a population source (a census),

$$
P(y \mid x) = \sum_{\mathbf z} P(y \mid x, \mathbf z, S = 1)\,P(\mathbf z \mid x).
$$

- **Weighting:** equivalently, if $Y \perp\!\!\!\perp S \mid \mathbf Z$ and $P(S = 1 \mid \mathbf z) > 0$ is known or estimable, weight selected units by $1/P(S = 1 \mid \mathbf z)$ (inverse probability of selection weighting; Day 14 used the same idea for missing data).
- **Odds ratios:** if selection depends only on $Y$ (and on nothing else that relates to $X$), the $X$–$Y$ odds ratio is unchanged by selection. That is why case-control studies estimate odds ratios, not risks.

### Causal effects under selection

For $P(y \mid \operatorname{do}(x))$ we need two things at once: a back-door set $\mathbf Z$, and a way to undo selection, for example $(Y \perp\!\!\!\perp S \mid X, \mathbf Z)$ plus population data on $\mathbf Z$:

$$
P(y \mid \operatorname{do}(x)) = \sum_{\mathbf z} P(y \mid x, \mathbf z, S = 1)\,P(\mathbf z).
$$

Randomisation removes confounding but **not** selection bias created after randomisation.

### Exercise 5 · Survivors

A study of 10-year-old firms finds that firms that started with high debt are as profitable as low-debt firms. Draw a DAG with debt $D$, survival $S$, managerial quality $Q$ (unobserved) and profitability $P$, and explain why the finding may mislead.

### Exercise 6 · Selection independent of the outcome

Prove that if $(Y \perp\!\!\!\perp S \mid X)$, then $P(y \mid x, S = 1) = P(y \mid x)$.

<details><summary>Show solution</summary>

**Step 1.** By the global Markov property, the d-separation implies $Y \perp\!\!\!\perp S \mid X$ in the distribution.

**Step 2.** So $P(y \mid x, s) = P(y \mid x)$ for every $s$ with $P(x, s) > 0$, in particular $s = 1$. $\blacksquare$

</details>

### Exercise 7 · Odds-ratio invariance

Suppose selection depends only on $Y$: $P(S = 1 \mid x, y) = s_y$. Prove that the odds ratio of $X$ and $Y$ is the same in the selected sample as in the population.

<details><summary>Show solution</summary>

**Step 1.** $P(x, y \mid S = 1) = \dfrac{P(x, y)\,s_y}{P(S = 1)}$.

**Step 2.** The odds ratio is $\dfrac{P(1, 1)P(0, 0)}{P(1, 0)P(0, 1)}$ (arguments $x, y$).

**Step 3.** In the sample, each cell is multiplied by $s_y/P(S = 1)$; numerator and denominator each contain one $y = 1$ and one $y = 0$ cell, so the factors $s_1 s_0/P(S = 1)^2$ cancel. $\blacksquare$

Risk ratios do not cancel this way, so case-control studies cannot estimate absolute risks without external information.

</details>

### Exercise 8 · Recovery with external covariate data

Prove: if $(Y \perp\!\!\!\perp S \mid X, \mathbf Z)$, then $P(y \mid x) = \sum_{\mathbf z} P(y \mid x, \mathbf z, S = 1)P(\mathbf z \mid x)$.

<details><summary>Show solution</summary>

**Step 1.** Law of total probability: $P(y \mid x) = \sum_{\mathbf z} P(y \mid x, \mathbf z)P(\mathbf z \mid x)$.

**Step 2.** By the d-separation and Exercise 6's argument, $P(y \mid x, \mathbf z) = P(y \mid x, \mathbf z, S = 1)$.

**Step 3.** Substitute. $\blacksquare$ The first factor comes from the sample, the second from the census.

### Exercise 9 · Inverse probability of selection weighting

Assume $Y \perp\!\!\!\perp S \mid \mathbf Z$ and $\pi(\mathbf z) = P(S = 1 \mid \mathbf z) > 0$. Prove $E[Y] = E\big[SY/\pi(\mathbf Z)\big]$.

<details><summary>Show solution</summary>

**Step 1.** By the LIE, $E[SY/\pi(\mathbf Z)] = E\big[E[SY \mid \mathbf Z]/\pi(\mathbf Z)\big]$.

**Step 2.** $E[SY \mid \mathbf Z] = E[S \mid \mathbf Z]E[Y \mid \mathbf Z] = \pi(\mathbf Z)E[Y \mid \mathbf Z]$, by conditional independence.

**Step 3.** So $E[SY/\pi(\mathbf Z)] = E[E[Y \mid \mathbf Z]] = E[Y]$. $\blacksquare$

</details>

### Exercise 10 · Randomisation does not protect against drop-out

In a trial, $X$ is randomised, $Y = \tau X + U$, and units stay in the study ($S = 1$) when $X + U + e > 0$ (the programme's support visits encourage staying, and healthier patients stay), with $U, e$ independent normals. Explain with d-separation why $E[Y \mid X = 1, S = 1] - E[Y \mid X = 0, S = 1] \ne \tau$ in general.

<details><summary>Show solution</summary>

**Step 1.** DAG: $X \to Y$, $U \to Y$, $X \to S$, $U \to S$.

**Step 2.** Conditioning on $S$ opens $X \to S \leftarrow U \to Y$.

**Step 3.** Here $X = 1$ makes staying easier, so treated units stay even with low $U$, while controls need a higher $U$ to stay. Hence $E[U \mid X = 1, S = 1] < E[U \mid X = 0, S = 1]$.

**Step 4.** The completers' contrast equals $\tau + E[U \mid X = 1, S = 1] - E[U \mid X = 0, S = 1]$. $\blacksquare$

</details>

### Exercise 11 · A case-control study of smoking and lung cancer

In the population, 1% have lung cancer $D$. Among cases 60% smoke; among non-cases 20% smoke. A case-control study samples cases and an equal number of controls. Compute the odds ratio and the risk ratio in the population and in the study.

<details><summary>Show solution</summary>

**Step 1.** Population OR (exposure odds among cases versus controls): $\dfrac{0.6/0.4}{0.2/0.8} = \dfrac{1.5}{0.25} = 6$.

**Step 2.** Population risks: $P(D \mid \text{smoke}) = \dfrac{0.01(0.6)}{0.01(0.6) + 0.99(0.2)} = \dfrac{0.006}{0.204} \approx 0.0294$; $P(D \mid \text{no smoke}) = \dfrac{0.004}{0.004 + 0.792} \approx 0.00503$. RR $\approx 5.85$.

**Step 3.** In the 1:1 study (per 1 case, 1 control): smokers $0.6$ cases and $0.2$ controls, non-smokers $0.4$ and $0.8$. OR $= (0.6 \times 0.8)/(0.4 \times 0.2) = 6$.

**Step 4.** Study "risks": $0.6/0.8 = 0.75$ and $0.4/1.2 \approx 0.333$; RR $= 2.25$.

**Answer:** the OR (6) is preserved; the risk ratio is not (2.25 in the sample vs 5.85). With a rare disease the OR approximates the RR.

### Exercise 15 · Studying successful entrepreneurs

A business school surveys founders of firms that reached \$10 million in revenue and finds that founders without a university degree performed as well as graduates, concluding that "education doesn't matter for entrepreneurial success". Draw the DAG and explain the flaw. What data would answer the question?

<details><summary>Show solution</summary>

**Step 1.** DAG: education $E \to$ success $S$, talent and luck $T \to S$; the sample is selected on $S$ itself (the outcome).

**Step 2.** Within the sample, the outcome no longer varies much, and among successes, founders without degrees must have had more talent or luck: conditioning on the collider opens $E \to S \leftarrow T$.

**Answer:** selecting on the outcome destroys the comparison. Follow a cohort of all founders (or all potential founders) and compare success rates by education, adjusting for pre-founding confounders.

</details>

## Key takeaways

- Being in the data is a variable; analysing a selected sample conditions on it.
- Selection on a collider, on a descendant of the outcome, or on a post-treatment variable biases associations, even in randomised trials.
- Recovery needs a d-separation between outcome and selection given what you can measure, plus external data or selection weights; odds ratios survive outcome-only selection.
