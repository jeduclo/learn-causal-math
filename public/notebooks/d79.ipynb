## Theory

### The missing-data problem

For a binary treatment $D$, $E[Y(1)] = E[Y \mid D = 1]P(D = 1) + E[Y(1) \mid D = 0]P(D = 0)$. The data reveal everything except $E[Y(1) \mid D = 0]$ (and, symmetrically, $E[Y(0) \mid D = 1]$).

### Worst-case (Manski) bounds

If $Y \in [y_L, y_U]$, the unobserved term lies in $[y_L, y_U]$, so

$$
E[Y(1)] \in \big[E[Y \mid D = 1]p + y_L(1 - p),\;\; E[Y \mid D = 1]p + y_U(1 - p)\big], \quad p = P(D = 1),
$$

and similarly for $E[Y(0)]$. The ATE lower bound is the lower bound of $E[Y(1)]$ minus the upper bound of $E[Y(0)]$; the ATE upper bound is the upper bound of $E[Y(1)]$ minus the lower bound of $E[Y(0)]$. Their width is always $y_U - y_L$, and they always contain zero: data alone cannot sign the effect.

### Monotonicity assumptions (Manski and Pepper)

- **Monotone treatment response (MTR):** $Y(1) \ge Y(0)$ for everyone. The ATE is at least 0.
- **Monotone treatment selection (MTS):** those who choose treatment have weakly higher potential outcomes: $E[Y(d) \mid D = 1] \ge E[Y(d) \mid D = 0]$. Then the naive difference in means is an **upper** bound on the ATE.
- Together: $0 \le ATE \le E[Y \mid D = 1] - E[Y \mid D = 0]$.

### Lee bounds for selective attrition

In a randomised experiment, the outcome (wages) is observed only for some units (the employed), and treatment changes who is observed. Assume **monotonicity of selection**: treatment can only make units observed, never unobserved. Then observed treated units contain all "always-observed" units plus some units observed only because of treatment, a share

$$
p = \frac{q_1 - q_0}{q_1}
$$

of treated respondents, where $q_d$ is the response rate in arm $d$. Trimming the top $p$ (or bottom $p$) of the treated respondents' outcome distribution gives the lowest (or highest) possible mean for always-observed units; subtracting the control respondents' mean bounds the effect for the always-observed.

### Inference

Sample bounds are estimates. For a 95% confidence interval covering the true parameter (not the whole set), Imbens and Manski use $[\hat L - c\,\operatorname{SE}_L,\; \hat U + c\,\operatorname{SE}_U]$ with $c$ between 1.645 (wide sets) and 1.96 (narrow sets).

### Exercise 6 · Worst-case bounds

Prove the bounds on $E[Y(1)]$ and that they are sharp (attainable).

<details><summary>Show solution</summary>

**Step 1.** Law of total expectation: $E[Y(1)] = E[Y \mid D = 1]p + E[Y(1) \mid D = 0](1 - p)$, using consistency for the treated.

**Step 2.** $E[Y(1) \mid D = 0] \in [y_L, y_U]$ since $Y(1)$ is bounded.

**Step 3.** Substitute the extremes. Each value in the interval is attained by a distribution of $Y(1)$ among the untreated that is consistent with the data, so the bounds are sharp. $\blacksquare$

### Exercise 9 · Lee bounds

Assume random assignment and monotone selection ($S(1) \ge S(0)$). Show that treated respondents are a mixture of always-observed units (share $1 - p$) and units observed only if treated (share $p$), and that trimming the top $p$ gives a lower bound for the always-observed treated mean.

<details><summary>Show solution</summary>

**Step 1.** Control respondents have $S(0) = 1$: always-observed only (by monotonicity). Their share of the population is $q_0$.

**Step 2.** Treated respondents have $S(1) = 1$: always-observed ($q_0$) plus marginal units ($q_1 - q_0$); the marginal share among them is $p$.

**Step 3.** The always-observed treated outcomes are an unknown $(1 - p)$ part of the treated respondents' distribution; the smallest possible mean of such a part is the mean of the lowest $1 - p$ (trim the top $p$), the largest is the mean after trimming the bottom $p$.

**Step 4.** Control respondents' mean is the always-observed untreated mean. Subtract. $\blacksquare$

</details>

### Exercise 10 · Covering the parameter, not the set

Explain why $[\hat L - 1.645\,\operatorname{SE},\; \hat U + 1.645\,\operatorname{SE}]$ covers the true parameter with probability at least about 95% when the identified set is wide.

<details><summary>Show solution</summary>

**Step 1.** The parameter lies in $[L, U]$. If it is near $L$, only the lower end can miss, which happens with probability about 5% using the one-sided 1.645.

**Step 2.** When the set is wide, the parameter cannot be near both ends at once, so at most one tail matters.

**Answer:** coverage of the parameter is about 95%; for narrow sets the critical value rises towards 1.96 (Imbens–Manski). $\blacksquare$

</details>

```python
import numpy as np

def manski(p, m1, m0, lo=0, hi=1):
    y1 = (m1 * p + lo * (1 - p), m1 * p + hi * (1 - p)); y0 = (m0 * (1 - p) + lo * p, m0 * (1 - p) + hi * p)
    return np.round(y1, 4), np.round(y0, 4), np.round((y1[0] - y0[1], y1[1] - y0[0]), 4)
for name, args in [("Ex 1", (.4, .7, .5)), ("Ex 12", (.25, .30, .15))]:
    y1, y0, ate = manski(*args); print(f"{name}: E[Y(1)] {y1.tolist()}, E[Y(0)] {y0.tolist()}, ATE {ate.tolist()}")
print("Ex 14:", np.round([50 / 1e4 - 600 / 1e4, 550 / 1e4 - 200 / 1e4], 4).tolist())

# Lee bounds: Exercise 11 by formula, then a simulated experiment
p = (.9 - .8) / .9
print(f"Ex 11: trim share {p:.3f}, bounds [{900 * (1 - p) / 2 - 380:.0f}, {(900 * p + 900) / 2 - 380:.0f}]")
rng = np.random.default_rng(79); n = 400_000
Z = rng.random(n) < .5
t = rng.choice([0, 1, 2], n, p=[.8, .1, .1])          # always employed, employed only if treated, never
w0 = rng.lognormal(6, .5, n)
w = np.where(Z, 1.1 * w0, w0); w = np.where(t == 1, .7 * w, w)
S = (t == 0) | ((t == 1) & Z)
q1, q0 = S[Z].mean(), S[~Z].mean(); p = (q1 - q0) / q1
wt = np.sort(w[Z & S]); k = int(round(p * len(wt))); c = w[~Z & S].mean()
print(f"trim {p:.3f}; Lee bounds [{wt[:len(wt) - k].mean() - c:.1f}, {wt[k:].mean() - c:.1f}]; "
      f"truth for always-employed {(.1 * w0[t == 0]).mean():.1f}; naive {w[Z & S].mean() - c:.1f}")
```

## Key takeaways

- With a bounded outcome and no assumptions, the ATE is bounded by an interval of width $y_U - y_L$ that contains zero.
- Monotone response and selection assumptions narrow the bounds transparently; MTS makes the naive difference an upper bound.
- Lee bounds trim the treated respondents' distribution to handle outcomes observed only for some units, assuming treatment only adds respondents.
