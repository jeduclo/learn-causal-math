# Day 94: Policy Learning and Empirical Welfare Maximisation

*Beginner to Ninja: Math for Causal Inference*

Heterogeneous effects matter because they let us decide whom to treat. A treatment rule maps characteristics to a decision; its value is the average outcome (net of costs) if everyone were treated according to it. Doubly robust scores give unbiased estimates of any rule's value, and empirical welfare maximisation picks the best rule within a simple, transparent class, with guarantees on regret.

**Learning outcomes**

- Define a treatment rule, its welfare and its regret relative to the best rule.
- Estimate the value of any rule with AIPW scores and prove unbiasedness.
- Derive the optimal rule with costs and with a budget constraint.
- Implement empirical welfare maximisation over threshold rules and recognise its in-sample optimism.

## Theory

### Rules and welfare

A **treatment rule** is $\pi: \mathcal X \to \{0, 1\}$. With a per-person cost $c$, its welfare (per capita, relative to treating no one) is

$$
W(\pi) = E\big[\pi(X)(\tau(X) - c)\big].
$$

The **first-best** rule is $\pi^*(x) = \mathbf 1\{\tau(x) > c\}$. In practice we restrict to a class $\Pi$ (thresholds, shallow trees) for transparency, fairness or legal reasons; the **regret** of a chosen $\hat\pi$ is $\max_{\pi \in \Pi}W(\pi) - W(\hat\pi)$.

### Estimating welfare

With AIPW scores $\hat\varphi_i$ (Day 92),

$$
\widehat W(\pi) = \frac1n\sum_i\pi(X_i)\big(\hat\varphi_i - c\big)
$$

is (approximately) unbiased for $W(\pi)$, for **any** fixed rule.

### Empirical welfare maximisation (EWM)

Kitagawa and Tetenov (2018) choose $\hat\pi = \arg\max_{\pi \in \Pi}\widehat W(\pi)$. Regret is bounded by twice the maximal estimation error over the class,

$$
W(\pi^\dagger) - W(\hat\pi) \le 2\sup_{\pi \in \Pi}\big|\widehat W(\pi) - W(\pi)\big| = O_p\big(\sqrt{v/n}\big),
$$

where $v$ measures the complexity (VC dimension) of $\Pi$ and $\pi^\dagger$ is the best rule in $\Pi$. Athey and Wager (2021) show doubly robust scores achieve this rate with observational data.

### Plug-in versus direct

Alternatively estimate $\hat\tau$ and treat if $\hat\tau(x) > c$. This is good when $\hat\tau$ is accurate near the threshold, but it optimises prediction accuracy everywhere rather than decisions where they matter.

### Budgets

With capacity to treat a share $q$, treat the units with the highest $\tau(x)$ until the budget binds (provided $\tau > c$ for them).

### Honest evaluation

The maximised in-sample $\widehat W(\hat\pi)$ is biased upwards (the winner's curse of Day 93). Evaluate the chosen rule on held-out data.

### Exercise 6 · The first-best rule

Show $\pi^*(x) = \mathbf 1\{\tau(x) > c\}$ maximises $W(\pi)$ over all rules.

<details><summary>Show solution</summary>

**Step 1.** $W(\pi) = E[\pi(X)(\tau(X) - c)]$, and pointwise $\pi(x)(\tau(x) - c) \le \max\{0, \tau(x) - c\}$.

**Step 2.** $\pi^*$ attains this maximum at every $x$. $\blacksquare$

### Exercise 7 · AIPW welfare is unbiased

Assume $E[\varphi \mid X] = \tau(X)$. Show $E[\pi(X)(\varphi - c)] = W(\pi)$ for fixed $\pi$.

<details><summary>Show solution</summary>

**Step 1.** Iterated expectations: $E[\pi(X)(\varphi - c)] = E[\pi(X)(E[\varphi \mid X] - c)] = E[\pi(X)(\tau(X) - c)]$. $\blacksquare$

</details>

### Exercise 8 · The regret bound

Let $\hat\pi$ maximise $\widehat W$ over $\Pi$ and $\pi^\dagger$ maximise $W$. Show $W(\pi^\dagger) - W(\hat\pi) \le 2\sup_{\pi \in \Pi}|\widehat W(\pi) - W(\pi)|$.

<details><summary>Show solution</summary>

**Step 1.** $W(\pi^\dagger) - W(\hat\pi) = [W(\pi^\dagger) - \widehat W(\pi^\dagger)] + [\widehat W(\pi^\dagger) - \widehat W(\hat\pi)] + [\widehat W(\hat\pi) - W(\hat\pi)]$.

**Step 2.** The middle bracket is $\le 0$ because $\hat\pi$ maximises $\widehat W$.

**Step 3.** Each outer bracket is $\le\sup|\widehat W - W|$. $\blacksquare$

</details>

### Exercise 9 · The budget rule

With capacity share $q$ and continuous $\tau(X)$, show that among rules with $E[\pi(X)] \le q$, the one treating the top $q$ share of $\tau$ (if those exceed $c$) maximises welfare.

<details><summary>Show solution</summary>

**Step 1.** Exchange argument: suppose a rule treats some $x_a$ and not $x_b$ with $\tau(x_b) > \tau(x_a)$.

**Step 2.** Swapping equal masses keeps the budget and changes welfare by $\tau(x_b) - \tau(x_a) > 0$.

**Step 3.** So an optimal rule has no such pair: it treats a top set of $\tau$ (and never units with $\tau < c$). $\blacksquare$

</details>

### Exercise 10 · Comparing two rules

Show $W(\pi_1) - W(\pi_2) = E[(\pi_1 - \pi_2)(\tau - c)]$ and explain why its estimate depends only on units where the rules disagree.

<details><summary>Show solution</summary>

**Step 1.** Linearity of $W$ in $\pi$.

**Step 2.** $\pi_1 - \pi_2 = 0$ wherever the rules agree, so only disagreeing units enter $\frac1n\sum(\pi_1 - \pi_2)(\hat\varphi_i - c)$. $\blacksquare$ Comparisons of similar rules are therefore noisy relative to their small differences.

</details>

## Verify in Python

*Your work:*

```python
import numpy as np

rng = np.random.default_rng(94)
tau = lambda X: 1.5 - 2 * X[:, 0] + .5 * X[:, 1]            # gain in $1,000
c = .8                                                      # cost per participant
n = 2000; X = rng.uniform(0, 1, size=(n, 3)); D = rng.binomial(1, .5, n)
Y = 2 + X[:, 2] + tau(X) * D + 2 * rng.normal(size=n)

Z = lambda A: np.column_stack([np.ones(len(A)), A])
phi = np.empty(n); folds = np.array_split(rng.permutation(n), 2)
for k in range(2):                                          # cross-fitted AIPW scores (e = 0.5)
    te, tr = folds[k], folds[1 - k]
    b1 = np.linalg.lstsq(Z(X[tr][D[tr] == 1]), Y[tr][D[tr] == 1], rcond=None)[0]
    b0 = np.linalg.lstsq(Z(X[tr][D[tr] == 0]), Y[tr][D[tr] == 0], rcond=None)[0]
    m1, m0 = Z(X[te]) @ b1, Z(X[te]) @ b0
    phi[te] = m1 - m0 + D[te] * (Y[te] - m1) / .5 - (1 - D[te]) * (Y[te] - m0) / .5

best = (-np.inf,)
for j in range(3):
    for t in np.linspace(0, 1, 101):
        for s in (1, -1):
            w = np.mean((s * X[:, j] <= s * t) * (phi - c))
            if w > best[0]: best = (w, j, t, s)
w, j, t, s = best
print(f"EWM rule: treat if {'<=' if s == 1 else '>='} {t:.2f} on x{j + 1}; in-sample welfare {w:.3f}")

Xt = rng.uniform(0, 1, size=(200_000, 3)); nb = tau(Xt) - c
val = lambda pi: float(np.mean(pi * nb))
print(f"true welfare: treat all {val(1):.3f}, EWM {val(s * Xt[:, j] <= s * t):.3f}, first-best {val(nb > 0):.3f}")
bb = np.linalg.lstsq(Z(X), phi, rcond=None)[0]
print("plug-in CATE coefficients", bb.round(2), f" welfare {val(Z(Xt) @ bb > c):.3f}")
print("Ex 1: treat all", round(.5 * (2 - .8) + .5 * (.5 - .8), 3), " group 1 only", round(.5 * 1.2, 3))
```

## Key takeaways

- A treatment rule's welfare can be estimated without bias from AIPW scores, for any fixed rule.
- EWM picks the best rule in a transparent class, with regret shrinking like √(complexity/n); the first-best treats whenever τ(x) exceeds the cost.
- In-sample welfare of the chosen rule is optimistic; evaluate on held-out data and report the price of any constraints.
