When crossing the cutoff raises the probability of treatment without determining it, the discontinuity becomes an instrument. The ratio of the jump in outcomes to the jump in take-up identifies a local average effect for compliers at the cutoff, and is estimated by 2SLS within a bandwidth.

**Learning outcomes**

- Define the fuzzy RD estimand as a ratio of two jumps.
- Prove that it equals the average effect for compliers at the cutoff under continuity and monotonicity.
- Implement fuzzy RD as 2SLS with the same bandwidth and controls in both stages.
- Apply it to class size, insurance at 65 and scholarship take-up, with attention to weak first stages.

## Theory

### Fuzzy designs

In a **fuzzy RD**, crossing the cutoff changes the probability of treatment by less than one:

$$
\lim_{x \downarrow c}P(D = 1 \mid X = x) - \lim_{x \uparrow c}P(D = 1 \mid X = x) = \pi \in (0, 1).
$$

Examples: eligibility rules with imperfect take-up, guidelines that doctors sometimes override, class-size caps that schools partly ignore.

### Estimand

$$
\tau_{FRD} = \frac{\lim_{x \downarrow c}E[Y \mid X = x] - \lim_{x \uparrow c}E[Y \mid X = x]}{\lim_{x \downarrow c}E[D \mid X = x] - \lim_{x \uparrow c}E[D \mid X = x]}.
$$

It is a Wald estimator (Day 62) using $Z = \mathbf 1[X \ge c]$ as the instrument, evaluated at the cutoff. A sharp RD is the special case $\pi = 1$.

### Interpretation (Hahn, Todd and van der Klaauw)

Let $D(x')$ denote treatment if the cutoff rule were set as if the unit's running variable were $x'$, with types defined at the cutoff. Under continuity of potential outcomes and type shares in $x$, monotonicity (crossing the cutoff never discourages treatment) and exclusion (the cutoff affects $Y$ only through $D$),

$$
\tau_{FRD} = E[Y(1) - Y(0) \mid \text{complier}, X = c].
$$

It is doubly local: compliers, and at the cutoff.

### Estimation

Within bandwidth $h$ (kernel-weighted), run 2SLS:

- First stage: $D_i = \gamma_0 + \pi Z_i + \gamma_1(X_i - c) + \gamma_2 Z_i(X_i - c) + v_i$.
- Second stage: $Y_i = \alpha_0 + \tau D_i + \alpha_1(X_i - c) + \alpha_2 Z_i(X_i - c) + \varepsilon_i$.

With the same bandwidth, kernel and controls in both stages, $\hat\tau$ equals the ratio of the two local linear jumps. Bandwidths are chosen for the ratio (or for the reduced form), and RBC inference extends to the ratio.

### Practical points

Always plot the first stage. A small jump in take-up is a weak instrument (Day 65): report the reduced form and weak-instrument-robust intervals.

### Exercise 6 · Identification of the complier effect

Assume that, near the cutoff, $Z = \mathbf 1[X \ge c]$ acts as a randomly assigned instrument in the limit: type shares and the conditional means of $Y(1), Y(0)$ given type are continuous in $x$ at $c$. Under monotonicity and exclusion, prove $\tau_{FRD} = E[Y(1) - Y(0) \mid \text{complier}, X = c]$.

<details><summary>Show solution</summary>

**Step 1.** Just above $c$, treated units are compliers and always-takers; just below, always-takers only.

**Step 2.** Outcome jump: by continuity, always-takers' and never-takers' contributions are the same on both sides and cancel; compliers contribute $\pi_c(c)\,E[Y(1) - Y(0) \mid c, X = c]$.

**Step 3.** Take-up jump: $\pi_c(c)$.

**Step 4.** Divide (the LATE theorem at the cutoff). $\blacksquare$

</details>

### Exercise 7 · 2SLS equals the ratio of jumps

Show that, with a uniform kernel and the same window and controls $(1, X - c, Z(X - c))$ in both stages, 2SLS gives $\hat\tau = \hat\rho/\hat\pi$, the ratio of the reduced-form and first-stage jumps.

<details><summary>Show solution</summary>

**Step 1.** The model is just identified, with $Z$ the excluded instrument and the same controls $\mathbf W$ in both stages.

**Step 2.** By FWL, partial $\mathbf W$ out of $Y$, $D$ and $Z$: $\hat\tau = \tilde Z^\top\tilde Y/\tilde Z^\top\tilde D$.

**Step 3.** The reduced-form coefficient on $Z$ is $\tilde Z^\top\tilde Y/\tilde Z^\top\tilde Z$ and the first-stage one $\tilde Z^\top\tilde D/\tilde Z^\top\tilde Z$; their ratio is the same expression. $\blacksquare$

</details>

### Exercise 8 · A delta-method standard error

With reduced-form jump $\hat\rho$, first-stage jump $\hat\pi$ and covariance matrix of the two, write the delta-method variance of $\hat\tau$.

$$

### Exercise 15 · Same bandwidth in both stages

Why must the first stage and reduced form use the same bandwidth, kernel and polynomial order?

<details><summary>Show solution</summary>

**Step 1.** The ratio estimates the complier effect only if both jumps are measured on the same units with the same weights.

**Step 2.** Different windows would compare different populations (and different bias), so the ratio would not have the LATE interpretation.

**Answer:** use the 2SLS implementation within a common window, or a bandwidth selected for the ratio.

## Verify in Python

*Your work:*

```python
import numpy as np

print("Ex 1:", 6 / .4, " Ex 2 types:", (.15, round(1 - .70, 2), round(.70 - .15, 2)), " Ex 11:", 2 / -8, " Ex 12:", round(.8 / .09, 2), " Ex 13:", .04 / .5)

# Simulated fuzzy RD with heterogeneous effects
rng = np.random.default_rng(76); n = 200_000
x = rng.uniform(-1, 1, n); z = (x >= 0).astype(float)
typ = rng.choice(["c", "a", "n"], n, p=[.55, .15, .30])
d = ((typ == "a") | ((typ == "c") & (z == 1))).astype(float)
gain = np.select([typ == "c", typ == "a", typ == "n"], [3.0, 1.0, 0.5]) + 2 * x        # effects vary with type and x
y = 10 + 4 * x + 2 * (typ == "a") + d * gain + rng.normal(0, 2, n)
h = .2; m = np.abs(x) <= h
W = np.column_stack([np.ones(m.sum()), x[m], z[m] * x[m]])
def coef_on(v, target):
    X = np.column_stack([W, v]); return np.linalg.lstsq(X, target, rcond=None)[0][-1]
rho, pi = coef_on(z[m], y[m]), coef_on(z[m], d[m])
Xs = np.column_stack([W, d[m]]); Zs = np.column_stack([W, z[m]])
Xh = Zs @ np.linalg.lstsq(Zs, Xs, rcond=None)[0]
tsls = np.linalg.lstsq(Xh, y[m], rcond=None)[0][-1]
print(f"reduced form {rho:.3f}, first stage {pi:.3f}, ratio {rho / pi:.3f}, 2SLS {tsls:.3f}  (complier effect at cutoff 3.0)")
```

## Key takeaways

- Fuzzy RD divides the outcome jump by the take-up jump; it is IV with the cutoff indicator as instrument.
- It identifies the average effect for compliers at the cutoff, under continuity, monotonicity and exclusion.
- Estimate by 2SLS with a common window and controls; plot the first stage and guard against weak jumps.
