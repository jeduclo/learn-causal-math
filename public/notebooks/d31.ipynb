## Theory

### Potential outcomes

Phase 1 and 2 gave you the tools; Phase 3 gives them a causal meaning. The **Rubin causal model** (Neyman 1923, Rubin 1974) starts from three ingredients:

- **Units** $i = 1, \dots, n$: workers, households, patients, villages.
- A **treatment** $T_i \in \{0, 1\}$: training or not, a cash transfer or not.
- **Potential outcomes** $Y_i(1)$ and $Y_i(0)$: what unit $i$'s outcome *would be* under each treatment, both defined before treatment is assigned.

The **unit-level causal effect** is $\tau_i = Y_i(1) - Y_i(0)$. Collected in one table, the potential outcomes and covariates of all units form the **science table**: everything there is to know about the causal question.

### The observed outcome

We only ever see the outcome under the treatment actually received:

$$
Y_i = T_iY_i(1) + (1 - T_i)Y_i(0) = Y_i(0) + T_i\tau_i.
$$

For every unit, one potential outcome is observed and the other is **missing**: the *fundamental problem of causal inference* (Holland 1986). Causal inference is therefore a missing-data problem (Day 14).

### Estimands

For a fixed set of $n$ units, the **finite-population ATE** is

$$
\tau_{fp} = \frac1n\sum_{i=1}^n\big(Y_i(1) - Y_i(0)\big) = \bar Y(1) - \bar Y(0).
$$

If the units are a random sample from a larger population, the **super-population ATE** is $E[Y(1) - Y(0)]$. Day 33 adds ATT, ATC and CATE.

### The assignment mechanism

The **assignment mechanism** is the rule that decides who gets treated: $P(T_1, \dots, T_n \mid \text{science table})$. It is the crux of every design:

- In a **randomised experiment** the researcher controls it and it does not depend on potential outcomes.
- In an **observational study** people (or officials) choose, often based on what they expect to gain.

With **complete randomisation**, exactly $n_1$ of $n$ units are treated and each of the $\binom{n}{n_1}$ assignments is equally likely. Then the difference in means $\hat\tau = \bar Y_1 - \bar Y_0$ is unbiased for $\tau_{fp}$ (Exercise 8).

### What counts as a treatment

"No causation without manipulation" (Holland): $Y(1)$ must be well defined. "The effect of a job-training voucher" is; "the effect of being obese" is not, because there are many ways to change weight (diet, exercise, surgery) with different consequences. Day 32 formalises this as SUTVA.

## Part 2 · Proofs

### Exercise 6 · The observed-outcome identity

Prove $T_iY_i(1) + (1 - T_i)Y_i(0) = Y_i(0) + T_i\tau_i$.

<details><summary>Show solution</summary>

**Step 1.** Expand the left side: $T_iY_i(1) + Y_i(0) - T_iY_i(0)$.

**Step 2.** Group the terms with $T_i$: $Y_i(0) + T_i\big(Y_i(1) - Y_i(0)\big) = Y_i(0) + T_i\tau_i$. $\blacksquare$

</details>

*Your work:*

### Exercise 7 · ATE as a difference of means

Prove $\tau_{fp} = \bar Y(1) - \bar Y(0)$.

<details><summary>Show solution</summary>

**Step 1.** $\tau_{fp} = \frac1n\sum_i(Y_i(1) - Y_i(0))$.

**Step 2.** Split the sum (linearity): $\frac1n\sum_iY_i(1) - \frac1n\sum_iY_i(0) = \bar Y(1) - \bar Y(0)$. $\blacksquare$

So we never need individual effects to learn the ATE, only the two means.

</details>

*Your work:*

### Exercise 8 · Complete randomisation is unbiased

Under complete randomisation with $n_1$ treated, prove $E[\bar Y_1 - \bar Y_0] = \tau_{fp}$, where the expectation is over assignments and the science table is fixed.

<details><summary>Show solution</summary>

**Step 1.** Write the treated mean with indicators: $\bar Y_1 = \frac{1}{n_1}\sum_iT_iY_i(1)$ (for treated units the observed outcome is $Y_i(1)$).

**Step 2.** Only $T_i$ is random, and by symmetry $E[T_i] = P(T_i = 1) = n_1/n$.

**Step 3.** So $E[\bar Y_1] = \frac{1}{n_1}\sum_i\frac{n_1}{n}Y_i(1) = \bar Y(1)$.

**Step 4.** Likewise $\bar Y_0 = \frac{1}{n_0}\sum_i(1 - T_i)Y_i(0)$ and $E[\bar Y_0] = \bar Y(0)$.

**Step 5.** Linearity and Exercise 7: $E[\bar Y_1 - \bar Y_0] = \bar Y(1) - \bar Y(0) = \tau_{fp}$. $\blacksquare$

</details>

### Exercise 10 · Post-treatment variables have potential values too

A training programme may change whether a worker moves city, $M$. Explain why $M$ has potential values $M(1), M(0)$, and why comparing trained and untrained workers *who did not move* is not a causal comparison.

<details><summary>Show solution</summary>

**Step 1.** Moving happens after assignment and can be affected by it, so each worker has $M_i(1)$ and $M_i(0)$.

**Step 2.** "Trained non-movers" are workers with $M_i(1) = 0$; "untrained non-movers" have $M_i(0) = 0$. These are generally different sets of people.

**Answer:** the comparison mixes the effect of training with differences between two selected groups. Only pre-treatment covariates (fixed before assignment) can be conditioned on safely; this is the "bad control" problem of Day 53.

</details>

Now suppose workers know their gains and enrol only if they gain (A and C). Compute the difference in means and compare with the ATT.

### Exercise 13 · Is the treatment well defined?

A ministry asks: "What is the effect of obesity on annual health-care costs?" Explain why this is ill-posed in the potential-outcomes framework, and reformulate it as a well-defined question.

<details><summary>Show solution</summary>

**Step 1.** $Y_i(\text{not obese})$ is ambiguous: losing weight through exercise, diet, medication or surgery would change costs differently.

**Answer:** causal questions need a specific intervention; the reformulation also matches what a policymaker can actually do.

### Exercise 14 · Finite or super population?

An evaluation studies all 120 municipalities of a province, randomising a governance reform. Another samples 2,000 households at random from a country and randomises a cash transfer. Which estimand is natural in each case?

<details><summary>Show solution</summary>

**Step 1.** The 120 municipalities *are* the population of interest; uncertainty comes only from which ones were assigned the reform: the finite-population ATE is natural.

**Step 2.** The 2,000 households stand in for millions; uncertainty comes from sampling *and* assignment: the super-population ATE is natural.

**Answer:** finite-population for the municipalities, super-population for the households. Day 36 shows the two lead to slightly different variance formulas.

</details>

### Exercise 15 · Who does a vaccine help?

Six people are classified by their potential infection outcomes: 1 "doomed" ($Y(0) = Y(1) = 1$), 2 "helped" ($Y(0) = 1$, $Y(1) = 0$), 3 "immune" ($Y(0) = Y(1) = 0$). Compute the infection rate with and without the vaccine and the ATE. Could a trial reveal the number helped?

<details><summary>Show solution</summary>

**Step 1.** Without the vaccine, infected = doomed + helped = 3 of 6: $\bar Y(0) = 0.5$.

**Step 2.** With the vaccine, infected = doomed = 1 of 6: $\bar Y(1) \approx 0.167$.

**Step 3.** ATE $= 0.167 - 0.5 \approx -0.333$: the vaccine cuts the infection rate by a third of the population.

**Step 4.** A trial identifies the two rates, hence the ATE. But a fourth type, "hurt" ($Y(0) = 0$, $Y(1) = 1$), would also be possible: 3 helped and 1 hurt give the same rates as 2 helped and 0 hurt.

**Answer:** ATE = −1/3. The share helped equals −ATE only if nobody is hurt (a monotonicity assumption); the data alone cannot tell.

</details>

## Verify in Python

*Your work:*

```python
import itertools
import numpy as np

y0 = np.array([10, 12, 8, 15.]); y1 = np.array([14, 12, 13, 15.])
print("unit effects:", y1 - y0, " ATE:", (y1 - y0).mean())

ests = []
for treated in itertools.combinations(range(4), 2):
    T = np.zeros(4, bool); T[list(treated)] = True
    Y = np.where(T, y1, y0)                         # observed outcomes
    ests.append(Y[T].mean() - Y[~T].mean())
    print("treated", [("ABCD")[i] for i in treated], "estimate", ests[-1])
print("average over all assignments:", np.mean(ests), "(= ATE)")
print("variance across assignments:", round(np.var(ests), 4))

# Self-selection (Exercise 12)
T = (y1 - y0) > 0; Y = np.where(T, y1, y0)
print("self-selected comparison:", Y[T].mean() - Y[~T].mean(), " ATT:", (y1 - y0)[T].mean())
```

## Key takeaways

- A causal effect compares two potential outcomes of the same unit; only one is ever observed.
- The assignment mechanism decides whether observed comparisons recover causal effects.
- Complete randomisation makes the difference in means unbiased for the ATE, averaging over all possible assignments.
