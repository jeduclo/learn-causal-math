## Theory

### Assignment is not receipt

In many experiments the researcher randomises an **offer** $Z$ (a voucher, an invitation, an encouragement letter), but people decide whether to take up the **treatment** $D$. Each unit has potential take-up $D(1), D(0)$ and potential outcomes. Four **compliance types**:

| Type | $D(1)$ | $D(0)$ |
|---|---|---|
| Complier | 1 | 0 |
| Always-taker | 1 | 1 |
| Never-taker | 0 | 0 |
| Defier | 0 | 1 |

Types are defined by potential take-up, so they are fixed before assignment and unaffected by it.

### Intention-to-treat

Random assignment of $Z$ identifies two **intention-to-treat** effects:

$$
\text{ITT}_D = E[D \mid Z = 1] - E[D \mid Z = 0], \qquad \text{ITT}_Y = E[Y \mid Z = 1] - E[Y \mid Z = 0].
$$

$\text{ITT}_Y$ is the effect of *offering* the programme: often exactly the policy-relevant quantity (governments can offer, not compel).

### The effect for compliers

Assume (i) $Z$ is randomised; (ii) **exclusion**: $Z$ affects $Y$ only through $D$; (iii) **monotonicity**: no defiers; (iv) $\text{ITT}_D > 0$. Then

$$
\text{CACE} = E[Y(1) - Y(0) \mid \text{complier}] = \frac{\text{ITT}_Y}{\text{ITT}_D}.
$$

This is the Wald (IV) estimator (Days 12 and 19); Day 63 develops it as the LATE. With **one-sided non-compliance** (no one assigned to control can get treated), there are no always-takers, so compliers are exactly the treated and CACE = ATT (Bloom 1984).

### What not to do

- **As-treated:** compare those who took the treatment with those who did not, ignoring $Z$. This is an observational comparison with self-selection.
- **Per-protocol:** compare those who followed their assignment in each arm. The arms then contain different types (e.g. compliers + always-takers versus compliers + never-takers), so randomisation is lost.

## Part 2 · Proofs

### Exercise 6 · ITT = complier share × CACE

Under randomisation, exclusion and no defiers, prove $\text{ITT}_Y = \pi_c\,\text{CACE}$ and $\text{ITT}_D = \pi_c$, where $\pi_c$ is the complier share.

<details><summary>Show solution</summary>

**Step 1.** By exclusion, write each unit's outcome under assignment $z$ as $Y(D(z))$. Then $Y_i(Z{=}1) - Y_i(Z{=}0) = Y_i(D_i(1)) - Y_i(D_i(0))$.

**Step 2.** By type: always-takers and never-takers have $D(1) = D(0)$, so their difference is 0. Compliers have $D(1) = 1$, $D(0) = 0$, so their difference is $Y_i(1) - Y_i(0)$. There are no defiers.

**Step 3.** Randomisation makes $\text{ITT}_Y$ the average of these differences: $\text{ITT}_Y = \pi_c\,E[Y(1) - Y(0) \mid \text{complier}] = \pi_c\,\text{CACE}$.

**Step 4.** Same argument for take-up: $D(1) - D(0)$ is 1 for compliers and 0 otherwise, so $\text{ITT}_D = \pi_c$.

**Step 5.** Divide: $\text{CACE} = \text{ITT}_Y/\text{ITT}_D$. $\blacksquare$

</details>

### Exercise 7 · Bloom's result

With one-sided non-compliance, prove CACE = ATT.

<details><summary>Show solution</summary>

**Step 1.** No one assigned to control can be treated, so $D(0) = 0$ for all: there are no always-takers or defiers.

**Step 2.** The treated ($D = 1$) are exactly those with $Z = 1$ and $D(1) = 1$: compliers assigned to treatment.

**Step 3.** Randomisation makes compliers assigned to treatment representative of all compliers, so $E[\tau \mid D = 1] = E[\tau \mid \text{complier}]$.

**Answer:** ATT = CACE $= \text{ITT}_Y/\text{ITT}_D$. $\blacksquare$

</details>

### Exercise 8 · Per-protocol is biased

In Exercise 4's population, show that the per-protocol comparison (offered who trained versus not-offered who did not train) compares different mixtures of types.

<details><summary>Show solution</summary>

**Step 1.** Offered who trained: compliers and always-takers, in proportions $0.5 : 0.1$.

### Exercise 9 · As-treated is an observational study

Show that the as-treated difference $E[Y \mid D = 1] - E[Y \mid D = 0]$ contains a selection-bias term even when $Z$ is randomised.

<details><summary>Show solution</summary>

**Step 1.** $D$ is chosen by people (types), not assigned: the treated are compliers-with-offer plus always-takers; the untreated are compliers-without-offer plus never-takers.

**Step 2.** Apply Day 34's decomposition with $D$ as the treatment: naive $=$ ATT $+ \{E[Y(0) \mid D = 1] - E[Y(0) \mid D = 0]\}$.

**Step 3.** The bracket compares $Y(0)$ across different type mixtures, so it is generally not zero.

**Answer:** as-treated analysis throws away the randomisation and reintroduces selection bias.

</details>

### Exercise 10 · When exclusion fails

Suppose the offer letter itself raises employment by $\delta$ for everyone (e.g. it signals that the government is watching). Show the Wald ratio equals CACE $+ \delta/\pi_c$.

<details><summary>Show solution</summary>

**Step 1.** Each unit's outcome difference becomes (its training effect if a complier, else 0) $+ \delta$.

**Step 2.** $\text{ITT}_Y = \pi_c\,\text{CACE} + \delta$, while $\text{ITT}_D = \pi_c$ is unchanged.

**Step 3.** Ratio: $\text{CACE} + \delta/\pi_c$. $\blacksquare$

A small direct effect is magnified when compliance is low: exclusion matters most for weak instruments.

</details>

### Exercise 13 · Adherence in a drug trial

In a heart-disease trial, patients who took at least 80% of their *placebo* pills had 15% mortality versus 28% for poor adherers. What does this imply for per-protocol analyses of the active drug?

<details><summary>Show solution</summary>

**Step 1.** A placebo has no effect, so the 13-point gap is pure selection: good adherers are healthier or more careful.

**Step 2.** Comparing adherent patients on the drug with all placebo patients (or with non-adherent ones) would credit the drug with this gap.

**Answer:** per-protocol and as-treated analyses can show large "effects" of pills that do nothing; analyse by assignment (ITT) and, if needed, estimate the CACE with assignment as the instrument. (This pattern was famously documented in the Coronary Drug Project.)

## Verify in Python

*Your work:*

```python
import numpy as np

rng = np.random.default_rng(41)
n = 400_000
types = rng.choice(["complier", "always", "never"], n, p=[0.5, 0.1, 0.4])
Z = rng.random(n) < 0.5
D = np.where(types == "always", 1, np.where(types == "never", 0, Z.astype(int)))
base = np.select([types == "always", types == "never"], [0.55, 0.30], 0.40)   # different Y(0) by type
effect = np.where(types == "complier", 0.10, 0.05)
Y = rng.random(n) < base + effect * D
itt_y = Y[Z].mean() - Y[~Z].mean(); itt_d = D[Z].mean() - D[~Z].mean()
print(f"ITT_Y {itt_y:.3f}, ITT_D {itt_d:.3f}, CACE {itt_y / itt_d:.3f} (truth 0.10)")
print(f"as-treated {Y[D == 1].mean() - Y[D == 0].mean():.3f}")
print(f"per-protocol {Y[Z & (D == 1)].mean() - Y[~Z & (D == 0)].mean():.3f}")
print("flu-shot CACE:", round((0.08 - 0.09) / (0.30 - 0.20), 3))
```

## Key takeaways

- Randomising an offer identifies intention-to-treat effects, often the policy-relevant quantity.
- Under exclusion and monotonicity, the effect for compliers is ITT_Y / ITT_D; with one-sided non-compliance it equals the ATT.
- As-treated and per-protocol comparisons discard the randomisation and reintroduce selection bias.
