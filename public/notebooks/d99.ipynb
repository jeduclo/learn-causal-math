# Day 99: Many Estimates at Once: Multiple Testing and Shrinkage

*Beginner to Ninja: Math for Causal Inference*

Modern causal studies produce many estimates: dozens of outcomes, subgroups, hospitals, teachers or policies. Testing each at 5% guarantees false discoveries, and the largest estimates are systematically too large. Family-wise error control (Bonferroni, Holm), false discovery rate control (Benjamini–Hochberg) and empirical Bayes shrinkage are the tools that keep many-estimate studies honest.

**Learning outcomes**

- Define the family-wise error rate and the false discovery rate and apply Bonferroni, Holm and Benjamini–Hochberg.
- Prove that Bonferroni and Holm control the FWER and that BH controls the FDR under independence.
- Derive the normal–normal empirical Bayes posterior mean and its risk.
- Correct the winner's curse when ranking hospitals, schools or programmes.

## Theory

### Error rates

Testing $m$ hypotheses, $m_0$ of them true nulls, let $V$ be the number of false rejections and $R$ the total number of rejections.

- **FWER** $= P(V \ge 1)$: the chance of any false discovery.
- **FDR** $= E[V/\max(R, 1)]$: the expected share of discoveries that are false.

With 1,000 tests at 5% and no true effects, we expect 50 false positives.

### Procedures

Order the p-values $p_{(1)} \le \dots \le p_{(m)}$.

- **Bonferroni:** reject if $p_i \le \alpha/m$. Controls FWER under any dependence.
- **Holm:** step down; reject $p_{(1)}, p_{(2)}, \dots$ while $p_{(k)} \le \alpha/(m - k + 1)$; stop at the first failure. Controls FWER and is uniformly more powerful than Bonferroni.
- **Benjamini–Hochberg (BH):** find the largest $k$ with $p_{(k)} \le qk/m$ and reject $p_{(1)}, \dots, p_{(k)}$. Controls FDR at $qm_0/m \le q$ under independence (and positive dependence).

Use FWER when a single false claim is costly (a confirmatory trial's primary outcomes); FDR for screening many candidates. Pre-specified indices of related outcomes reduce the number of tests in the first place.

### Shrinkage and the winner's curse

Suppose estimates $\hat\theta_j \sim N(\theta_j, s_j^2)$ and true effects vary as $\theta_j \sim N(\mu, \tau^2)$. The posterior mean is

$$
E[\theta_j \mid \hat\theta_j] = \mu + (1 - B_j)(\hat\theta_j - \mu), \qquad B_j = \frac{s_j^2}{s_j^2 + \tau^2}.
$$

Noisy estimates (large $s_j$) are shrunk most. **Empirical Bayes** estimates $\mu$ and $\tau^2$ from all the $\hat\theta_j$ (e.g. by maximum likelihood). The largest raw estimate is usually large partly because of noise (the **winner's curse**); shrinkage corrects this and lowers the total mean squared error (Stein's paradox).

### Exercise 6 · Bonferroni controls the FWER

Show $P(V \ge 1) \le \alpha$ for any dependence among the tests.

<details><summary>Show solution</summary>

**Step 1.** $\{V \ge 1\} = \bigcup_{i \in H_0}\{p_i \le \alpha/m\}$.

**Step 2.** Union bound: $P \le \sum_{i \in H_0}P(p_i \le \alpha/m) \le m_0\alpha/m \le \alpha$. $\blacksquare$

</details>

*Your work:*

### Exercise 7 · Holm controls the FWER

Show Holm's procedure has FWER $\le \alpha$.

<details><summary>Show solution</summary>

**Step 1.** Let $p^* = \min_{i \in H_0}p_i$. Suppose Holm makes a false rejection; consider the first true null rejected, at step $k$.

**Step 2.** Before it, at most $m - m_0$ hypotheses (all false nulls) were rejected, so $k \le m - m_0 + 1$ and the step-$k$ threshold is $\alpha/(m - k + 1) \le \alpha/m_0$.

**Step 3.** Hence a false rejection implies $p^* \le \alpha/m_0$, and $P(p^* \le \alpha/m_0) \le m_0\cdot\alpha/m_0 = \alpha$ by the union bound. $\blacksquare$

</details>

### Exercise 8 · BH controls the FDR

Assume independent p-values with true-null p-values uniform. Let $R^{(i)}$ be the number of BH rejections when $p_i$ is replaced by 0. Show $\text{FDR} = qm_0/m$.

<details><summary>Show solution</summary>

**Step 1.** BH rejects $i$ iff $p_i \le qR/m$, so $\text{FDR} = \sum_{i \in H_0}\sum_{k = 1}^m\frac1kP(p_i \le qk/m, R = k)$.

**Step 2.** On $\{p_i \le qk/m\}$, $R = k$ iff $R^{(i)} = k$ (lowering an already-rejected p-value to 0 does not change the cut-off).

**Step 3.** $R^{(i)}$ depends only on the other p-values, independent of $p_i$: $P(p_i \le qk/m, R^{(i)} = k) = \frac{qk}m P(R^{(i)} = k)$.

**Step 4.** So each null contributes $\sum_k\frac qmP(R^{(i)} = k) = q/m$; summing over $m_0$ nulls gives $qm_0/m$. $\blacksquare$

</details>

### Exercise 9 · The posterior mean

With $\hat\theta \mid \theta \sim N(\theta, s^2)$ and $\theta \sim N(\mu, \tau^2)$, derive $E[\theta \mid \hat\theta]$.

<details><summary>Show solution</summary>

**Step 1.** $(\theta, \hat\theta)$ is jointly normal with $\mathrm{Cov}(\theta, \hat\theta) = \tau^2$, $\mathrm{Var}(\hat\theta) = \tau^2 + s^2$.

**Step 2.** $E[\theta \mid \hat\theta] = \mu + \frac{\tau^2}{\tau^2 + s^2}(\hat\theta - \mu) = \mu + (1 - B)(\hat\theta - \mu)$. $\blacksquare$

</details>

*Your work:*

### Exercise 10 · Shrinkage lowers the risk

Show $E[(\theta - E[\theta \mid \hat\theta])^2] = \tau^2s^2/(\tau^2 + s^2) < s^2 = E[(\theta - \hat\theta)^2]$.

<details><summary>Show solution</summary>

**Step 1.** The posterior variance of jointly normal variables is $\mathrm{Var}(\theta) - \mathrm{Cov}^2/\mathrm{Var}(\hat\theta) = \tau^2 - \tau^4/(\tau^2 + s^2) = \tau^2s^2/(\tau^2 + s^2)$.

**Step 2.** This equals $s^2(1 - B) < s^2$. $\blacksquare$ The gain is largest for the noisiest units.

</details>

## Verify in Python

*Your work:*

```python
import numpy as np
from scipy.stats import norm

rng = np.random.default_rng(99)
m, m1 = 1000, 100; mu = np.r_[np.full(m1, 3.0), np.zeros(m - m1)]
def procedures(p, a=.05):
    o = np.argsort(p); ps = p[o]; k = np.arange(1, m + 1)
    holm = np.zeros(m, bool); fail = ps > a / (m - k + 1)
    holm[o[:np.argmax(fail) if fail.any() else m]] = True
    bh = np.zeros(m, bool); ok = np.nonzero(ps <= a * k / m)[0]
    if len(ok): bh[o[:ok.max() + 1]] = True
    return {"unadjusted": p < a, "Bonferroni": p < a / m, "Holm": holm, "BH": bh}
res = {}
for _ in range(200):
    z = mu + rng.normal(size=m)
    for name, r in procedures(2 * norm.sf(abs(z))).items():
        V, S = r[m1:].sum(), r[:m1].sum(); res.setdefault(name, []).append((S, V, V / max(V + S, 1), V > 0))
for name, v in res.items():
    v = np.array(v, float)
    print(f"{name:10s} true {v[:, 0].mean():5.1f}  false {v[:, 1].mean():5.1f}  FDR {v[:, 2].mean():.3f}  FWER {v[:, 3].mean():.2f}")

p5 = np.array([.001, .008, .012, .04, .30]); k5 = np.arange(1, 6)
print("Ex 2-3: Bonferroni", (p5 <= .01).sum(), " Holm thresholds", (.05 / (6 - k5)).round(4), " BH thresholds", (.05 * k5 / 5).round(3))

H = 60; theta = rng.normal(10, 2, H); se = rng.uniform(1, 4, H); y = theta + se * rng.normal(size=H)
t2 = 1.0
for _ in range(200):                                   # maximum-likelihood fixed point for tau^2
    w = 1 / (se ** 2 + t2); mu_ = np.sum(w * y) / w.sum()
    t2 = max(np.sum(w ** 2 * ((y - mu_) ** 2 - se ** 2)) / np.sum(w ** 2), 0)
B = se ** 2 / (se ** 2 + t2); post = mu_ + (1 - B) * (y - mu_); top = np.argmax(y)
print(f"tau2 {t2:.2f} (true variance {theta.var():.2f})  MSE raw {np.mean((y - theta) ** 2):.2f}  MSE EB {np.mean((post - theta) ** 2):.2f}")
print(f"worst raw hospital: raw {y[top]:.2f} (SE {se[top]:.2f}), EB {post[top]:.2f} (B {B[top]:.2f}), true {theta[top]:.2f}")
```

## Key takeaways

- Many tests at 5% guarantee false discoveries: control the FWER (Bonferroni, Holm) for confirmatory claims and the FDR (BH) for screening.
- Pre-specified indices reduce the number of tests in the first place.
- Empirical Bayes shrinkage corrects the winner's curse and lowers mean squared error, especially for noisy units.
