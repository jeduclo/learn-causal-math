Single-world intervention graphs split each treated node into the value it would have taken and the value we set, so potential outcomes appear in the graph. d-separation in the split graph reads off Phase 3's ignorability assumptions directly, unifying the two languages of causal inference.

**Learning outcomes**

- Read conditional ignorability, and its failure, from d-separation in a SWIG.
- Derive the adjustment formula and the g-formula from SWIG independences plus consistency.
- Carry a policy question through the whole Phase 4 toolkit (DAG, identification, adjustment set, bad controls, selection and mediation).

## Theory

### Two languages, one model

Phase 3 wrote assumptions with potential outcomes ($Y(a) \perp\!\!\!\perp A \mid L$). Phase 4 wrote them with graphs (back-door paths). **Single-world intervention graphs** (Richardson and Robins, 2013) put the potential outcomes into the graph, so ignorability can be checked by d-separation.

### Node splitting

To intervene on $A$ and set it to $a$:

1. Split $A$ into two halves: the **random half** $A$ (the value it would naturally take), which keeps all incoming arrows, and the **fixed half** $a$, which keeps all outgoing arrows.
2. Relabel every descendant $V$ of $a$ as the potential outcome $V(a)$.

Example: $L \to A \to Y$, $L \to Y$ becomes

$$
L \to A, \qquad a \to Y(a), \qquad L \to Y(a).
$$

There is no edge between $A$ and $Y(a)$: the natural treatment and the potential outcome are linked only through $L$.

### Reading ignorability

Apply d-separation in the SWIG (fixed nodes are constants and can be treated as conditioned on):

$$
\big(Y(a) \perp\!\!\!\perp A \mid L\big)_{\text{SWIG}} \;\Longrightarrow\; Y(a) \perp\!\!\!\perp A \mid L.
$$

With consistency ($A = a \Rightarrow Y(a) = Y$) and positivity, this gives the adjustment formula

$$
E[Y(a)] = \sum_l E[Y \mid A = a, L = l]\,P(l).
$$

The back-door criterion holds for $L$ exactly when $L$ d-separates $A$ from $Y(a)$ in the SWIG.

### Several interventions

For sequential treatments $A_1 \to L \to A_2 \to Y$, split both: the SWIG contains $L(a_1)$, $A_2(a_1)$ and $Y(a_1, a_2)$. **Sequential ignorability**, $Y(a_1, a_2) \perp\!\!\!\perp A_1$ and $Y(a_1, a_2) \perp\!\!\!\perp A_2(a_1) \mid L(a_1), A_1$, can be read off, and it yields the **g-formula** of Day 55.

### Single world, by design

A SWIG represents one intervention at a time. It cannot express **cross-world** statements like $Y(1, m) \perp\!\!\!\perp M(0)$, which natural mediation effects require (Day 58). That is a feature: assumptions that no experiment could ever check stay visibly separate.

### Exercise 5 · What a SWIG cannot show

Why can't a SWIG display the assumption $Y(1, m) \perp\!\!\!\perp M(0)$ used for natural direct effects?

<details><summary>Show solution</summary>

**Step 1.** A SWIG contains variables from a single intervention: either $M(0)$ (intervening on $A$ with $a = 0$) or $Y(1, m)$ (intervening on $A$ and $M$ with $a = 1$), not both.

**Step 2.** The assumption relates two different interventions on the same unit.

### Exercise 6 · From ignorability to the adjustment formula

Assume $Y(a) \perp\!\!\!\perp A \mid L$, consistency and positivity. Prove $E[Y(a)] = \sum_l E[Y \mid A = a, L = l]P(l)$.

<details><summary>Show solution</summary>

**Step 1.** LIE: $E[Y(a)] = \sum_l E[Y(a) \mid L = l]P(l)$.

**Step 2.** Ignorability: $E[Y(a) \mid L = l] = E[Y(a) \mid A = a, L = l]$ (positivity makes the conditioning event possible).

**Step 3.** Consistency: on $\{A = a\}$, $Y(a) = Y$, so this is $E[Y \mid A = a, L = l]$.

**Step 4.** Substitute. $\blacksquare$

</details>

### Exercise 7 · Back-door implies SWIG separation

Suppose $L$ satisfies the back-door criterion for $(A, Y)$ in $G$. Argue that $L$ d-separates $A$ from $Y(a)$ in the SWIG.

<details><summary>Show solution</summary>

**Step 1.** In the SWIG, the random half $A$ has only incoming edges, so every path from $A$ to $Y(a)$ starts with an edge into $A$: these are exactly the back-door paths of $G$ (with descendants of $a$ relabelled).

**Step 2.** Directed paths from $A$ in $G$ now start at the fixed node $a$, not at $A$, so they are not $A$–$Y(a)$ paths.

**Step 3.** $L$ contains no descendant of $A$, so in the SWIG it contains no potential outcome; conditioning on it opens or blocks the same nodes as in $G$.

**Step 4.** $L$ blocks every back-door path in $G$, hence every $A$–$Y(a)$ path in the SWIG. $\blacksquare$

</details>

### Exercise 8 · The g-formula from sequential ignorability

For the SWIG of Exercise 4 (with no $U$–$A_2$ edge), assume $Y(a_1, a_2) \perp\!\!\!\perp A_1$, $Y(a_1, a_2) \perp\!\!\!\perp A_2(a_1) \mid A_1, L(a_1)$, $L(a_1) \perp\!\!\!\perp A_1$, and consistency. Prove $E[Y(a_1, a_2)] = \sum_l E[Y \mid a_1, l, a_2]P(l \mid a_1)$.

**Step 3.** Consistency: when $A_1 = a_1$ and $A_2 = a_2$, $Y(a_1, a_2) = Y$.

**Step 4.** Substitute. $\blacksquare$

### Exercise 9 · M-bias in a SWIG

In $U_1 \to A$, $U_1 \to M$, $U_2 \to M$, $U_2 \to Y$, $A \to Y$, show with the SWIG that $Y(a) \perp\!\!\!\perp A$ but not $Y(a) \perp\!\!\!\perp A \mid M$.

<details><summary>Show solution</summary>

**Step 1.** SWIG: $U_1 \to A$, $U_1 \to M$, $U_2 \to M$, $U_2 \to Y(a)$, $a \to Y(a)$. $M$ is not a descendant of $a$, so it is not relabelled.

**Step 2.** The only $A$–$Y(a)$ path: $A \leftarrow U_1 \to M \leftarrow U_2 \to Y(a)$, with a collider at $M$.

**Step 3.** Unconditioned: blocked, so $Y(a) \perp\!\!\!\perp A$. Given $M$: open.

**Answer:** treatment is marginally ignorable but not given $M$: the potential-outcome version of M-bias. $\blacksquare$

</details>

### Exercise 10 · Why the front door is not "ignorability"

In the front-door graph $U \to A$, $U \to Y$, $A \to M \to Y$, show that no observed set makes $Y(a) \perp\!\!\!\perp A$, and explain how identification still succeeds.

<details><summary>Show solution</summary>

**Step 1.** SWIG: $U \to A$, $U \to Y(a)$, $a \to M(a) \to Y(a)$.

**Step 2.** The path $A \leftarrow U \to Y(a)$ is open, and the only observed candidate, $M(a)$, is not on it.

**Step 3.** So no conditioning set gives ignorability of $A$.

**Step 4.** Identification works through two separate ignorabilities: $M(a) \perp\!\!\!\perp A$ (in this SWIG) and $Y(m) \perp\!\!\!\perp M \mid A$ (in the SWIG for intervening on $M$), chained as in Day 54. $\blacksquare$

</details>

### Exercise 15 · Reading a SWIG for a vaccine study

Observational vaccine data: age $G \to$ vaccination $A$, $G \to$ infection $Y$; health-seeking behaviour $H$ (unobserved) $\to A$ and $H \to$ testing $T$; $Y \to T$; only tested people are analysed. Draw the SWIG for $A$ and decide whether $Y(a) \perp\!\!\!\perp A \mid G$, and what the testing restriction does.

<details><summary>Show solution</summary>

**Step 1.** SWIG: $G \to A$, $H \to A$, $G \to Y(a)$, $a \to Y(a)$, $H \to T(a)$, $Y(a) \to T(a)$ (testing is a descendant of $a$ through $Y$).

**Step 2.** Paths from $A$ to $Y(a)$: $A \leftarrow G \to Y(a)$ (blocked by $G$) and $A \leftarrow H \to T(a) \leftarrow Y(a)$ (collider at $T(a)$, blocked if not conditioned).

**Step 3.** So $Y(a) \perp\!\!\!\perp A \mid G$ in the full population.

**Step 4.** Restricting to tested people conditions on $T$, opening $A \leftarrow H \to T \leftarrow Y$.

**Answer:** age adjustment suffices in population data, but tested-only data are biased by health-seeking behaviour (the motivation for test-negative designs, which make further assumptions).

</details>

## Verify in Python

```python
import numpy as np
import networkx as nx

dsep = getattr(nx, "is_d_separator", None) or nx.d_separated   # name changed in networkx 3.3

def swig(G, A):
    """Split node A into random half A and fixed half 'a'; relabel descendants of A as V(a)."""
    desc = nx.descendants(G, A)
    name = lambda v: f"{v}(a)" if v in desc else v
    S = nx.DiGraph(); S.add_nodes_from([name(v) for v in G] + ["a"])
    for u, v in G.edges:
        S.add_edge("a" if u == A else name(u), name(v))
    return S

G = nx.DiGraph([("L", "A"), ("L", "Y"), ("A", "Y"), ("A", "M"), ("M", "Y")])
S = swig(G, "A")
print("SWIG edges:", sorted(S.edges))
print("Y(a) _||_ A:", dsep(S, {"Y(a)"}, {"A"}, {"a"}), "| given L:", dsep(S, {"Y(a)"}, {"A"}, {"a", "L"}))

Mb = swig(nx.DiGraph([("U1", "A"), ("U1", "M"), ("U2", "M"), ("U2", "Y"), ("A", "Y")]), "A")
print("M-bias: Y(a) _||_ A:", dsep(Mb, {"Y(a)"}, {"A"}, {"a"}), "| given M:", dsep(Mb, {"Y(a)"}, {"A"}, {"a", "M"}))

V = swig(nx.DiGraph([("G", "A"), ("H", "A"), ("G", "Y"), ("A", "Y"), ("H", "T"), ("Y", "T")]), "A")
print("vaccine: Y(a) _||_ A | G:", dsep(V, {"Y(a)"}, {"A"}, {"a", "G"}), "| G, T(a):", dsep(V, {"Y(a)"}, {"A"}, {"a", "G", "T(a)"}))

# Exercise 13: sequential treatment and the g-formula
rng = np.random.default_rng(60); n = 2_000_000
def clinic(a1=None, a2=None):
    U = rng.random(n) < .5
    A1 = rng.random(n) < .5 if a1 is None else np.full(n, a1, bool)
    L = rng.random(n) < .2 + .4 * A1 + .3 * U                       # high CD4
    A2 = rng.random(n) < (.8 - .6 * L) if a2 is None else np.full(n, a2, bool)
    Y = rng.random(n) < .1 + .2 * A1 + .25 * A2 + .3 * U             # viral suppression
    return A1, L, A2, Y
A1, L, A2, Y = clinic()
truth = clinic(1, 1)[3].mean() - clinic(0, 0)[3].mean()
g = lambda a1, a2: sum(Y[(A1 == a1) & (L == l) & (A2 == a2)].mean() * (L[A1 == a1] == l).mean() for l in (0, 1))
b = np.linalg.lstsq(np.column_stack([np.ones(n), A1, L, A2]).astype(float), Y.astype(float), rcond=None)[0]
print(f"truth {truth:.3f}  naive {Y[A1 & A2].mean() - Y[~A1 & ~A2].mean():.3f}  regression {b[1] + b[3]:.3f}  g-formula {g(1, 1) - g(0, 0):.3f}")
```

## Key takeaways

- A SWIG splits each intervened node into its natural and fixed halves and relabels descendants as potential outcomes.
- Ignorability is d-separation in the SWIG; with consistency and positivity it yields the adjustment formula and, for sequential treatments, the g-formula.
- SWIGs keep single-world assumptions testable in principle and cross-world assumptions visibly separate. Phase 5 now turns to designs (instruments, differences-in-differences, discontinuities) for when no admissible set is observed.
