## Theory

### Two decompositions

With $Y = Y(0) + T\tau$ (Day 31) and $p = P(T = 1)$:

$$
\underbrace{E[Y \mid T = 1] - E[Y \mid T = 0]}_{\text{naive}} = \text{ATT} + \underbrace{E[Y(0) \mid T = 1] - E[Y(0) \mid T = 0]}_{\text{selection bias}}
$$

$$
\phantom{E[Y \mid T = 1] - E[Y \mid T = 0]} = \text{ATE} + \text{selection bias} + \underbrace{(1 - p)(\text{ATT} - \text{ATC})}_{\text{heterogeneity bias}}.
$$

- **Selection bias** (selection on levels): the treated would have had different outcomes even untreated.
- **Heterogeneity bias** (selection on gains): the treated benefit more or less than the untreated.

Randomisation sets both to zero (Day 35).

### The Roy model

Roy (1951) asked how workers choose occupations. Applied to programmes: unit $i$ chooses treatment if its gain exceeds its cost,

$$
T_i = \mathbf 1\{Y_i(1) - Y_i(0) > c_i\}.
$$

Choice now depends on potential outcomes, the opposite of an experiment. Two typical patterns:

- **Positive selection on gains:** those who gain most enrol, so ATT > ATE > ATC.
- **Negative selection on levels:** programmes target, or attract, people with poor prospects: $E[Y(0) \mid T = 1] < E[Y(0) \mid T = 0]$, so naive comparisons understate effects (training for the unemployed, hospital care for the sick).

If gains are normal, $\tau \sim N(\mu, \sigma^2)$, and $T = \mathbf 1\{\tau > 0\}$, then

$$
\text{ATT} = \mu + \sigma\frac{\phi(\mu/\sigma)}{\Phi(\mu/\sigma)}.
$$

### Ashenfelter's dip

Trainees' earnings often *fall* just before they enrol (they enrol because they lost a job). A before–after comparison then credits the programme with the natural recovery (Ashenfelter, 1978).

### Exercise 6 · The two-term decomposition

Prove naive $=$ ATT $+$ selection bias.

<details><summary>Show solution</summary>

**Step 1.** Observed outcomes: $E[Y \mid T = 1] = E[Y(1) \mid T = 1]$ and $E[Y \mid T = 0] = E[Y(0) \mid T = 0]$.

**Step 2.** Add and subtract $E[Y(0) \mid T = 1]$:

$$
\big(E[Y(1) \mid T = 1] - E[Y(0) \mid T = 1]\big) + \big(E[Y(0) \mid T = 1] - E[Y(0) \mid T = 0]\big).
$$

**Step 3.** The first bracket is the ATT; the second is the selection bias. $\blacksquare$

### Exercise 8 · Randomisation kills both biases

Prove that if $T$ is independent of $(Y(0), Y(1))$, selection bias and heterogeneity bias are both zero.

<details><summary>Show solution</summary>

**Step 1.** Independence of $T$ and $Y(0)$: $E[Y(0) \mid T = 1] = E[Y(0) \mid T = 0]$, so selection bias $= 0$.

### Exercise 9 · Selection on gains raises the ATT

In the Roy model with $T = \mathbf 1\{\tau > c\}$ (and $0 < P(T = 1) < 1$), prove ATT $\ge$ ATE $\ge$ ATC.

<details><summary>Show solution</summary>

**Step 1.** Every treated unit has $\tau > c$, so ATT $= E[\tau \mid \tau > c] > c$.

**Step 2.** Every untreated unit has $\tau \le c$, so ATC $= E[\tau \mid \tau \le c] \le c$.

**Step 3.** So ATT $>$ ATC, and ATE $= p\,\text{ATT} + (1 - p)\,\text{ATC}$ lies between them. $\blacksquare$

</details>

### Exercise 10 · The Roy ATT with normal gains

If $\tau \sim N(\mu, \sigma^2)$ and $T = \mathbf 1\{\tau > 0\}$, prove ATT $= \mu + \sigma\phi(\mu/\sigma)/\Phi(\mu/\sigma)$.

<details><summary>Show solution</summary>

**Step 1.** Standardise: $\tau = \mu + \sigma Z$, and $\tau > 0 \iff Z > -\mu/\sigma$.

**Step 2.** Truncated-normal mean: $E[Z \mid Z > a] = \phi(a)/(1 - \Phi(a))$ with $a = -\mu/\sigma$.

**Step 3.** By symmetry, $\phi(-\mu/\sigma) = \phi(\mu/\sigma)$ and $1 - \Phi(-\mu/\sigma) = \Phi(\mu/\sigma)$.

**Step 4.** ATT $= \mu + \sigma E[Z \mid Z > a] = \mu + \sigma\phi(\mu/\sigma)/\Phi(\mu/\sigma)$. $\blacksquare$

</details>

**Step 4.** ATE $= \mu = 2$.

**Answer:** 75% enrol; ATT ≈ \$3,280 versus ATE \$2,000. An evaluation of current trainees overstates what compulsory training would achieve.

### Exercise 12 · Ashenfelter's dip

Trainees usually earn \$15k a year. In the year before training, after losing jobs, they earn \$12k. After training they earn \$16k. Without training, their earnings would have recovered to \$15k. Compare the before–after estimate with the true effect.

<details><summary>Show solution</summary>

**Step 1.** Before–after: $16 - 12 = 4$.

**Step 2.** True effect: $Y(1) - Y(0) = 16 - 15 = 1$.

**Step 3.** The extra 3 is the recovery from the dip (mean reversion) that would have happened anyway.

**Answer:** before–after says \$4k; the truth is \$1k. Comparing with a pre-dip baseline or with a matched control group that also lost jobs avoids this.

</details>

### Exercise 15 · Which direction?

A city opens a free clinic, and clinic users have lower blood pressure than non-users. An analyst concludes the clinic works. List the threats using the decomposition, and propose a design that avoids them.

<details><summary>Show solution</summary>

**Step 1.** Selection on levels: health-conscious people use clinics and would have lower blood pressure anyway (positive bias in the naive "effect", i.e. too good).

**Step 2.** Selection on levels the other way: people with symptoms seek care (negative bias). The net sign is unknown.

**Step 3.** Selection on gains: those who expect to benefit attend, so even an unbiased ATT would overstate the ATE of universal access.

**Step 4.** Design: randomise invitations or outreach (an encouragement design, Day 41), or exploit a clinic opening date that varies across neighbourhoods (difference-in-differences, Phase 5).

**Answer:** the naive comparison mixes the clinic's effect with who chooses to attend; randomised encouragement isolates it.

</details>

## Verify in Python

*Your work:*

```python
import numpy as np
from scipy import stats

# Roy model with normal gains (Exercise 11)
mu, sd = 2.0, 3.0
p = stats.norm.cdf(mu / sd)
att = mu + sd * stats.norm.pdf(mu / sd) / p
print(f"share enrolling {p:.3f}, ATT {att:.3f}, ATE {mu}")

rng = np.random.default_rng(34)
n = 1_000_000
y0 = rng.normal(20, 5, n)
tau = rng.normal(mu, sd, n)
T = tau > 0                                         # select on gains only
y = y0 + T * tau
naive = y[T].mean() - y[~T].mean()
sb = y0[T].mean() - y0[~T].mean()
att_s, atc_s = tau[T].mean(), tau[~T].mean()
print(f"simulated ATT {att_s:.3f}, ATC {atc_s:.3f}, ATE {tau.mean():.3f}")
print(f"naive {naive:.3f} = ATE {tau.mean():.3f} + SB {sb:.3f} + HB {(1 - T.mean()) * (att_s - atc_s):.3f}")
```

## Key takeaways

- Naive comparison = ATE + selection bias + heterogeneity bias.
- In the Roy model people select on gains, so ATT > ATE > ATC; programmes for the disadvantaged also select on levels.
- Before–after comparisons are vulnerable to Ashenfelter's dip and mean reversion.
