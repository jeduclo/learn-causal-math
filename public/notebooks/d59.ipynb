## Theory

### The ladder of causation

| Rung | Question | Notation | Needs |
|---|---|---|---|
| 1. Association | What do I expect to see if I observe $x$? | $P(y \mid x)$ | data |
| 2. Intervention | What happens if I *do* $x$? | $P(y \mid \operatorname{do}(x))$ | data + causal graph (or an experiment) |
| 3. Counterfactual | Given that $x$ and $y$ happened, what would $y$ have been under $x'$? | $P(y_{x'} \mid x, y)$ | an SCM (functions, not only a graph) |

Each rung needs more assumptions than the one below.

### Counterfactuals in an SCM

For unit $u$ (a value of the disturbances $\mathbf U$), the counterfactual $Y_x(u)$ is the value of $Y$ in the modified model $\mathcal M_x$. These are Phase 3's potential outcomes, now derived from structural equations. **Consistency**: if $X(u) = x$ then $Y_x(u) = Y(u)$.

### Three steps

To compute $P(y_{x'} \mid \text{evidence } e)$:

1. **Abduction:** update the disturbances using the evidence: $P(\mathbf u) \to P(\mathbf u \mid e)$.
2. **Action:** replace $X$'s equation by $X := x'$.
3. **Prediction:** compute $Y$ in the modified model, using $P(\mathbf u \mid e)$.

### Linear models: the same effect for everyone

If $Y := \beta X + U_Y$ (whatever the dependence between $X$ and $U_Y$), then $Y_{x'}(u) - Y_x(u) = \beta(x' - x)$ for every unit. Unit-level counterfactuals are easy; they are hard precisely when effects vary across units.

### Attribution: necessity and sufficiency

For binary $X, Y$ (write $y$ for $Y = 1$, $y'$ for $Y = 0$):

- **Probability of necessity:** $PN = P(Y_0 = 0 \mid X = 1, Y = 1)$: among exposed people who got the disease, the share who would not have without exposure ("but-for" causation in law).
- **Probability of sufficiency:** $PS = P(Y_1 = 1 \mid X = 0, Y = 0)$.
- **Probability of necessity and sufficiency:** $PNS = P(Y_1 = 1, Y_0 = 0)$: the share of units who respond.

### Identification under extra assumptions

If $X$ is **exogenous** (randomised) and the effect is **monotone** (no one is harmed: $Y_1 \ge Y_0$), then

$$
PN = \frac{P(y \mid x) - P(y \mid x')}{P(y \mid x)} = 1 - \frac{1}{RR}, \qquad PNS = P(y \mid x) - P(y \mid x').
$$

Without monotonicity, experiments give only **bounds**, for example

$$
\max\{0,\; P(y_x) - P(y_{x'})\} \le PNS \le \min\{P(y_x),\; P(y'_{x'})\}.
$$

### Exercise 6 · Consistency from the SCM

Prove that if $X(u) = x$ then $Y_x(u) = Y(u)$.

<details><summary>Show solution</summary>

**Step 1.** $\mathcal M_x$ differs from $\mathcal M$ only in $X$'s equation.

**Step 2.** For unit $u$, the original equation already produces $X = x$, so replacing it with $X := x$ changes no value.

**Step 3.** All other equations and disturbances are identical, so every variable, including $Y$, takes the same value. $\blacksquare$

</details>

### Exercise 7 · Linear counterfactuals

Let $Y := \beta X + U_Y$, with $U_Y$ arbitrarily dependent on $X$. Prove $E[Y_{x'} \mid X = x, Y = y] = y + \beta(x' - x)$.

<details><summary>Show solution</summary>

**Step 1.** Abduction: given $X = x$, $Y = y$, we know $U_Y = y - \beta x$ exactly.

**Step 2.** Action and prediction: $Y_{x'} = \beta x' + U_Y = \beta x' + y - \beta x$.

**Step 3.** This is non-random given the evidence, so its expectation is itself. $\blacksquare$ (Rung 3 collapses to $\beta$, a rung-2 quantity, in linear additive models.)

</details>

### Exercise 8 · Probability of necessity under exogeneity and monotonicity

Assume $X \perp\!\!\!\perp (Y_0, Y_1)$ and $Y_1 \ge Y_0$. Prove $PN = [P(y \mid x) - P(y \mid x')]/P(y \mid x)$.

<details><summary>Show solution</summary>

**Step 1.** $PN = \dfrac{P(Y_0 = 0, X = 1, Y = 1)}{P(X = 1, Y = 1)} = \dfrac{P(Y_0 = 0, Y_1 = 1, X = 1)}{P(Y_1 = 1, X = 1)}$ (consistency).

**Step 2.** By exogeneity, $X$ factors out: $PN = P(Y_0 = 0, Y_1 = 1)/P(Y_1 = 1)$.

**Step 3.** Monotonicity: $Y_1 = 1$ whenever $Y_0 = 1$, so $P(Y_0 = 0, Y_1 = 1) = P(Y_1 = 1) - P(Y_0 = 1)$.

**Step 4.** By exogeneity $P(Y_x = 1) = P(y \mid x)$. Substitute. $\blacksquare$

</details>

### Exercise 9 · Bounds on PNS

From experimental data alone, prove $\max\{0, P(y_x) - P(y_{x'})\} \le PNS \le \min\{P(y_x), P(y'_{x'})\}$.

<details><summary>Show solution</summary>

**Step 1.** Four response types: $(Y_1, Y_0) \in \{(1, 1), (1, 0), (0, 1), (0, 0)\}$ with probabilities $p_{11}, p_{10}, p_{01}, p_{00}$; $PNS = p_{10}$.

**Step 2.** Experiments give $P(y_x) = p_{11} + p_{10}$ and $P(y_{x'}) = p_{11} + p_{01}$.

**Step 3.** Upper: $p_{10} \le p_{11} + p_{10} = P(y_x)$ and $p_{10} \le p_{10} + p_{00} = 1 - P(y_{x'}) = P(y'_{x'})$.

**Step 4.** Lower: $p_{10} = P(y_x) - P(y_{x'}) + p_{01} \ge P(y_x) - P(y_{x'})$, and $p_{10} \ge 0$. $\blacksquare$

</details>

### Exercise 10 · Same experiments, different counterfactuals

Construct two SCMs for binary $X, Y$ with the same $P(y \mid \operatorname{do}(x))$ for both $x$ but different $PNS$.

<details><summary>Show solution</summary>

**Step 1.** Model A: $U \sim \text{Bernoulli}(0.5)$, $Y := U$. Then $P(Y = 1 \mid \operatorname{do}(x)) = 0.5$ for both $x$, and nobody responds: $PNS = 0$.

**Step 2.** Model B: $Y := X \oplus U$ (exclusive or). Again $P(Y = 1 \mid \operatorname{do}(x)) = 0.5$ for both $x$.

**Step 3.** In Model B, units with $U = 0$ have $Y_1 = 1, Y_0 = 0$ (helped) and units with $U = 1$ have $Y_1 = 0, Y_0 = 1$ (harmed): $PNS = 0.5$.

**Answer:** identical at rungs 1 and 2, different at rung 3. $\blacksquare$ A zero average effect can hide large offsetting individual effects.

</details>

### Exercise 11 · A worker's counterfactual wage

In Day 50's SCM, $\log W := 1.5 + 0.08S + 0.10A + U_W$, a worker has 12 years of schooling and $\log W = 2.6$. What would her log wage have been with 16 years? Does the answer depend on her (unobserved) ability?

### Exercise 12 · But-for causation in court

A worker exposed to asbestos develops mesothelioma and sues. A cohort study (exposure as good as random given the covariates) gives a lifetime risk of 2.0% for exposed and 0.5% for comparable unexposed workers. Assuming asbestos never prevents the disease, is it "more likely than not" that the exposure caused this worker's disease?

<details><summary>Show solution</summary>

**Step 1.** Exogeneity and monotonicity give $PN = (0.020 - 0.005)/0.020 = 0.75$.

**Step 2.** Equivalently, $RR = 4$ and $PN = 1 - 1/4 = 0.75$.

**Answer:** 75% > 50%: the civil standard is met. The rule "RR > 2 implies PN > 0.5" holds under these assumptions; without monotonicity only bounds are available.

### Exercise 13 · How many people does a vaccine protect?

A trial finds infection risks of 2% with vaccine and 10% with placebo. Let $y$ = "not infected". Bound the share of people who are protected (not infected with the vaccine, infected without), without assuming monotonicity. What does monotonicity add?

<details><summary>Show solution</summary>

**Step 1.** $P(y_1) = 0.98$ and $P(y_0) = 0.90$; $P(y'_0) = 0.10$.

**Step 2.** Lower bound: $\max\{0, 0.98 - 0.90\} = 0.08$.

**Step 3.** Upper bound: $\min\{0.98, 0.10\} = 0.10$.

**Step 4.** With monotonicity (the vaccine never causes infection), $PNS = 0.98 - 0.90 = 0.08$ exactly.

### Exercise 14 · Effect of training on the trained

Use Day 52's training data: half previously employed ($Z = 1$); training rates 30% ($Z = 1$) and 70% ($Z = 0$); employment $P(E = 1 \mid T, Z)$: $Z = 1$: 0.60 untrained, 0.70 trained; $Z = 0$: 0.20 untrained, 0.35 trained. Compute $E[E_0 \mid T = 1]$ and the ATT.

<details><summary>Show solution</summary>

**Step 1.** Composition of the trained: $P(Z = 1 \mid T = 1) = \dfrac{0.5(0.3)}{0.5} = 0.3$.

**Step 3.** $E[E \mid T = 1] = 0.3(0.70) + 0.7(0.35) = 0.455$.

**Step 4.** $ATT = 0.455 - 0.32 = 0.135$.

**Answer:** training raised the employment of those trained by 13.5 points, slightly more than the ATE of 12.5 points, because the trained are mostly people for whom training helps more.

### Exercise 15 · A patient's counterfactual

A patient refused a drug and died. Her family asks whether she would have survived with it. A trial shows survival of 70% with the drug and 60% without. What can be said?

<details><summary>Show solution</summary>

**Step 1.** The question is rung 3: $P(\text{survive}_1 \mid X = 0, \text{died})$, the probability of sufficiency.

**Step 2.** Patients who refuse may differ from trial participants (refusal is not random), so even population-level quantities require care.

**Step 3.** Treating her as exchangeable with trial patients and assuming monotonicity: $PS = \dfrac{P(y \mid x) - P(y \mid x')}{1 - P(y \mid x')} = \dfrac{0.70 - 0.60}{0.40} = 0.25$.

**Answer:** about a 25% chance she would have survived, under strong assumptions. Without monotonicity, only bounds are available; the honest answer is a range.

</details>

## Key takeaways

- Rung 3 conditions counterfactuals on what actually happened; computing them takes abduction, action and prediction in an SCM.
- In linear additive models every unit has the same effect; heterogeneity is what makes counterfactuals hard.
- Probabilities of necessity and sufficiency answer attribution questions; they are identified under exogeneity and monotonicity and otherwise bounded.
