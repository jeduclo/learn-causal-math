Seeing is not doing. An intervention do(X = x) replaces X's structural equation, cutting the arrows into X. The truncated factorisation turns this into a formula for the interventional distribution, which differs from ordinary conditioning whenever X has open back-door paths.

- Evaluate deterministic and stochastic policy interventions in health and economics.

## Theory

### Intervening on a model

An **intervention** $\operatorname{do}(X = x)$ replaces the structural equation $X := f_X(\operatorname{pa}(X), U_X)$ by the constant $X := x$, leaving every other equation unchanged. The modified model is $\mathcal M_x$; its graph $G_{\overline X}$ is the **mutilated graph**, with all arrows *into* $X$ removed.

The distribution of $Y$ in $\mathcal M_x$ is written $P(y \mid \operatorname{do}(x))$. It answers "what would happen to $Y$ if we *set* $X$ to $x$ for everyone?"

### Seeing versus doing

$P(y \mid x)$ describes $Y$ among units that *happen* to have $X = x$. Those units may differ in other ways (they have different parents of $X$). $P(y \mid \operatorname{do}(x))$ describes the whole population when $X$ is *set* to $x$. They coincide only when nothing that affects $Y$ also affects $X$.

### The truncated factorisation

For a Markovian model, delete the factor for $X$ and plug in $X = x$ everywhere else:

$$
P(\mathbf v \mid \operatorname{do}(x)) = \prod_{i: V_i \ne X} P\big(v_i \mid \operatorname{pa}(v_i)\big)\Big|_{X = x} \quad \text{for } \mathbf v \text{ consistent with } x.
$$

### Adjusting for the parents

Summing the truncated factorisation over everything except $Y$ and the parents $\mathbf{PA} = \operatorname{pa}(X)$ gives the **adjustment formula**

$$
P(y \mid \operatorname{do}(x)) = \sum_{\mathbf{pa}} P(y \mid x, \mathbf{pa})\,P(\mathbf{pa}).
$$

Compare with conditioning, which weights by $P(\mathbf{pa} \mid x)$ instead of $P(\mathbf{pa})$:

$$
P(y \mid x) = \sum_{\mathbf{pa}} P(y \mid x, \mathbf{pa})\,P(\mathbf{pa} \mid x).
$$

The difference is exactly confounding. (When the parents are unobserved, Days 52–56 look for other sets.)

### Causal effects

The **average causal effect** of a binary $X$ is $E[Y \mid \operatorname{do}(X = 1)] - E[Y \mid \operatorname{do}(X = 0)]$. It is the ATE of Phase 3: in an SCM, the potential outcome is $Y_i(x) = Y$ computed in $\mathcal M_x$ with unit $i$'s disturbances (Day 59).

### Stochastic (policy) interventions

Real policies rarely force $X$ on everyone; they shift its distribution, for example a subsidy that raises vaccination rates. Replacing $P(x \mid \operatorname{pa})$ by a new $q(x \mid \operatorname{pa})$ gives

$$
P_q(y) = \sum_{\mathbf{pa}, x} P(y \mid x, \mathbf{pa})\,q(x \mid \mathbf{pa})\,P(\mathbf{pa}).
$$

### Exercise 7 · The parent adjustment formula

Derive $P(y \mid \operatorname{do}(x)) = \sum_{\mathbf{pa}} P(y \mid x, \mathbf{pa})P(\mathbf{pa})$ from the truncated factorisation.

<details><summary>Show solution</summary>

**Step 1.** Multiply and divide the truncated product by $P(x \mid \mathbf{pa})$: $P(\mathbf v \mid \operatorname{do}(x)) = \dfrac{P(\mathbf v)}{P(x \mid \mathbf{pa})}$ for $\mathbf v$ consistent with $x$.

**Step 2.** Write $P(\mathbf v) = P(\mathbf{pa})P(x \mid \mathbf{pa})P(\mathbf r \mid x, \mathbf{pa})$, where $\mathbf R$ is all remaining variables, including $Y$.

**Step 3.** So $P(\mathbf v \mid \operatorname{do}(x)) = P(\mathbf{pa})P(\mathbf r \mid x, \mathbf{pa})$.

**Step 4.** Sum over $\mathbf{pa}$ and over all components of $\mathbf r$ other than $y$: $P(y \mid \operatorname{do}(x)) = \sum_{\mathbf{pa}} P(\mathbf{pa})P(y \mid x, \mathbf{pa})$. $\blacksquare$

</details>

### Exercise 8 · Non-descendants are unaffected

Prove that for any set $\mathbf N$ of non-descendants of $X$, $P(\mathbf n \mid \operatorname{do}(x)) = P(\mathbf n)$.

<details><summary>Show solution</summary>

**Step 1.** Order the variables topologically with all non-descendants of $X$ first. This is possible: no non-descendant has a descendant of $X$ as an ancestor.

**Step 2.** The equations of non-descendants involve only other non-descendants and their own disturbances.

**Step 3.** The intervention changes only $X$'s equation, and $X$ is not an input to any non-descendant's equation (directly or indirectly).

**Step 4.** So the non-descendants are the same functions of the same disturbances in $\mathcal M$ and $\mathcal M_x$: their joint distribution is unchanged. $\blacksquare$

</details>

### Exercise 9 · Conditional ignorability from the SCM

In a Markovian SCM, prove that the potential outcome $Y(x)$ is independent of $X$ given $\operatorname{pa}(X)$.

<details><summary>Show solution</summary>

**Step 1.** In $\mathcal M_x$, $X$'s equation is gone, so $Y(x)$ is a function of the disturbances other than $U_X$: $Y(x) = g(\mathbf U_{-X})$.

**Step 2.** $\operatorname{pa}(X)$ is a function of the disturbances of its ancestors, also in $\mathbf U_{-X}$.

**Step 3.** $X = f_X(\operatorname{pa}(X), U_X)$. Given $\operatorname{pa}(X) = \mathbf p$, $X = f_X(\mathbf p, U_X)$ depends only on $U_X$.

**Step 4.** $U_X$ is independent of $\mathbf U_{-X}$, also after conditioning on an event defined by $\mathbf U_{-X}$ (the event $\operatorname{pa}(X) = \mathbf p$).

**Step 5.** So given $\operatorname{pa}(X)$, $X$ is independent of $Y(x)$. $\blacksquare$ This is Phase 3's unconfoundedness, derived from a graph.

</details>

Show that $P(y \mid x) - P(y \mid \operatorname{do}(x)) = \sum_{\mathbf{pa}} P(y \mid x, \mathbf{pa})\big[P(\mathbf{pa} \mid x) - P(\mathbf{pa})\big]$, and give two conditions under which it is zero.

## Key takeaways

- $\operatorname{do}(X = x)$ deletes $X$'s equation and the arrows into $X$; everything else stays the same.
- With observed parents, $P(y \mid \operatorname{do}(x)) = \sum_{\mathbf{pa}} P(y \mid x, \mathbf{pa})P(\mathbf{pa})$; conditioning instead weights by $P(\mathbf{pa} \mid x)$.
- Policies are often stochastic interventions: they shift $P(x \mid \operatorname{pa})$ rather than fix $x$.
