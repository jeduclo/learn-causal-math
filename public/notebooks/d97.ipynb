SUTVA assumes that one unit's treatment does not affect another's outcome. Vaccines, deworming, cash transfers and information campaigns violate it: benefits spill over to neighbours. With interference, 'the' treatment effect splits into direct, indirect (spillover), total and overall effects, and they are identified by designs that randomise the share of treated units across clusters.

**Learning outcomes**

- Write potential outcomes under interference and state the partial-interference assumption.
- Define direct, indirect, total and overall effects at a treatment saturation, and prove their decomposition.
- Analyse a two-stage randomised saturation design and see why naive comparisons mislead.
- Use exposure mappings and Horvitz–Thompson estimators for network interference.

## Theory

### Potential outcomes under interference

With interference, unit $i$'s outcome depends on the whole assignment vector: $Y_i(\mathbf d)$. This is unmanageable without structure.

- **Partial interference:** units are grouped (villages, schools, households); spillovers occur within but not between groups.
- **Exposure mapping:** outcomes depend on $\mathbf d$ only through $(d_i, e_i)$, where $e_i$ summarises neighbours' treatment, e.g. the number or share of treated neighbours.

### Two-stage randomisation

Hudgens and Halloran (2008): first randomise each cluster to a **saturation** $\alpha$ (share to be treated), then randomise individuals within the cluster. Define, at saturation $\alpha$:

| Effect | Definition | Question |
|---|---|---|
| Direct $DE(\alpha)$ | $E[Y(1) \mid \alpha] - E[Y(0) \mid \alpha]$ | Effect of own treatment, holding the village saturation fixed |
| Indirect $IE(\alpha)$ | $E[Y(0) \mid \alpha] - E[Y(0) \mid 0]$ | Spillover on untreated units |
| Total $TE(\alpha)$ | $E[Y(1) \mid \alpha] - E[Y(0) \mid 0]$ | Treated in a treated village versus a pure control |
| Overall $OE(\alpha)$ | $E[\bar Y \mid \alpha] - E[\bar Y \mid 0]$ | Policy effect of treating share $\alpha$ |

Pure control clusters ($\alpha = 0$) are essential for indirect effects. Inference is at the cluster level.

### What goes wrong without the design

A standard individually randomised trial compares treated and control units that both enjoy spillovers, so it estimates the direct effect at the trial's saturation, which understates the policy effect when spillovers are positive; pooling across saturations mixes different estimands.

### Network interference

With a network, define exposures (e.g. "treated, at least one treated friend"). Under randomisation, each unit's probability $\pi_i(e)$ of receiving exposure $e$ is computable, and the **Horvitz–Thompson** estimator

$$
\hat\mu(e) = \frac1n\sum_i\frac{\mathbf 1\{E_i = e\}Y_i}{\pi_i(e)}
$$

is unbiased for the average potential outcome under exposure $e$ (Aronow and Samii, 2017).

Show $E\big[\mathbf 1\{E_i = e\}Y_i/\pi_i(e)\big] = Y_i(e)$ when $Y_i$ depends on $\mathbf d$ only through $E_i$ and $\pi_i(e) > 0$.

## Verify in Python

*Your work:*

```python
import numpy as np

rng = np.random.default_rng(97)
G, m, tau, gam = 600, 8, 1.0, 2.0                       # villages, households per village, direct, spillover
sat = rng.choice([0.0, .25, .75], size=G)               # stage 1: saturation per village
D = np.concatenate([rng.permutation(np.r_[np.ones(int(a * m)), np.zeros(m - int(a * m))]) for a in sat])
vid = np.repeat(np.arange(G), m); s = np.repeat(sat, m)
k = np.bincount(vid, weights=D)[vid]                    # treated in own village
Y = 5 + tau * D + gam * (k - D) / (m - 1) + np.repeat(.5 * rng.normal(size=G), m) + rng.normal(size=G * m)

print("pooled treated vs untreated:", round(float(Y[D == 1].mean() - Y[D == 0].mean()), 3))
y0 = Y[s == 0].mean()
for a in (.25, .75):
    S, kk = s == a, a * m
    est = [Y[S & (D == 1)].mean() - Y[S & (D == 0)].mean(), Y[S & (D == 0)].mean() - y0,
           Y[S & (D == 1)].mean() - y0, Y[S].mean() - y0]
    tru = [tau - gam / (m - 1), gam * kk / (m - 1), tau + gam * (kk - 1) / (m - 1),
           a * (tau + gam * (kk - 1) / (m - 1)) + (1 - a) * gam * kk / (m - 1)]
    print(f"alpha={a}: DE, IE, TE, OE estimated {np.round(est, 2)}  truth {np.round(tru, 2)}")
print("Ex 3 OE:", round(.75 * 7.43 + .25 * 6.71 - 5, 2))
```

## Key takeaways

- Interference breaks SUTVA; with partial interference, effects are defined at a treatment saturation.
- Direct + indirect = total; the overall effect is the policy-relevant comparison of villages.
- Two-stage randomised saturation designs with pure controls identify all four; analyse at the cluster level, and use exposure mappings for networks.
