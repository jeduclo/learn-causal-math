## Theory

### Random vectors

For a random vector $X = (X_1, \dots, X_k)^{\top}$, the **mean vector** is $\mu = E[X]$ and the **variance–covariance matrix** is

$$
\Sigma = \operatorname{Var}(X) = E\big[(X - \mu)(X - \mu)^{\top}\big], \qquad \Sigma_{jj} = \operatorname{Var}(X_j), \quad \Sigma_{j\ell} = \operatorname{Cov}(X_j, X_\ell).
$$

$\Sigma$ is symmetric. Its correlation matrix is $D^{-1/2}\Sigma D^{-1/2}$ with $D = \operatorname{diag}(\Sigma)$.

### The two rules you will use every week

$$
\operatorname{Var}(a^{\top}X) = a^{\top}\Sigma a, \qquad \operatorname{Var}(AX + b) = A\Sigma A^{\top}.
$$

They give the variance of a portfolio, a contrast between treatment arms, a cumulative event-study effect, or a predicted value $x_0^{\top}\hat\beta$: just plug in $\operatorname{Var}(\hat\beta)$.

### Positive (semi-)definiteness

Because $a^{\top}\Sigma a$ is a variance, it can never be negative: **every covariance matrix is PSD**. It is PD unless some combination of the variables is constant (e.g. shares that sum to one). A symmetric matrix that is not PSD (e.g. a "correlation matrix" assembled pairwise from different samples) cannot be the covariance matrix of anything.

### Cholesky factor and simulation

A PD matrix factors as $\Sigma = LL^{\top}$ with $L$ lower-triangular (the **Cholesky** factor). If $Z \sim N(0, I)$, then

$$
X = \mu + LZ \sim N(\mu, \Sigma).
$$

This is how simulations generate correlated covariates, errors or potential outcomes.

### Sample covariance

With the centred data matrix $\tilde X$ ($n \times k$), $S = \frac{1}{n - 1}\tilde X^{\top}\tilde X$, which is PSD by Day 17.

Find the Cholesky factor of $\Sigma$.

### Exercise 6 · Variance of a linear combination

Prove $\operatorname{Var}(a^{\top}X) = a^{\top}\Sigma a$.

<details><summary>Show solution</summary>

$a^{\top}X - E[a^{\top}X] = a^{\top}(X - \mu)$, a scalar equal to its own transpose. So

$$
\operatorname{Var}(a^{\top}X) = E\big[a^{\top}(X - \mu)(X - \mu)^{\top}a\big] = a^{\top}E\big[(X - \mu)(X - \mu)^{\top}\big]a = a^{\top}\Sigma a.
$$

</details>

### Exercise 11 · Portfolio risk

Two assets have return sds 10% and 20% and correlation 0.3. With weights $w = (0.6, 0.4)$, find the portfolio sd.

<details><summary>Show solution</summary>

**Step 1.** Covariance: $0.3 \times 10 \times 20 = 60$, so $\Sigma = \begin{pmatrix}100 & 60\\60 & 400\end{pmatrix}$.

**Step 2.** $w^{\top}\Sigma w = 0.6^2(100) + 0.4^2(400) + 2(0.6)(0.4)(60) = 36 + 64 + 28.8 = 128.8$.

**Step 3.** sd $= \sqrt{128.8} \approx 11.35$.

**Answer:** portfolio sd ≈ 11.35%.

### Exercise 12 · Minimum-variance weights

For the same assets, find the weight $w_1$ that minimises portfolio variance (with $w_2 = 1 - w_1$).

<details><summary>Show solution</summary>

**Step 1.** Variance as a function of $w_1$: $V(w_1) = w_1^2\sigma_1^2 + (1 - w_1)^2\sigma_2^2 + 2w_1(1 - w_1)\sigma_{12}$.

**Step 2.** Set $V'(w_1) = 2w_1\sigma_1^2 - 2(1 - w_1)\sigma_2^2 + 2(1 - 2w_1)\sigma_{12} = 0$ and solve:

$$
w_1 = \frac{\sigma_2^2 - \sigma_{12}}{\sigma_1^2 + \sigma_2^2 - 2\sigma_{12}}.
$$

**Step 3.** Plug in: $w_1 = (400 - 60)/(100 + 400 - 120) = 340/380 \approx 0.895$.

**Answer:** about 89.5% in the less risky asset (Day 26 derives the general version with a Lagrange multiplier).

</details>

### Exercise 14 · A cumulative effect

An event study estimates three post-period effects, each with variance 0.01 and pairwise covariance 0.005. Find the SE of their sum.

<details><summary>Show solution</summary>

**Step 1.** The sum is $\mathbf 1^{\top}\hat\beta$, so its variance is $\mathbf 1^{\top}V\mathbf 1$ = the sum of *all* entries of $V$.

**Step 2.** Diagonal: $3 \times 0.01 = 0.03$. Off-diagonal: 6 entries × 0.005 = 0.03.

**Step 3.** Total 0.06, SE $= \sqrt{0.06} \approx 0.245$.

**Answer:** 0.245; assuming independence would give 0.173, far too small.

</details>

### Exercise 15 · An impossible set of survey correlations

A report on a household survey lists pairwise correlations computed from different sub-samples: income–education 0.9, income–health 0.9, and education–health −0.9. Could these be the correlations of three real variables?

<details><summary>Show solution</summary>

**Step 1.** A valid correlation matrix must be PSD, so its determinant (the product of the eigenvalues) cannot be negative.

**Step 2.** Cofactor expansion of $R = \begin{pmatrix}1 & 0.9 & 0.9\\0.9 & 1 & -0.9\\0.9 & -0.9 & 1\end{pmatrix}$:

$$
\det R = 1(1 - 0.81) - 0.9(0.9 + 0.81) + 0.9(-0.81 - 0.9) = 0.19 - 1.539 - 1.539 \approx -2.89.
$$

**Step 3.** Negative, so $R$ has a negative eigenvalue (−0.8, from the code).

**Answer:** impossible. If income is strongly correlated with both education and health, those two cannot be strongly negatively correlated. Pairwise estimates from different samples can produce such incoherent matrices.

</details>

## Verify in Python

*Your work:*

```python
import numpy as np

S = np.array([[4, 2], [2, 9.]])
for a in ([1, 1], [1, -1]):
    a = np.array(a); print("Var(a'X) for a =", a, ":", a @ S @ a)
L = np.linalg.cholesky(S); print("Cholesky L =\n", L.round(3))

rng = np.random.default_rng(22)
Z = rng.normal(size=(2, 200_000)); X = (L @ Z).T
print("simulated covariance:\n", np.cov(X.T).round(2))

w = np.array([0.6, 0.4]); Sig = np.array([[100, 60], [60, 400.]])
print("portfolio sd:", round(np.sqrt(w @ Sig @ w), 2))
V = np.array([[0.01, 0.005, 0.005], [0.005, 0.01, 0.005], [0.005, 0.005, 0.01]])
print("SE of cumulative effect:", round(np.sqrt(np.ones(3) @ V @ np.ones(3)), 3))
R = np.array([[1, 0.9, 0.9], [0.9, 1, -0.9], [0.9, -0.9, 1]])
print("eigenvalues of the 'correlation matrix':", np.linalg.eigvalsh(R).round(3))
```

## Key takeaways

- $\operatorname{Var}(a^{\top}X) = a^{\top}\Sigma a$ and $\operatorname{Var}(AX + b) = A\Sigma A^{\top}$ give the variance of any linear combination.
- Covariance matrices are PSD; a matrix that is not PSD cannot describe real data.
- Cholesky factors turn independent normals into correlated ones, the basis of most simulation studies.
