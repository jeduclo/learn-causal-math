A DAG is also a statement about probability. Each variable depends only on its parents, so the joint distribution factorises into small pieces and implies conditional independences we can test.

**Learning outcomes**

- State and prove the local Markov property and use it to list testable independences.

## Theory

### From the chain rule to the DAG

For any ordering of variables, the **chain rule** of probability gives

$$
P(v_1, \dots, v_p) = \prod_{i=1}^p P(v_i \mid v_1, \dots, v_{i-1}).
$$

A DAG says more: in a topological order, each variable depends on its predecessors **only through its parents**. The joint distribution then **factorises**:

$$
P(v_1, \dots, v_p) = \prod_{i=1}^p P\big(v_i \mid \operatorname{pa}(v_i)\big).
$$

A distribution that factorises this way is **Markov** with respect to the DAG. Each factor $P(v_i \mid \operatorname{pa}(v_i))$ is a **conditional probability table (CPT)** for discrete variables, or a regression model for continuous ones.

### Fewer parameters

With $p$ binary variables, an unrestricted joint distribution needs $2^p - 1$ numbers. Under the factorisation, node $i$ needs one probability for each configuration of its parents, $2^{|\operatorname{pa}(v_i)|}$, so the total is

$$
\sum_{i=1}^p 2^{|\operatorname{pa}(v_i)|}.
$$

With 20 variables and at most two parents each: at most 80 numbers instead of about a million.

### The local Markov property

Factorisation is equivalent to the **local Markov property**:

$$
V_i \perp\!\!\!\perp \operatorname{nd}(V_i) \setminus \operatorname{pa}(V_i) \;\big|\; \operatorname{pa}(V_i),
$$

where $\operatorname{nd}(V_i)$ are the non-descendants of $V_i$. "Once you know a variable's direct causes, its other non-effects tell you nothing more about it."

### Three building blocks

| Structure | Factorisation | Implied independence |
|---|---|---|
| Chain $Z \to X \to Y$ | $P(z)P(x \mid z)P(y \mid x)$ | $Y \perp\!\!\!\perp Z \mid X$ |
| Fork $Z \leftarrow X \to Y$ | $P(x)P(z \mid x)P(y \mid x)$ | $Y \perp\!\!\!\perp Z \mid X$ |
| Collider $Z \to X \leftarrow Y$ | $P(z)P(y)P(x \mid z, y)$ | $Y \perp\!\!\!\perp Z$ (but not given $X$) |

Day 48 studies these in depth.

### Faithfulness

The Markov property says which independences *must* hold. A distribution may have *extra* independences, for instance when two causal channels cancel exactly. **Faithfulness** assumes it does not: every independence in the data comes from the graph. Causal discovery (Day 98) relies on it.

### Testable implications

Each implied independence is a **testable implication** of the DAG. If data reject $Y \perp\!\!\!\perp Z \mid X$, the graph is wrong. If data agree, the graph survives, but other graphs may imply the same independences.

### Exercise 6 · The chain screens off

For $Z \to X \to Y$, prove $P(y \mid x, z) = P(y \mid x)$ from the factorisation.

<details><summary>Show solution</summary>

**Step 1.** By definition $P(y \mid x, z) = \dfrac{P(z, x, y)}{P(z, x)}$.

**Step 2.** Numerator: $P(z)P(x \mid z)P(y \mid x)$.

**Step 3.** Denominator: sum the numerator over $y$: $P(z)P(x \mid z)\sum_y P(y \mid x) = P(z)P(x \mid z)$.

**Step 4.** The ratio is $P(y \mid x)$, which does not involve $z$. $\blacksquare$

</details>

### Exercise 7 · The fork screens off

For $Z \leftarrow X \to Y$, prove $Y \perp\!\!\!\perp Z \mid X$.

<details><summary>Show solution</summary>

**Step 1.** Factorisation: $P(x, y, z) = P(x)P(z \mid x)P(y \mid x)$.

**Step 2.** Divide by $P(x)$: $P(y, z \mid x) = P(z \mid x)\,P(y \mid x)$.

**Step 3.** That is the definition of conditional independence. $\blacksquare$

</details>

### Exercise 8 · The collider separates its causes

For $Z \to X \leftarrow Y$, prove $Z \perp\!\!\!\perp Y$ marginally.

<details><summary>Show solution</summary>

**Step 1.** Factorisation: $P(z, x, y) = P(z)P(y)P(x \mid z, y)$.

**Step 2.** Sum over $x$: $P(z, y) = P(z)P(y)\sum_x P(x \mid z, y) = P(z)P(y)$.

**Step 3.** So $Z$ and $Y$ are independent. $\blacksquare$ (Conditioning on $X$ generally destroys this; see Exercise 14.)

</details>

### Exercise 9 · The factorisation is a valid distribution

Prove that $\prod_i P(v_i \mid \operatorname{pa}(v_i))$ sums to 1 over all values, for any choice of CPTs.

<details><summary>Show solution</summary>

**Step 1.** Order the nodes topologically, $V_1, \dots, V_p$. The last node $V_p$ is a parent of no one, so it appears only in its own factor.

**Step 2.** Summing over $v_p$ first: $\sum_{v_p} P(v_p \mid \operatorname{pa}(v_p)) = 1$, which removes the last factor.

**Step 3.** What remains is the same product for $V_1, \dots, V_{p-1}$, in which $V_{p-1}$ is now last.

**Step 4.** Repeat until no factors remain: the total is 1. $\blacksquare$

</details>

### Exercise 10 · Local Markov implies factorisation

Prove that if each $V_i$ satisfies the local Markov property, then the joint distribution factorises over the DAG.

<details><summary>Show solution</summary>

**Step 1.** Take a topological order and apply the chain rule: $P(v) = \prod_i P(v_i \mid v_1, \dots, v_{i-1})$.

**Step 2.** The predecessors $\{V_1, \dots, V_{i-1}\}$ contain all parents of $V_i$ and no descendants of $V_i$ (descendants come later in the order).

**Step 3.** So the predecessors are $\operatorname{pa}(V_i)$ plus some other non-descendants.

**Step 4.** By local Markov, those other non-descendants can be dropped: $P(v_i \mid v_1, \dots, v_{i-1}) = P(v_i \mid \operatorname{pa}(v_i))$.

**Step 5.** Substituting gives the factorisation. $\blacksquare$

</details>

### Exercise 11 · Vaccination and age

In a flu season, being older ($O = 1$, 30% of the population) affects both vaccination $V$ and infection $I$, and vaccination affects infection: $O \to V$, $O \to I$, $V \to I$. $P(V = 1 \mid O = 1) = 0.8$, $P(V = 1 \mid O = 0) = 0.4$. Infection risks $P(I = 1 \mid V, O)$: young unvaccinated 0.10, young vaccinated 0.04, old unvaccinated 0.30, old vaccinated 0.12. Compute $P(I = 1)$, $P(I = 1 \mid V = 1)$ and $P(I = 1 \mid V = 0)$. Compare the crude risk ratio with the age-specific ones.

<details><summary>Show solution</summary>

**Step 1.** Joint probabilities with $I = 1$, using $P(o)P(v \mid o)P(I = 1 \mid v, o)$:

- young, vaccinated: $0.7 \times 0.4 \times 0.04 = 0.0112$;
- young, unvaccinated: $0.7 \times 0.6 \times 0.10 = 0.0420$;
- old, vaccinated: $0.3 \times 0.8 \times 0.12 = 0.0288$;
- old, unvaccinated: $0.3 \times 0.2 \times 0.30 = 0.0180$.

**Step 2.** $P(I = 1) = 0.0112 + 0.0420 + 0.0288 + 0.0180 = 0.100$.

**Step 4.** $P(I = 1 \mid V = 0) = (0.0420 + 0.0180)/0.48 = 0.06/0.48 = 0.125$.

**Step 5.** Crude risk ratio: $0.077/0.125 \approx 0.62$. Age-specific: $0.04/0.10 = 0.4$ and $0.12/0.30 = 0.4$.

**Answer:** the vaccine cuts risk by 60% in each age group, but the crude comparison suggests only 38%, because the vaccinated are disproportionately old (the fork $V \leftarrow O \to I$).

### Exercise 14 · Hospital admissions (Berkson)

Flu $F$ (10% of people) and injury $J$ (5%) are independent. Hospital admission $H$ depends on both: $P(H = 1 \mid F, J)$ is 0.01 for neither, 0.3 for flu only, 0.5 for injury only and 0.8 for both. Among admitted patients, compute $P(J = 1)$, $P(J = 1 \mid F = 1)$ and $P(J = 1 \mid F = 0)$.

<details><summary>Show solution</summary>

**Step 1.** Joint probabilities with $H = 1$:

- neither: $0.9 \times 0.95 \times 0.01 = 0.00855$;
- flu only: $0.1 \times 0.95 \times 0.3 = 0.0285$;
- injury only: $0.9 \times 0.05 \times 0.5 = 0.0225$;
- both: $0.1 \times 0.05 \times 0.8 = 0.004$.

**Step 2.** $P(H = 1) = 0.06355$.

**Step 3.** $P(J = 1 \mid H = 1) = (0.0225 + 0.004)/0.06355 \approx 0.417$.

**Step 4.** $P(J = 1 \mid H = 1, F = 1) = 0.004/(0.0285 + 0.004) \approx 0.123$.

**Step 5.** $P(J = 1 \mid H = 1, F = 0) = 0.0225/(0.00855 + 0.0225) \approx 0.725$.

**Answer:** in the population flu and injury are independent, but among hospital patients those with flu are far *less* likely to be injured. Conditioning on the collider $H$ creates the association.

### Exercise 15 · Microfinance testable implications

A DAG for micro-entrepreneurs: education $E \to$ loan $L$, $E \to$ income $I$, $L \to$ business start $B$, $B \to I$, rainfall $R \to I$. List the local Markov independences, and say which one you would test first with survey data.

<details><summary>Show solution</summary>

**Step 1.** $E$: no parents; non-descendants other than itself: $R$. So $E \perp\!\!\!\perp R$.

**Step 2.** $R$: no parents; non-descendants: $E, L, B$. So $R \perp\!\!\!\perp \{E, L, B\}$.

**Step 3.** $L$: parent $E$; non-descendants other than $E$: $R$. So $L \perp\!\!\!\perp R \mid E$.

**Step 4.** $B$: parent $L$; non-descendants: $E, R$. So $B \perp\!\!\!\perp \{E, R\} \mid L$.

**Step 5.** $I$: parents $E, B, R$; non-descendant $L$. So $I \perp\!\!\!\perp L \mid E, B, R$.

**Answer:** test $I \perp\!\!\!\perp L \mid E, B, R$ first: it says loans raise income *only* by starting businesses. Loans used for consumption smoothing or to pay off costlier debt would violate it.

</details>

## Key takeaways

- A DAG factorises the joint distribution into one conditional distribution per node given its parents.
- This is equivalent to the local Markov property: each variable is independent of its non-descendants given its parents.
- Every implied independence is a testable implication; colliders are the exception that creates dependence when conditioned on.
