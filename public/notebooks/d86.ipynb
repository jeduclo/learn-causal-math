With one treated unit there is no sampling distribution in the usual sense. Placebo (permutation) inference asks whether the treated unit's gap is unusual compared with gaps obtained by pretending each donor was treated. Synthetic difference-in-differences combines unit weights, time weights and fixed effects, nesting DiD and synthetic control.

**Learning outcomes**

- Carry out placebo-in-space and placebo-in-time analyses and compute permutation p-values from RMSPE ratios.
- Prove the validity and the resolution limit of permutation p-values.
- Define synthetic DiD and show how it reduces to DiD and relates to synthetic control.
- Compare DiD, SC and SDID when parallel trends fails.

## Theory

### Placebos in space

Re-run the synthetic control analysis treating each donor $j$ as if it had been treated (using the other units as its donor pool), and record its gap path. If the treated unit's post-period gap is extreme relative to these placebo gaps, the effect is unlikely to be chance.

Because placebos with poor pre-treatment fit produce large gaps anyway, compare the **ratio**

$$
r_j = \frac{\text{RMSPE}_{post, j}}{\text{RMSPE}_{pre, j}},
$$

and report the permutation p-value $p = \#\{j: r_j \ge r_1\}/(J + 1)$ (counting the treated unit itself).

### Placebos in time

Pretend the policy started earlier, using only pre-treatment data; a large "effect" at a fake date signals poor fit or a pre-existing break.

### What the p-value means

If treatment assignment were random across the $J + 1$ units (exchangeability), the treated unit's ratio would be equally likely to have any rank, so $P(p \le k/(J + 1)) = k/(J + 1)$. The smallest attainable p-value is $1/(J + 1)$: with 19 donors, 0.05. It is a design-based test of "no effect anywhere" (a sharp null), in the spirit of Fisher (Day 37).

### Synthetic difference-in-differences (Arkhangelsky et al., 2021)

1. **Unit weights** $\omega$ (non-negative, summing to one) make the weighted donors' pre-period path *parallel* to the treated unit's (an intercept is allowed, with a small ridge penalty for stability).
2. **Time weights** $\lambda$ over pre-periods make the weighted pre-period resemble the post-period for donors.
3. The estimate is a weighted DiD:

$$
\hat\tau_{SDID} = \Big(\bar Y_{1, post} - \sum_t\lambda_tY_{1t}\Big) - \sum_j\omega_j\Big(\bar Y_{j, post} - \sum_t\lambda_tY_{jt}\Big).
$$

With uniform $\omega$ and $\lambda$ this is DiD; without time weights and without the intercept (levels must match) it is close to synthetic control. Like DiD it tolerates level differences; like SC it chooses comparison units to match trends.

## Verify in Python

*Your work:*

```python
import numpy as np
from scipy.optimize import minimize

r = np.array([4.1, 1.2, .8, 2.3, 4.5, 1.1, .9, 1.6, 2.0, 1.4, .7, 3.2, 1.0, 1.3, 2.8])
print("Ex 12 p-value:", round((r >= r[0]).mean(), 3), " Ex 1:", round(1 / 39, 3))

rng = np.random.default_rng(86)
J, T0, T = 20, 15, 25
mu = rng.normal(size=(J + 1, 2)); mu[1:4, 1] = [1.8, 1.4, 1.6]
lam = rng.normal(size=(T, 2)); lam[:, 1] = np.linspace(-2, 2, T)
alpha = rng.normal(0, 2, J + 1)
mu[0] = .5 * mu[1] + .3 * mu[2] + .2 * mu[3]; alpha[0] = .5 * alpha[1] + .3 * alpha[2] + .2 * alpha[3] + 1.5
Y = alpha[:, None] + np.linspace(0, 3, T)[None, :] + mu @ lam.T + .2 * rng.normal(size=(J + 1, T))
Y[0, T0:] += -2

def simplex_ls(A, b, intercept=False, ridge=0.0):
    k = A.shape[1]
    def f(z):
        r = A @ z[:k] + (z[k] if intercept else 0) - b
        return r @ r + ridge * (z[:k] @ z[:k])
    z0 = np.r_[np.full(k, 1 / k), [0.0] if intercept else []]
    bnds = [(0, 1)] * k + ([(None, None)] if intercept else [])
    return minimize(f, z0, bounds=bnds, constraints=({"type": "eq", "fun": lambda z: z[:k].sum() - 1},),
                    method="SLSQP", options={"maxiter": 500}).x[:k]
def sc_gap(Y, i):
    don = [j for j in range(Y.shape[0]) if j != i]
    w = simplex_ls(Y[don, :T0].T, Y[i, :T0]); return Y[i] - w @ Y[don]
def sdid(Y):
    w = simplex_ls(Y[1:, :T0].T, Y[0, :T0], intercept=True, ridge=T0 * .05)
    l = simplex_ls(Y[1:, :T0], Y[1:, T0:].mean(1), intercept=True)
    return (Y[0, T0:].mean() - l @ Y[0, :T0]) - w @ (Y[1:, T0:].mean(1) - Y[1:, :T0] @ l)
did = (Y[0, T0:].mean() - Y[0, :T0].mean()) - (Y[1:, T0:].mean() - Y[1:, :T0].mean())
print(f"DiD {did:.2f}   SC {sc_gap(Y, 0)[T0:].mean():.2f}   SDID {sdid(Y):.2f}   (truth -2)")
ratios = np.array([np.sqrt((g[T0:] ** 2).mean() / (g[:T0] ** 2).mean()) for g in (sc_gap(Y, i) for i in range(J + 1))])
print(f"treated ratio {ratios[0]:.1f}; rank {(ratios >= ratios[0]).sum()} of {J + 1}; p = {(ratios >= ratios[0]).mean():.3f}")
```

## Key takeaways

- Placebo-in-space inference compares the treated unit's post/pre RMSPE ratio with those of donors; its p-value is valid under exchangeability and cannot go below 1/(J+1).
- Placebos in time and leave-one-out analyses check that the synthetic unit tracks the treated unit for the right reasons.
- Synthetic DiD adds unit and time weights to DiD (or an intercept to SC), tolerating level gaps while matching trends.
