# Day 62: The Wald Estimator

*Beginner to Ninja: Math for Causal Inference*

With a binary instrument, IV is a ratio of two differences in means. The Wald estimator, its grouped-data generalisation, its delta-method standard error, and the special case of one-sided non-compliance where it recovers the effect on the treated.

**Learning outcomes**

- Compute the Wald estimator from group means and relate it to the covariance formula.
- Read IV graphically as the slope through group means.
- Derive a delta-method standard error for a ratio of differences.
- Prove that with one-sided non-compliance the Wald estimator equals the effect of treatment on the treated.

## Theory

### Binary instrument

Let $Z \in \{0, 1\}$ (won a lottery, born in an eligible year, offered a voucher). The **Wald estimator** is

$$
\hat\beta_{W} = \frac{\bar Y_1 - \bar Y_0}{\bar X_1 - \bar X_0},
$$

where $\bar Y_z, \bar X_z$ are means among units with $Z = z$. The numerator is the **reduced form** (intention-to-treat effect on $Y$), the denominator the **first stage** (effect of $Z$ on $X$; for binary $X$, the difference in take-up rates).

### Same as IV

For binary $Z$, $\operatorname{Cov}(Z, Y) = p(1 - p)(\bar Y_1 - \bar Y_0)$ with $p = P(Z = 1)$, and similarly for $X$. The factor $p(1 - p)$ cancels, so the Wald estimator is the IV estimator.

### The picture

Plot the two group means $(\bar X_0, \bar Y_0)$ and $(\bar X_1, \bar Y_1)$: the Wald estimate is the slope of the line through them. With several groups (quarters of birth, lottery cohorts), fitting a line through the group means, weighted by group size, is **2SLS with group dummies as instruments** (Angrist's grouping estimator).

### Standard error

By the delta method, with reduced form $\hat\rho$, first stage $\hat\pi$ and $\hat\beta = \hat\rho/\hat\pi$,

$$
\operatorname{Var}(\hat\beta) \approx \frac{\operatorname{Var}(\hat\rho) + \beta^2\operatorname{Var}(\hat\pi) - 2\beta\operatorname{Cov}(\hat\rho, \hat\pi)}{\pi^2}.
$$

A small first stage $\pi$ inflates the variance quadratically (Day 65).

### One-sided non-compliance

If nobody with $Z = 0$ can get treated (a programme only lottery winners can join), then under exclusion the Wald estimator equals the **effect of treatment on the treated** (Bloom's result). Phase 3, Day 41, called the same ratio the CACE; Day 63 develops the general case.

### Exercise 6 · Wald equals IV

For binary $Z$ with $p = P(Z = 1)$, prove $\operatorname{Cov}(Z, Y) = p(1 - p)(E[Y \mid Z = 1] - E[Y \mid Z = 0])$, and conclude that the Wald estimand equals $\operatorname{Cov}(Z, Y)/\operatorname{Cov}(Z, X)$.

<details><summary>Show solution</summary>

**Step 1.** $\operatorname{Cov}(Z, Y) = E[ZY] - E[Z]E[Y] = pE[Y \mid Z = 1] - p\big(pE[Y \mid Z = 1] + (1 - p)E[Y \mid Z = 0]\big)$.

**Step 2.** $= p(1 - p)E[Y \mid Z = 1] - p(1 - p)E[Y \mid Z = 0]$.

**Step 3.** The same holds for $X$; the factor $p(1 - p)$ cancels in the ratio. $\blacksquare$

</details>

### Exercise 7 · Delta method for a ratio

Let $g(\rho, \pi) = \rho/\pi$. Derive the delta-method variance in the theory.

<details><summary>Show solution</summary>

**Step 1.** Gradient: $\partial g/\partial\rho = 1/\pi$, $\partial g/\partial\pi = -\rho/\pi^2 = -\beta/\pi$.

**Step 2.** $\operatorname{Var}(g) \approx \nabla g^\top\boldsymbol\Sigma\nabla g = \frac{1}{\pi^2}\operatorname{Var}(\hat\rho) + \frac{\beta^2}{\pi^2}\operatorname{Var}(\hat\pi) - \frac{2\beta}{\pi^2}\operatorname{Cov}(\hat\rho, \hat\pi)$.

**Step 3.** Factor out $1/\pi^2$. $\blacksquare$ (Day 12.)

</details>

### Exercise 8 · One-sided non-compliance gives the ATT

Let $Z$ be random, $D \in \{0, 1\}$ with $D = 0$ whenever $Z = 0$, and exclusion: $Y = Y(D)$ does not depend on $Z$ otherwise. Prove the Wald estimand equals $E[Y(1) - Y(0) \mid D = 1]$.

<details><summary>Show solution</summary>

**Step 1.** Reduced form: $E[Y \mid Z = 1] - E[Y \mid Z = 0] = E[Y(0) + D(Y(1) - Y(0)) \mid Z = 1] - E[Y(0) \mid Z = 0]$.

**Step 2.** Randomisation: $E[Y(0) \mid Z = 1] = E[Y(0) \mid Z = 0]$, so the reduced form is $E[D(Y(1) - Y(0)) \mid Z = 1]$.

**Step 3.** $= P(D = 1 \mid Z = 1)\,E[Y(1) - Y(0) \mid D = 1, Z = 1]$.

**Step 4.** First stage: $P(D = 1 \mid Z = 1) - 0$. Ratio: $E[Y(1) - Y(0) \mid D = 1, Z = 1]$.

**Step 5.** Everyone with $D = 1$ has $Z = 1$, so this is the ATT. $\blacksquare$

</details>

### Exercise 9 · Grouping estimator with two groups

Prove that the IV estimator with a binary instrument equals the slope of the line through the two sample group means.

<details><summary>Show solution</summary>

**Step 1.** Exercise 6 holds in the sample too: $\widehat{\operatorname{Cov}}(Z, Y) = \hat p(1 - \hat p)(\bar Y_1 - \bar Y_0)$ (with the same divisor for both covariances).

**Step 2.** The ratio is $(\bar Y_1 - \bar Y_0)/(\bar X_1 - \bar X_0)$, which is the slope through the two points. $\blacksquare$

### Exercise 10 · A direct effect of the instrument

If $Z$ also affects $Y$ directly by $\delta$ (exclusion fails), show the Wald estimand is $\beta + \delta/\pi$ in the constant-effect model $Y = \alpha + \beta X + \delta Z + \varepsilon$, $E[\varepsilon \mid Z] = 0$.

<details><summary>Show solution</summary>

**Step 2.** Divide by $\pi$: $\beta + \delta/\pi$. $\blacksquare$

</details>

## Verify in Python

*Your work:*

```python
import numpy as np

rng = np.random.default_rng(62)
# A simulated lottery: Z random, compliers/always/never takers, effect -2800 for those who serve
n = 400_000
Z = rng.random(n) < .3
kind = rng.choice(["c", "a", "n"], size=n, p=[.16, .19, .65])
D = (kind == "a") | ((kind == "c") & Z)
Y0 = rng.normal(16_000 + 600 * (kind == "a"), 5_000, n)
Y = Y0 - 2_800 * D
rf = Y[Z].mean() - Y[~Z].mean(); fs = D[Z].mean() - D[~Z].mean()
iv = np.cov(Z, Y)[0, 1] / np.cov(Z, D)[0, 1]
print(f"reduced form {rf:.0f}, first stage {fs:.3f}, Wald {rf / fs:.0f}, covariance IV {iv:.0f}")

# Delta-method SE versus the bootstrap
var_rf = Y[Z].var() / Z.sum() + Y[~Z].var() / (~Z).sum()
var_fs = D[Z].var() / Z.sum() + D[~Z].var() / (~Z).sum()
b = rf / fs
cov = np.cov(Y[Z], D[Z])[0, 1] / Z.sum() + np.cov(Y[~Z], D[~Z])[0, 1] / (~Z).sum()
print(f"delta-method SE {np.sqrt(var_rf + b**2 * var_fs - 2 * b * cov) / abs(fs):.0f}")
boot = []
for _ in range(200):
    i = rng.integers(0, n, n); z, d, y = Z[i], D[i], Y[i]
    boot.append((y[z].mean() - y[~z].mean()) / (d[z].mean() - d[~z].mean()))
print(f"bootstrap SE {np.std(boot):.0f}")

# Exercises 11, 13, 15
b = -450 / .16
print("draft:", b, " SE", round(np.sqrt(200**2 + b**2 * .01**2) / .16), " CI", np.round([b - 1.96 * 1262.3, b + 1.96 * 1262.3]))
x = np.array([12.7, 12.8, 12.85, 12.9]); y = np.array([5.892, 5.900, 5.904, 5.908])
print("quarter-of-birth slope:", round(np.polyfit(x, y, 1)[0], 4))
```

## Key takeaways

- With a binary instrument, IV is the Wald ratio: difference in outcomes over difference in treatment between instrument groups.
- Graphically, it is the slope through group means; with many groups, a size-weighted line through the means is 2SLS.
- Its delta-method variance is inflated by $1/\pi^2$; with one-sided non-compliance it estimates the effect on the treated.
