# Day 77: RD Validity: Density Tests & Covariate Balance

*Beginner to Ninja: Math for Causal Inference*

RD fails when units can sort precisely around the cutoff or when something else changes there. The checks that make an RD credible are a smooth density of the running variable, no jumps in pre-determined covariates, no effects at placebo cutoffs, and robustness to dropping observations right at the threshold.

**Learning outcomes**

- Explain why imprecise control over the running variable makes RD as good as a local experiment, and precise control breaks it.
- Carry out and interpret a density (manipulation) test.
- Use covariate balance, placebo cutoffs and donut RD as falsification checks.
- Quantify the bias from sorting and diagnose it in health, education and political applications.

## Theory

### Why RD can work: imprecise control

Lee (2008): if units influence their running variable but **cannot control it precisely** (there is a random component with a continuous density), then the distribution of their characteristics, and hence of potential outcomes, is continuous at the cutoff. Units just above and just below are then as good as randomly assigned. If units **can** place themselves precisely on the preferred side (teachers rounding scores up to a pass mark, firms adjusting reported size to stay under a regulation threshold), those just above differ systematically from those just below.

### Test 1: the density of the running variable

Under no manipulation, the density of $X$ is continuous at $c$. **McCrary (2008)** and **Cattaneo, Jansson and Ma** estimate the density on each side by local polynomial methods and test

$$
\theta = \ln f(c^+) - \ln f(c^-) = 0.
$$

A spike just above $c$ and a dip just below indicate sorting. The test has no power against manipulation that moves units in both directions equally.

### Test 2: covariate balance

Run the RD with **pre-determined covariates** (measured before treatment, or fixed: sex, prior scores, parental income) as outcomes. They should not jump. A jump means the units on either side differ in more than treatment.

### Test 3: placebo cutoffs and outcomes

Estimate "effects" at fake cutoffs where nothing happens (for instance, the median of $X$ on each side), and effects on outcomes the treatment cannot affect. Both should be zero.

### Robustness: donut RD

If manipulation or **heaping** (values piling up at round numbers, such as birth weights recorded at 1,500 g) is concentrated very close to $c$, drop observations within a small "donut" around $c$ and re-estimate. Stability reassures; sensitivity reveals problems (at the cost of more extrapolation).

### Other threats

Compound treatments (several policies switch at the same age), and sorting on the outcome side (selection into the sample that depends on crossing the cutoff).

### Exercise 6 · Imprecise control gives continuity

Let $X = X^* + \eta$, where units choose $X^*$ (possibly based on their potential outcomes) and $\eta$ is independent noise with a continuous density $g$. Show that the density of $X$ given any type of unit is continuous, so the share of each type is continuous at $c$.

<details><summary>Show solution</summary>

**Step 1.** For a unit type $u$ with chosen $X^* = x^*_u$, the density of $X$ is $g(x - x^*_u)$, continuous in $x$.

**Step 2.** Mixing over types: $f(x \mid u)$ continuous, and $P(u \mid X = x) = f(x \mid u)P(u)/f(x)$ is continuous wherever $f(x) > 0$.

**Step 3.** So the composition of units, and $E[Y(d) \mid X = x]$, is continuous at $c$. $\blacksquare$

</details>

### Exercise 7 · SE of a log count ratio

With independent Poisson counts $n_1, n_2$, show $\operatorname{Var}(\ln n_2 - \ln n_1) \approx 1/n_1 + 1/n_2$.

<details><summary>Show solution</summary>

**Step 1.** Delta method: $\operatorname{Var}(\ln n) \approx \operatorname{Var}(n)/E[n]^2 = E[n]/E[n]^2 = 1/E[n]$ for a Poisson count.

**Step 2.** Independence: add the two variances. $\blacksquare$

### Exercise 8 · The bias from sorting

Suppose a share $s$ of units just above the cutoff are "movers" who would naturally have been just below, with mean untreated outcome $\mu_0 + \Delta$ instead of $\mu_0$, and the effect is $\tau$ for everyone. Show the RD estimate converges to $\tau + s\Delta$ if movers come from a thin band just below (whose loss does not change the left limit).

<details><summary>Show solution</summary>

**Step 1.** Right limit: $(1 - s)(\mu_0 + \tau) + s(\mu_0 + \Delta + \tau) = \mu_0 + \tau + s\Delta$.

**Step 2.** Left limit: $\mu_0$ (by assumption).

**Step 3.** Difference: $\tau + s\Delta$. $\blacksquare$ Positive selection of movers ($\Delta > 0$) overstates the effect.

</details>

### Exercise 9 · Covariate balance is implied by continuity

Show that if the composition of units is continuous at $c$ (Exercise 6), then $E[W \mid X = x]$ is continuous at $c$ for every pre-determined $W$.

<details><summary>Show solution</summary>

**Step 1.** $E[W \mid X = x] = \sum_u E[W \mid u]P(u \mid X = x)$, with $W$ fixed for each unit type.

**Step 2.** Each $P(u \mid X = x)$ is continuous, so the sum is. $\blacksquare$ A jump in $E[W \mid X]$ therefore contradicts the continuity assumption that justifies the RD.

</details>

### Exercise 10 · Two-sided manipulation with a smooth density

Construct an example in which the density of $X$ is continuous at $c$ but the composition jumps.

<details><summary>Show solution</summary>

**Step 1.** Start with a uniform $X$ for two equal-sized groups, A and B.

**Step 2.** A teacher moves A students from $[c - \epsilon, c)$ to $[c, c + \epsilon)$, and an equal number of B students from $[c, c + \epsilon)$ to $[c - \epsilon, c)$.

**Step 3.** Total counts on each side are unchanged, but just above $c$ there are only A students and just below only B students.

**Answer:** the density test passes while covariate balance fails. $\blacksquare$

</details>

## Verify in Python

*Your work:*

```python
import numpy as np

z = lambda n1, n2: np.log(n2 / n1) / np.sqrt(1 / n1 + 1 / n2)
print(f"Ex 1: log ratio {np.log(620 / 480):.3f}, z {z(480, 620):.1f} | Ex 12: log ratio {np.log(2100 / 900):.2f}, z {z(900, 2100):.1f}")

rng = np.random.default_rng(77); n = 40_000
x = rng.uniform(-1, 1, n); y0 = 10 + 4 * x + rng.normal(0, 2, n)
mover = (x < 0) & (x > -.05) & (y0 - 10 - 4 * x > 0) & (rng.random(n) < .6)
xm = x.copy(); xm[mover] = rng.uniform(0, .05, mover.sum())
y = y0 + 3 * (xm >= 0)
def rd(x, y, h=.3):
    def side(xx, yy):
        w = np.clip(1 - np.abs(xx) / h, 0, None); m = w > 0
        X = np.column_stack([np.ones(m.sum()), xx[m]]); W = w[m]
        return np.linalg.solve(X.T @ (W[:, None] * X), X.T @ (W * yy[m]))[0]
    r = x >= 0
    return side(x[r], y[r]) - side(-x[~r], y[~r])
nl, nr = ((xm < 0) & (xm >= -.05)).sum(), ((xm >= 0) & (xm < .05)).sum()
print(f"movers {mover.sum()}; density bins {nl} vs {nr}: log ratio {np.log(nr / nl):.2f}, z {z(nl, nr):.1f}")
keep = np.abs(xm) > .05
print(f"RD with sorting {rd(xm, y):.2f}  donut {rd(xm[keep], y[keep]):.2f}  no sorting {rd(x, y0 + 3 * (x >= 0)):.2f}  (truth 3)")
```

## Key takeaways

- RD is credible when units cannot precisely control which side of the cutoff they land on.
- Check the density of the running variable, balance of pre-determined covariates, placebo cutoffs and outcomes, and robustness to a donut.
- Sorting by units with different potential outcomes biases the jump; two-sided manipulation can hide from density tests but not from covariate checks.
