# Day 73: Sharp Regression Discontinuity

*Beginner to Ninja: Math for Causal Inference*

When treatment switches on at a known threshold of a running variable, units just above and just below are comparable. The jump in the outcome at the cutoff identifies the effect there, under one assumption: potential outcomes are continuous in the running variable.

**Learning outcomes**

- Define the sharp RD estimand and state the continuity assumption.
- Estimate the jump with separate linear fits on each side and explain why a raw difference in means is biased.
- Prove identification and explain why the effect is local to the cutoff.
- Apply RD to drinking age, incumbency, scholarships and Medicare eligibility.

## Theory

### Design

A **running variable** $X$ (age, test score, vote margin, income) determines treatment by a rule: $D = \mathbf 1[X \ge c]$. There is no overlap: every unit with $X \ge c$ is treated and every unit below is not. Units cannot be compared at the same $X$, but they can be compared arbitrarily close to $c$.

### Estimand and assumption

$$
\tau_{SRD} = \lim_{x \downarrow c}E[Y \mid X = x] - \lim_{x \uparrow c}E[Y \mid X = x].
$$

**Continuity:** $E[Y(0) \mid X = x]$ and $E[Y(1) \mid X = x]$ are continuous at $x = c$. Then

$$
\tau_{SRD} = E[Y(1) - Y(0) \mid X = c].
$$

Continuity fails if something else also changes at the cutoff (another policy at age 21), or if units manipulate $X$ to land on the desired side (Day 77).

### Estimation

Fit $E[Y \mid X]$ separately on each side near $c$ and take the difference of the two fitted values at $c$. The standard choice is **local linear regression** within a bandwidth $h$ (Days 74–75):

$$
Y_i = \alpha + \tau D_i + \beta_1(X_i - c) + \beta_2 D_i(X_i - c) + \varepsilon_i, \qquad |X_i - c| \le h.
$$

A **raw difference in means** within $[c - h, c + h]$ is biased whenever $E[Y \mid X]$ has a slope; **high-order global polynomials** fitted to the whole range give erratic estimates and should be avoided (Gelman and Imbens).

### What RD identifies

The effect for units at the cutoff, an extreme form of locality. Away from $c$ the data say nothing without further assumptions. Heterogeneity in effects along $X$ can be studied only with extrapolation or multiple cutoffs.

### Two views

- **Continuity:** limits of conditional expectations, estimated by local polynomials.
- **Local randomisation:** within a small window, treatment is as good as random (for instance when $X$ includes a random component that units cannot control precisely); analyse as an experiment.

### Exercise 6 · Identification

Prove $\tau_{SRD} = E[Y(1) - Y(0) \mid X = c]$ under continuity.

<details><summary>Show solution</summary>

**Step 1.** For $x \ge c$, $E[Y \mid X = x] = E[Y(1) \mid X = x]$; for $x < c$, $E[Y \mid X = x] = E[Y(0) \mid X = x]$.

**Step 2.** By continuity, $\lim_{x \downarrow c}E[Y(1) \mid X = x] = E[Y(1) \mid X = c]$ and $\lim_{x \uparrow c}E[Y(0) \mid X = x] = E[Y(0) \mid X = c]$.

**Step 3.** Subtract. $\blacksquare$

### Exercise 8 · Separate regressions

Show that the pooled regression $Y = \alpha + \tau D + \beta_1(X - c) + \beta_2 D(X - c)$ within the window gives the same $\hat\tau$ as fitting separate lines on each side and differencing their intercepts at $c$.

Construct two models consistent with the same observed $E[Y \mid X]$ but with different effects at $x = c + 1$.

### Exercise 10 · Covariates do not change the estimand

Let $W$ be a pre-treatment covariate with $E[W \mid X]$ continuous at $c$. Explain why adding $W$ linearly to the local linear regression leaves the estimand unchanged (in large samples) but can reduce variance.

<details><summary>Show solution</summary>

**Step 1.** Including $W$ changes $\hat\tau$ by its coefficient times the jump in $E[W \mid X]$ at $c$ (FWL).

**Step 2.** Continuity of $E[W \mid X]$ makes that jump zero in the limit.

**Step 3.** $W$ explains outcome variance, shrinking residuals and the standard error.

**Answer:** same estimand, possibly more precision; a jump in $W$ would instead signal a violation (Day 77). $\blacksquare$

</details>

## Verify in Python

*Your work:*

```python
import numpy as np

rng = np.random.default_rng(73)
c, tau = 0.0, 5.0
mu0 = lambda x: 50 + 8 * x + 6 * x ** 2 - 4 * x ** 3          # smooth untreated outcome
def sample(n=1000):
    x = rng.uniform(-1, 1, n); y = mu0(x) + tau * (x >= c) + rng.normal(0, 3, n)
    return x, y
def local_linear(x, y, h):
    m = np.abs(x - c) <= h; d = (x >= c).astype(float)
    X = np.column_stack([np.ones(m.sum()), d[m], x[m] - c, d[m] * (x[m] - c)])
    return np.linalg.lstsq(X, y[m], rcond=None)[0][1]
def window_means(x, y, h):
    return y[(x >= c) & (x <= c + h)].mean() - y[(x < c) & (x >= c - h)].mean()
def global_poly(x, y, p=4):
    d = (x >= c).astype(float)
    X = np.column_stack([np.ones_like(x), d] + [x ** k for k in range(1, p + 1)] + [d * x ** k for k in range(1, p + 1)])
    return np.linalg.lstsq(X, y, rcond=None)[0][1]
res = {"local linear h=0.3": [], "window means h=0.3": [], "global quartic": []}
for _ in range(1000):
    x, y = sample()
    res["local linear h=0.3"].append(local_linear(x, y, .3))
    res["window means h=0.3"].append(window_means(x, y, .3))
    res["global quartic"].append(global_poly(x, y))
for k, v in res.items():
    print(f"{k:20s} mean {np.mean(v):6.2f}  sd {np.std(v):5.2f}  (truth {tau})")
print("Ex 4 bias s*h:", 2 * 1, " | Ex 11:", 103 - 94, f"({(103 - 94) / 94:.1%})")
```

## Key takeaways

- Sharp RD identifies the effect at the cutoff as the jump in $E[Y \mid X]$, assuming potential outcomes are continuous there.
- Estimate with local linear fits on each side; raw window means are biased by the slope, global polynomials are fragile.
- The effect is local; extrapolating away from the cutoff needs extra assumptions.
