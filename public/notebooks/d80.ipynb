## Theory

### Start from the variation

The first question is not "which estimator?" but "what moves treatment for reasons unrelated to outcomes?"

| Source of variation | Design | Estimand | Key assumption | Diagnostics |
|---|---|---|---|---|
| You randomise | RCT (Phase 3) | ATE (or ITT) | randomisation, no attrition bias | balance, attrition, compliance |
| A random or as-good-as-random nudge to take-up | IV (61–66) | LATE for compliers | independence, exclusion, monotonicity | first stage F, balance on Z, AR intervals, complier profile |
| A policy changes for some groups at some time | DiD / event study (67–72) | ATT for treated groups | parallel trends, no anticipation | pre-trends, placebo outcomes, sensitivity bounds, robust staggered estimators |
| A rule assigns treatment at a threshold | RD (73–77) | effect at the cutoff (compliers there if fuzzy) | continuity at the cutoff | density test, covariate balance, placebo cutoffs, bandwidth sensitivity |
| A rule changes the slope of a schedule | RKD (78) | marginal effect at the kink | smoothness | slope placebo, density derivative |
| None of the above is credible | bounds (79), sensitivity (95) | identified set | only what you state | transparency about assumptions |

### Estimands differ

Each design answers a question about a particular group: compliers with a lottery, treated states, students at a test-score cutoff. With heterogeneous effects, credible designs applied to the same policy can legitimately disagree. Before comparing numbers, compare estimands.

### Combining designs

- **Wald-DiD (fuzzy DiD):** if a policy raises take-up in treated groups, divide the DiD in outcomes by the DiD in take-up.
- **Difference in discontinuities:** if a jump already existed at a cutoff before a policy started applying there, subtract the old jump from the new one.
- **RD + bounds, IV + DiD, event studies with instruments**: assumptions combine, and so do diagnostics.

### A workflow

1. State the causal question and the estimand (whose effect, of what, on what, when).
2. Describe the assignment mechanism with institutional detail; draw the DAG.
3. Choose the design that matches the variation; state its assumptions plainly.
4. Run the diagnostics; report the reduced form or raw comparisons graphically.
5. Report estimates with robust inference and sensitivity analyses.
6. Discuss external validity: for whom does the estimate speak?

### Exercise 6 · Wald-DiD

Suppose a reform raises take-up $D$ more in group 1 than in group 0, and $Y = \alpha_g + \lambda_t + \tau D + \varepsilon$ with a constant effect $\tau$ and parallel trends in $\varepsilon$. Prove that the DiD in $Y$ divided by the DiD in $D$ equals $\tau$.

<details><summary>Show solution</summary>

**Step 1.** Group-period means: $\bar Y_{gt} = \alpha_g + \lambda_t + \tau\bar D_{gt} + \bar\varepsilon_{gt}$.

**Step 2.** DiD in $Y$: the $\alpha_g$ and $\lambda_t$ cancel; parallel trends cancels the $\varepsilon$ terms; left with $\tau\times$ DiD in $D$.

**Step 3.** Divide (DiD in $D$ non-zero). $\blacksquare$ With heterogeneous effects, extra assumptions are needed for a LATE interpretation (de Chaisemartin and D'Haultfœuille).

</details>

### Exercise 7 · Difference in discontinuities

A cutoff at age 65 had a jump $J_0$ in employment before a pension reform (from other age-65 rules) and $J_1$ after the reform also started at 65. Show that $J_1 - J_0$ identifies the reform's effect at 65 if the other rules' jump is unchanged.

<details><summary>Show solution</summary>

**Step 1.** $J_0 = \delta$ (other rules); $J_1 = \delta + \tau$ (other rules plus reform).

**Step 2.** $J_1 - J_0 = \tau$ when $\delta$ is the same in both periods. $\blacksquare$ This combines RD's local comparison with DiD's removal of a stable confounder.

</details>

### Exercise 9 · OLS or IV by mean squared error

OLS has bias $b$ and variance $v_O$; IV is unbiased with variance $v_I > v_O$. Show IV has lower MSE iff $b^2 > v_I - v_O$.

<details><summary>Show solution</summary>

**Step 1.** $\text{MSE}_{OLS} = b^2 + v_O$; $\text{MSE}_{IV} = v_I$.

**Step 2.** $v_I < b^2 + v_O \iff b^2 > v_I - v_O$. $\blacksquare$ With a weak instrument, $v_I$ can be so large that a biased OLS is closer to the truth, but its bias is unknown, so IV's confidence interval is the honest summary.

### Exercise 10 · Reweighting for external validity

Suppose effects vary only with a covariate $X$: $\tau(x)$, and a study identifies $\tau(x)$ for each $x$ in its sample. Show the ATE for a target population with covariate distribution $Q$ is $\int\tau(x)\,dQ(x)$, and that the study's own average uses its own distribution $P$.

<details><summary>Show solution</summary>

**Step 1.** Target ATE: $E_Q[Y(1) - Y(0)] = E_Q[E[Y(1) - Y(0) \mid X]] = E_Q[\tau(X)]$.

**Step 2.** The study's average is $E_P[\tau(X)]$; they differ when $P \ne Q$ and $\tau$ varies.

**Answer:** reweighting the study's conditional effects to the target's covariate mix transports the estimate, if effects depend only on observed $X$ (and $Q$'s support is covered). $\blacksquare$

</details>

### Exercise 15 · One world, three estimands

In the simulation below, a programme's effect is larger for poorer households: $\tau(x) = 2 - 1.5x$ for a poverty score $x \in [-1, 1]$ (lower is poorer). Households with $x < 0$ are eligible; a lottery among eligible households offers places, and poorer households comply more. <details><summary>Show solution</summary>

## Verify in Python

*Your work:*

```python
import numpy as np

rng = np.random.default_rng(80); n = 2_000_000
x = rng.uniform(-1, 1, n)
tau = 2 - 1.5 * x
y0 = 5 + x + rng.normal(0, 1, n)
eligible = x < 0
print(f"population ATE {tau.mean():.3f}   eligible ATE {tau[eligible].mean():.3f}")

# Sharp RD at the cutoff with universal take-up for the eligible
d_rd = eligible.astype(float); y = y0 + d_rd * tau
def side(xx, yy, h=.2):
    w = np.clip(1 - np.abs(xx) / h, 0, None); m = w > 0
    X = np.column_stack([np.ones(m.sum()), xx[m]]); W = w[m]
    return np.linalg.solve(X.T @ (W[:, None] * X), X.T @ (W * yy[m]))[0]
rd = side(-x[eligible], y[eligible]) - side(x[~eligible], y[~eligible])
print(f"RD at cutoff {rd:.3f} (tau(0) = 2)")

# Lottery among the eligible; poorer households comply more
e = eligible
Z = rng.random(n) < .5
comply = rng.random(n) < (0.3 - 0.6 * x)          # poorer households (x near -1) comply with prob 0.9, those near 0 with 0.3
D = e & Z & comply
yL = y0 + D * tau
wald = (yL[e & Z].mean() - yL[e & ~Z].mean()) / (D[e & Z].mean() - D[e & ~Z].mean())
print(f"lottery LATE {wald:.3f}  (true complier average {tau[e & comply].mean():.3f})")

# Exercise 9: OLS vs IV by MSE
b, vO, vI = .05, .0001, .0016
print("IV preferred by MSE:", b ** 2 > vI - vO)
```

## Key takeaways

- Start from the source of exogenous variation; it determines the design, its assumptions and its estimand.
- Every design has testable implications; run them, plot the raw comparisons and report robust inference.
- Credible designs can disagree because they target different people; designs can also be combined (Wald-DiD, difference in discontinuities) when one alone is not enough.
