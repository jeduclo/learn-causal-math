How much of an effect runs through a particular mechanism? Controlled and natural direct effects, natural indirect effects, the mediation formula that identifies them, and the strong no-confounding assumptions they need, including the case where regression "controlling for the mediator" goes badly wrong.

**Learning outcomes**

- Choose the right mediation quantity for a policy question.

## Theory

### Nested potential outcomes

Let $X$ be the treatment, $M$ a mediator and $Y$ the outcome. Write $M(x)$ for the mediator under treatment $x$, and $Y(x, m)$ for the outcome when $X$ is set to $x$ and $M$ to $m$. The observed outcome under treatment $x$ is $Y(x) = Y(x, M(x))$.

### Four effects (binary $X$)

- **Total effect:** $TE = E[Y(1, M(1)) - Y(0, M(0))]$.
- **Controlled direct effect** at level $m$: $CDE(m) = E[Y(1, m) - Y(0, m)]$. Fix the mediator for everyone.
- **Natural direct effect:** $NDE = E[Y(1, M(0)) - Y(0, M(0))]$. Change the treatment but keep each unit's mediator at the value it *would have had* without treatment.
- **Natural indirect effect:** $NIE = E[Y(1, M(1)) - Y(1, M(0))]$. Keep the treatment on and shift only the mediator.

They add up: $TE = NDE + NIE$. The **proportion mediated** is $NIE/TE$.

### Identification: the mediation formula

Assume, possibly within levels of pre-treatment covariates $C$: no unmeasured confounding of (1) $X$ and $Y$, (2) $X$ and $M$, (3) $M$ and $Y$, and (4) no confounder of $M$ and $Y$ that is itself affected by $X$. Then (Pearl's **mediation formula**)

$$
NDE = \sum_m \big(E[Y \mid X = 1, m] - E[Y \mid X = 0, m]\big)\,P(m \mid X = 0),
$$

$$
NIE = \sum_m E[Y \mid X = 1, m]\,\big(P(m \mid X = 1) - P(m \mid X = 0)\big).
$$

$CDE(m) = E[Y \mid X = 1, m] - E[Y \mid X = 0, m]$ needs only assumptions (1) and (3) (and can be identified even when (4) fails, by the g-formula).

### Linear models with interaction

If $E[M \mid x] = \beta_0 + \beta_1 x$ and $E[Y \mid x, m] = \theta_0 + \theta_1 x + \theta_2 m + \theta_3 xm$:

$$
CDE(m) = \theta_1 + \theta_3 m, \qquad NDE = \theta_1 + \theta_3\beta_0, \qquad NIE = (\theta_2 + \theta_3)\beta_1.
$$

Without interaction ($\theta_3 = 0$): $NDE = CDE = \theta_1$ and $NIE = \theta_2\beta_1$, the "product of coefficients" of Baron and Kenny.

### The trap

"Regress $Y$ on $X$ and $M$ and call the $X$ coefficient the direct effect" is right only under assumption (3). If something unobserved affects both $M$ and $Y$, $M$ is a collider on $X \to M \leftarrow U \to Y$, and the coefficient is biased (Day 53, Exercise 9); it can even flip sign.

### Exercise 7 · Deriving the mediation formula for the NDE

Assume (a) $Y(x, m) \perp\!\!\!\perp X$, (b) $M(x) \perp\!\!\!\perp X$, (c) $Y(x, m) \perp\!\!\!\perp M \mid X$, and (d) $Y(x, m) \perp\!\!\!\perp M(x')$ (the cross-world assumption). Prove the NDE formula.

<details><summary>Show solution</summary>

**Step 1.** $E[Y(1, M(0))] = \sum_m E[Y(1, m) \mid M(0) = m]\,P(M(0) = m)$.

**Step 2.** By (d), $E[Y(1, m) \mid M(0) = m] = E[Y(1, m)]$.

**Step 3.** By (a) and (c), $E[Y(1, m)] = E[Y(1, m) \mid X = 1] = E[Y(1, m) \mid X = 1, M = m] = E[Y \mid X = 1, M = m]$ (consistency).

**Step 4.** By (b), $P(M(0) = m) = P(M(0) = m \mid X = 0) = P(m \mid X = 0)$.

**Step 5.** So $E[Y(1, M(0))] = \sum_m E[Y \mid 1, m]P(m \mid 0)$; similarly $E[Y(0, M(0))] = \sum_m E[Y \mid 0, m]P(m \mid 0)$. Subtract. $\blacksquare$

</details>

### Exercise 8 · Linear formulas

With $E[M \mid x] = \beta_0 + \beta_1 x$ and $E[Y \mid x, m] = \theta_0 + \theta_1 x + \theta_2 m + \theta_3 xm$, derive $NDE$ and $NIE$ from the mediation formula.

<details><summary>Show solution</summary>

**Step 1.** $E[Y \mid 1, m] - E[Y \mid 0, m] = \theta_1 + \theta_3 m$.

**Step 2.** $NDE = \sum_m(\theta_1 + \theta_3 m)P(m \mid 0) = \theta_1 + \theta_3 E[M \mid X = 0] = \theta_1 + \theta_3\beta_0$.

**Step 3.** $NIE = \sum_m(\theta_0 + \theta_1 + (\theta_2 + \theta_3)m)\big(P(m \mid 1) - P(m \mid 0)\big)$.

**Step 4.** Constants sum to zero against a difference of probabilities; the $m$ term gives $(\theta_2 + \theta_3)(E[M \mid 1] - E[M \mid 0]) = (\theta_2 + \theta_3)\beta_1$. $\blacksquare$

### Exercise 10 · A treatment-induced mediator–outcome confounder

Let $X \to L \to M \to Y$, $L \to Y$, $X \to M$, $X \to Y$ (all observed). Explain why $CDE(m)$ is identified but the $NDE$ is not, without further assumptions.

<details><summary>Show solution</summary>

**Step 1.** $CDE(m) = E[Y \mid \operatorname{do}(x, m)]$ is a joint intervention; with $X$ unconfounded and $\{X, L\}$ blocking the back-door paths of $M$, the g-formula gives $E[Y \mid \operatorname{do}(x, m)] = \sum_l E[Y \mid x, l, m]P(l \mid x)$ (Day 55, Exercise 13).

**Step 2.** The NDE needs $E[Y(1, M(0))]$, which involves $L(1)$ (through the direct path $X \to L \to Y$) jointly with $M(0)$, which depends on $L(0)$.

**Step 3.** The joint distribution of $L(1)$ and $L(0)$ for the same unit is never observed and is not fixed by any observable distribution (a "recanting witness").

**Answer:** CDEs are identified; natural effects need extra assumptions or are reported with bounds. $\blacksquare$

</details>

### Exercise 13 · Smoking, birth weight and infant mortality

Return to Day 48's paradox in mediation form. In a linear model, $M$ = birth weight (standardised), $U$ = birth defects (unobserved): $M = -X - U + e_M$, $Y = 0.2X - 0.5M + 1.0U + e_Y$, with $\operatorname{Var}(U) = \operatorname{Var}(e_M) = 1$. What is the true direct effect of smoking, and what does the regression of mortality on smoking and birth weight report?

<details><summary>Show solution</summary>

**Step 1.** True direct effect: $\theta_1 = 0.2$ (harmful).

**Step 2.** Given $X$ and $M$, $-U + e_M = M + X$ is known, so $E[U \mid X, M] = -k(M + X)$ with $k = 1/(1 + 1) = 0.5$.

**Step 3.** $E[Y \mid X, M] = 0.2X - 0.5M + 1.0 \times (-0.5)(M + X) = -0.3X - 1.0M$.

**Answer:** the regression reports $-0.3$: smoking looks *protective* holding birth weight fixed, because birth weight is a collider of smoking and defects. The simulation confirms −0.3.

### Exercise 14 · Occupation and the gender pay gap

A study regresses wages on gender, "controlling for occupation", and calls the remaining gap "discrimination" (a direct effect). Draw the DAG with gender $G$, occupation $O$, wage $W$ and unobserved ability or preferences $A$ that affect both occupation and wages. What is wrong, and which quantity could be defended?

<details><summary>Show solution</summary>

**Step 1.** $G \to O \to W$, $G \to W$, $A \to O$, $A \to W$.

**Step 2.** Occupation is a mediator *and* a collider ($G \to O \leftarrow A$). Conditioning on it opens $G \to O \leftarrow A \to W$, so the "within-occupation gap" mixes discrimination with ability differences induced by selection into occupations.

**Step 3.** The total gap $TE$ is identified (gender is not caused by the other variables); the CDE given occupation needs $A$ measured.

**Answer:** report the total gap, and treat within-occupation gaps as descriptive unless the occupation–wage confounders are measured. Note also that occupational sorting can itself reflect discrimination, so the "indirect" part is not necessarily benign.

</details>

### Exercise 15 · Obesity, hypertension and stroke

A health ministry can fund either a weight-loss programme or a hypertension-control programme. Researchers estimate the effect of obesity $X$ on stroke $Y$ and the part mediated by hypertension $M$. Which mediation quantity answers "how many strokes would remain among obese people if everyone's blood pressure were controlled?" What assumptions does it need?

<details><summary>Show solution</summary>

**Step 1.** Controlling blood pressure fixes the mediator at a normal level $m_0$ for everyone: $CDE(m_0)$ is the effect of obesity that remains.

**Step 2.** Strokes prevented by hypertension control among the obese: $E[Y(1)] - E[Y(1, m_0)]$, which needs the same assumptions.

**Step 3.** Assumptions: no unmeasured confounding of obesity and stroke, and of hypertension and stroke given obesity and covariates (e.g. diet, smoking, kidney function); no cross-world assumption is needed.

**Answer:** the CDE answers the policy question; natural effects answer the mechanism question and need stronger assumptions.

</details>

## Verify in Python

*Your work:*

```python
import numpy as np

# Exercise 11: mediation formula with a binary mediator
pM = {0: .4, 1: .6}; EY = {(0, 0): 10, (0, 1): 20, (1, 0): 12, (1, 1): 24}
P = lambda m, x: pM[x] if m else 1 - pM[x]
TE = sum(EY[1, m] * P(m, 1) for m in (0, 1)) - sum(EY[0, m] * P(m, 0) for m in (0, 1))
NDE = sum((EY[1, m] - EY[0, m]) * P(m, 0) for m in (0, 1))
NIE = sum(EY[1, m] * (P(m, 1) - P(m, 0)) for m in (0, 1))
print(f"TE {TE:.2f}  NDE {NDE:.2f}  NIE {NIE:.2f}  proportion mediated {NIE / TE:.2f}")

# Exercises 3 and 12: linear formulas, checked by simulating nested potential outcomes
def effects(b0, b1, t1, t2, t3, n=1_000_000, seed=58):
    rng = np.random.default_rng(seed)
    eM = rng.normal(size=n); eY = rng.normal(size=n)
    M = lambda x: b0 + b1 * x + eM
    Y = lambda x, m: t1 * x + t2 * m + t3 * x * m + eY
    return (Y(1, M(0)) - Y(0, M(0))).mean(), (Y(1, M(1)) - Y(1, M(0))).mean()
for args in [(1, .5, 1, 2, .5), (50, 20, .10, .004, .002)]:
    b0, b1, t1, t2, t3 = args
    nde, nie = effects(*args)
    print(f"simulated NDE {nde:.3f} (formula {t1 + t3 * b0:.3f}), NIE {nie:.3f} (formula {(t2 + t3) * b1:.3f})")

# Exercise 13: regression 'direct effect' with a confounded mediator
rng = np.random.default_rng(13); n = 1_000_000
X = rng.normal(size=n); U = rng.normal(size=n)
M = -X - U + rng.normal(size=n); Y = .2 * X - .5 * M + 1.0 * U + rng.normal(size=n)
coef = np.linalg.lstsq(np.column_stack([np.ones(n), X, M]), Y, rcond=None)[0]
print(f"coefficient on smoking given birth weight: {coef[1]:.3f} (true direct effect 0.2)")
```

## Key takeaways

- $TE = NDE + NIE$; controlled direct effects fix the mediator by policy, natural effects describe mechanisms.
- The mediation formula identifies natural effects under no confounding of $X$–$Y$, $X$–$M$ and $M$–$Y$, plus a cross-world assumption that fails when treatment affects a mediator–outcome confounder.
- "Controlling for the mediator" gives a direct effect only if the mediator is unconfounded with the outcome; otherwise it is collider bias.
