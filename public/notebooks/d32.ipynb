When one unit's treatment affects another's outcome, or the "treatment" comes in several versions. How to spot it, define new estimands, and design around it.

**Learning outcomes**

## Theory

### The stable unit treatment value assumption

Writing $Y_i(1)$ and $Y_i(0)$ quietly assumed two things, together called **SUTVA** (Rubin 1980):

1. **No interference:** unit $i$'s outcome depends only on its own treatment, not on who else is treated.
2. **No hidden versions (consistency):** there is only one version of each treatment level, so $Y_i = Y_i(T_i)$.

Without (1), unit $i$ has a potential outcome for every assignment vector $\mathbf t = (t_1, \dots, t_n)$: $2^n$ of them. Nothing can be learned without some structure.

### Where interference comes from

| Mechanism | Example |
|---|---|
| Contagion | vaccination protects the unvaccinated (herd immunity); deworming cuts reinfection |
| Information | farmers copy treated neighbours' techniques |
| Competition / displacement | job-search assistance helps treated jobseekers take jobs controls would have got |
| Prices (general equilibrium) | cash transfers raise local prices, affecting non-recipients |

### Partial interference and exposure mappings

A common, testable structure: units interact **within** groups (villages, schools, markets) but not across them, and only through the **share** of their group that is treated. Then

$$
Y_{ig} = Y_{ig}(t_{ig},\, s_g), \qquad s_g = \text{saturation (share treated in group } g).
$$

Effects at saturation $s$ (averages over units):

- **Direct effect:** $\text{DE}(s) = E[Y(1, s) - Y(0, s)]$ (own treatment, holding others' fixed).
- **Spillover (indirect) effect:** $\text{SE}(s) = E[Y(0, s) - Y(0, 0)]$ (on the untreated, of others being treated).
- **Total effect:** $E[Y(1, s) - Y(0, 0)] = \text{DE}(s) + \text{SE}(s)$.

### Designs

- **Cluster randomisation** (Day 40) puts the whole group in one arm, so spillovers stay inside: estimates the total policy effect.
- **Saturation (two-stage) designs** first randomise groups to saturations (0%, 50%, 100%…), then individuals within groups (Hudgens & Halloran 2008; Baird et al. 2018). Comparing untreated units in treated groups with pure controls identifies spillovers.

### Exercise 6 · SUTVA collapses the outcomes

Show that under no interference $Y_i(\mathbf t) = Y_i(t_i)$, and that consistency then gives $Y_i = T_iY_i(1) + (1 - T_i)Y_i(0)$.

<details><summary>Show solution</summary>

**Step 1.** No interference says $Y_i(\mathbf t) = Y_i(\mathbf t')$ whenever $t_i = t'_i$, so $Y_i(\mathbf t)$ depends on $\mathbf t$ only through $t_i$: write it $Y_i(t_i)$.

**Step 2.** Consistency says the observed outcome equals the potential outcome of the treatment received: $Y_i = Y_i(T_i)$.

**Step 3.** If $T_i = 1$, $Y_i = Y_i(1)$; if $T_i = 0$, $Y_i = Y_i(0)$. Both cases are $T_iY_i(1) + (1 - T_i)Y_i(0)$. $\blacksquare$

</details>

### Exercise 7 · Within-group randomisation identifies the direct effect

Assume partial interference with exposure $(t, s)$. In groups assigned saturation $s$, individuals are treated completely at random. Show $E[Y \mid T = 1, s] - E[Y \mid T = 0, s] = \text{DE}(s)$.

<details><summary>Show solution</summary>

**Step 1.** In a saturation-$s$ group, every unit has exposure $(T_i, s)$, so $Y_i = Y_i(T_i, s)$.

**Step 2.** Within those groups, $T_i$ is randomised, so it is independent of $(Y_i(1, s), Y_i(0, s))$.

**Step 3.** Hence $E[Y \mid T = 1, s] = E[Y(1, s)]$ and $E[Y \mid T = 0, s] = E[Y(0, s)]$ (Day 6, Exercise 14).

**Step 4.** The difference is $\text{DE}(s)$. $\blacksquare$

</details>

### Exercise 8 · Total = direct + spillover

Prove $E[Y(1, s) - Y(0, 0)] = \text{DE}(s) + \text{SE}(s)$.

<details><summary>Show solution</summary>

**Step 1.** Add and subtract $Y(0, s)$: $Y(1, s) - Y(0, 0) = [Y(1, s) - Y(0, s)] + [Y(0, s) - Y(0, 0)]$.

**Step 2.** Take expectations (linearity): $\text{DE}(s) + \text{SE}(s)$. $\blacksquare$

</details>

### Exercise 9 · Contaminated controls

Individuals are randomised within every village at saturation $s$, and the analyst treats the within-village difference in means as "the effect". Show it equals the total effect minus the spillover on controls.

<details><summary>Show solution</summary>

**Step 1.** By Exercise 7 the comparison estimates $E[Y(1, s)] - E[Y(0, s)]$.

**Step 2.** Rewrite using Exercise 8: $E[Y(1, s)] - E[Y(0, s)] = \underbrace{E[Y(1, s) - Y(0, 0)]}_{\text{total}} - \underbrace{E[Y(0, s) - Y(0, 0)]}_{\text{spillover on controls}}$.

**Answer:** positive spillovers (herd immunity) make the within-village comparison *understate* the policy's total benefit; negative spillovers (displacement) make it *overstate* it.

</details>

### Exercise 10 · Hidden versions average out

A "treatment" has two versions, $a$ and $b$, delivered with probabilities $\pi$ and $1 - \pi$ independently of potential outcomes. Show the difference in means estimates $\pi\,\text{ATE}_a + (1 - \pi)\,\text{ATE}_b$.

<details><summary>Show solution</summary>

**Step 1.** A treated unit's outcome is $Y_i(a)$ with probability $\pi$ and $Y_i(b)$ otherwise, so $E[Y \mid T = 1] = \pi E[Y(a)] + (1 - \pi)E[Y(b)]$.

**Step 2.** $E[Y \mid T = 0] = E[Y(0)]$ by randomisation.

**Step 3.** Difference: $\pi(E[Y(a)] - E[Y(0)]) + (1 - \pi)(E[Y(b)] - E[Y(0)])$. $\blacksquare$

The estimate is a mixture: it describes *this* blend of versions, not either version alone.

</details>

### Exercise 11 · Deworming and reinfection

In a school deworming trial, infection rates are: treated pupils in treatment schools 0.20; untreated pupils in treatment schools 0.35; pupils in control schools 0.50. Compute the direct, spillover and total effects, and what a within-school comparison would report.

<details><summary>Show solution</summary>

**Step 1.** Direct effect (within treatment schools): $0.20 - 0.35 = -0.15$.

**Step 2.** Spillover on untreated pupils: $0.35 - 0.50 = -0.15$.

**Step 3.** Total effect for treated pupils: $0.20 - 0.50 = -0.30$.

**Step 4.** A within-school comparison reports the direct effect, −0.15.

**Answer:** the within-school comparison captures only half the benefit; deworming also protects untreated classmates. This was a central finding of Miguel & Kremer (2004).

### Exercise 12 · Job-search assistance and displacement

Job-finding rates within 6 months: assisted jobseekers 0.40; unassisted jobseekers in labour markets where half were assisted 0.28; jobseekers in untreated markets 0.32. Compute the direct and spillover effects, and the net effect on the average jobseeker in a 50%-assisted market.

<details><summary>Show solution</summary>

**Step 1.** Direct effect: $0.40 - 0.28 = 0.12$.

**Step 2.** Spillover on the unassisted: $0.28 - 0.32 = -0.04$ (they lose jobs to assisted competitors).

**Step 4.** Net effect versus untreated markets: $0.34 - 0.32 = 0.02$.

**Answer:** the direct effect (12 points) vastly overstates the 2-point net gain: most of the "effect" is reshuffling who gets the jobs (as in Crépon et al., 2013).

When 80% of a town is vaccinated, the infection risk of *unvaccinated* residents falls from 0.10 to 0.04. The vaccine reduces a vaccinated person's risk to 0.01 at that coverage. Compute the indirect and direct effects at 80% coverage.

<details><summary>Show solution</summary>

## Verify in Python

*Your work:*

```python
import numpy as np

rng = np.random.default_rng(32)
G, m = 400, 20                                  # villages, households per village
sat = rng.choice([0.0, 0.5], G)                 # half the villages untreated, half 50% treated
DE, SPILL = 20.0, 8.0                           # direct effect and spillover at 50% saturation
rows = []
for g in range(G):
    T = np.zeros(m, bool)
    T[rng.choice(m, int(sat[g] * m), replace=False)] = True
    y = 92 + DE * T + SPILL * (sat[g] > 0) + rng.normal(0, 15, m)   # everyone in a treated village gets the spillover
    rows += [(sat[g], t, v) for t, v in zip(T, y)]
s, t, y = map(np.array, zip(*rows))
direct = y[(s == 0.5) & t].mean() - y[(s == 0.5) & ~t].mean()
spill = y[(s == 0.5) & ~t].mean() - y[s == 0].mean()
total = y[(s == 0.5) & t].mean() - y[s == 0].mean()
print(f"direct {direct:.1f} (truth 20), spillover {spill:.1f} (truth 8), total {total:.1f} (truth 28)")

# Job-search displacement (Exercise 12)
print("net effect in a 50% market:", round(0.5 * 0.40 + 0.5 * 0.28 - 0.32, 3))
```

## Key takeaways

- SUTVA = no interference + no hidden versions; without it, "the" effect is not even defined.
- With spillovers, distinguish direct, spillover and total effects; within-group comparisons capture only the direct effect.
- Saturation designs and cluster randomisation turn spillovers from a nuisance into an estimand.
