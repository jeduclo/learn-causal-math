Compare how outcomes changed in a treated group with how they changed in an untreated group. Under parallel trends, the difference in changes is the effect of treatment on the treated. The estimand, its regression form, and the assumptions that make or break it.

**Learning outcomes**

- Compute the 2×2 difference-in-differences estimate from four means and from a regression with an interaction.
- Prove that DiD identifies the ATT under parallel trends and no anticipation.
- Show that parallel trends depend on the scale of the outcome (levels versus logs).
- Apply DiD to minimum wages, health insurance expansions and training programmes, and spot failures such as the Ashenfelter dip.

## Theory

### Set-up

Two groups ($G = 1$ treated, $G = 0$ comparison) and two periods ($T = 0$ before, $T = 1$ after). Only the treated group in period 1 receives treatment. The target is the **ATT** in the post period:

$$
ATT = E[Y_{1}(1) - Y_{1}(0) \mid G = 1],
$$

where $Y_t(d)$ is the potential outcome in period $t$. The problem: $E[Y_1(0) \mid G = 1]$, what would have happened to the treated without treatment, is unobserved.

### Two naive comparisons

- **Before–after** in the treated group: $E[Y_1 - Y_0 \mid G = 1]$, biased by anything else that changed over time.
- **Treated–comparison** after: $E[Y_1 \mid G = 1] - E[Y_1 \mid G = 0]$, biased by permanent group differences.

### Parallel trends

**Assumption (parallel trends):** without treatment, the average outcome would have changed by the same amount in both groups:

$$
E[Y_1(0) - Y_0(0) \mid G = 1] = E[Y_1(0) - Y_0(0) \mid G = 0].
$$

**No anticipation:** $Y_0 = Y_0(0)$ for the treated (treatment does not affect pre-period outcomes).

### The DiD estimand

Under both assumptions,

$$
ATT = \big(E[Y_1 \mid G = 1] - E[Y_0 \mid G = 1]\big) - \big(E[Y_1 \mid G = 0] - E[Y_0 \mid G = 0]\big).
$$

The comparison group's change is the counterfactual change for the treated.

### Regression form

$$
Y_{it} = \alpha + \gamma G_i + \lambda T_t + \delta\,(G_i \times T_t) + \varepsilon_{it}.
$$

The four cell means are $\alpha$, $\alpha + \lambda$, $\alpha + \gamma$, $\alpha + \gamma + \lambda + \delta$; so $\hat\delta$ is exactly the DiD of sample means. With panel data, regressing $\Delta Y_i$ on $G_i$ gives the same number. Covariates can be added (for instance pre-treatment characteristics interacted with $T$), and standard errors should be clustered at the level at which treatment is assigned.

### Scale matters

Parallel trends in levels and in logs cannot both hold unless baseline levels are equal or there is no trend. The choice of scale is a substantive assumption, not a technicality.

### Threats

Different shocks to the two groups in the post period; selection into treatment based on transitory outcomes (the **Ashenfelter dip**: people join training after a temporary earnings drop, which would have recovered anyway); compositional changes in repeated cross-sections; spillovers to the comparison group.

### Exercise 6 · DiD identifies the ATT

Prove that under parallel trends and no anticipation the DiD estimand equals the ATT.

<details><summary>Show solution</summary>

**Step 1.** Observed outcomes: $E[Y_1 \mid G = 1] = E[Y_1(1) \mid G = 1]$; $E[Y_0 \mid G = 1] = E[Y_0(0) \mid G = 1]$ (no anticipation); comparison outcomes are untreated potential outcomes.

**Step 2.** DiD $= E[Y_1(1) - Y_0(0) \mid G = 1] - E[Y_1(0) - Y_0(0) \mid G = 0]$.

**Step 3.** By parallel trends, replace the second term by $E[Y_1(0) - Y_0(0) \mid G = 1]$.

**Step 4.** DiD $= E[Y_1(1) - Y_1(0) \mid G = 1] = ATT$. $\blacksquare$

</details>

### Exercise 7 · The interaction coefficient

Prove that OLS of the saturated regression $Y = \alpha + \gamma G + \lambda T + \delta GT + \varepsilon$ gives $\hat\delta = (\bar Y_{11} - \bar Y_{10}) - (\bar Y_{01} - \bar Y_{00})$, with $\bar Y_{gt}$ the cell means.

<details><summary>Show solution</summary>

**Step 1.** The regression has four parameters and four cells: it is saturated, so fitted values equal cell means.

**Step 2.** Fitted values: $\bar Y_{00} = \hat\alpha$, $\bar Y_{01} = \hat\alpha + \hat\lambda$, $\bar Y_{10} = \hat\alpha + \hat\gamma$, $\bar Y_{11} = \hat\alpha + \hat\gamma + \hat\lambda + \hat\delta$.

**Step 3.** Solve: $\hat\delta = \bar Y_{11} - \bar Y_{10} - \bar Y_{01} + \bar Y_{00}$. $\blacksquare$

</details>

### Exercise 10 · Scale dependence

Suppose untreated outcomes grow by the same proportion $g > 0$ in both groups and baseline means differ ($\mu_1 \ne \mu_0$). Show that parallel trends in levels fails.

<details><summary>Show solution</summary>

**Step 1.** Untreated change for group $k$: $\mu_k(1 + g) - \mu_k = g\mu_k$.

**Step 2.** Difference between groups: $g(\mu_1 - \mu_0) \ne 0$.

**Answer:** level trends differ by $g(\mu_1 - \mu_0)$, which is exactly the level-DiD bias. $\blacksquare$

</details>

### Exercise 15 · The Ashenfelter dip

Trainees' earnings were \$20,000 two years before training, \$15,000 the year before (after job losses) and \$19,000 the year after. Comparison workers earned \$20,000, \$19,800 and \$20,000. Compute DiD using the year before as baseline, then using two years before. Which is more credible?

<details><summary>Show solution</summary>

**Step 1.** Baseline = year before: treated $+4{,}000$; comparison $+200$; DiD $= 3{,}800$.

## Verify in Python

*Your work:*

```python
import numpy as np

did = lambda t0, t1, c0, c1: (t1 - t0) - (c1 - c0)
print("Ex 1:", did(40, 52, 35, 41), " NJ:", round(did(20.44, 21.03, 23.33, 21.17), 2), " Medicaid:", did(320, 310, 330, 328))
print("logs vs levels:", did(20, 25, 10, 12), round(np.log(25 / 20) - np.log(12 / 10), 4))
print("bed nets placebo:", did(150, 140, 160, 155), " main:", did(140, 100, 155, 140))
print("Ashenfelter:", did(15_000, 19_000, 19_800, 20_000), did(20_000, 19_000, 20_000, 20_000))

# Regression form with simulated restaurants: the interaction equals the DiD of means
rng = np.random.default_rng(67)
n = 400; G = np.repeat([0, 1], n // 2)
y0 = 23 - 3 * G + rng.normal(0, 8, n)                      # before
y1 = y0 - 2 + 2.75 * G + rng.normal(0, 4, n)                # after: common trend -2, effect 2.75
Y = np.r_[y0, y1]; T = np.r_[np.zeros(n), np.ones(n)]; GG = np.r_[G, G]
X = np.column_stack([np.ones(2 * n), GG, T, GG * T])
b = np.linalg.lstsq(X, Y, rcond=None)[0]
means = did(y0[G == 1].mean(), y1[G == 1].mean(), y0[G == 0].mean(), y1[G == 0].mean())
fd = np.polyfit(G, y1 - y0, 1)[0]
print(f"interaction {b[3]:.3f}  DiD of means {means:.3f}  first-difference regression {fd:.3f}")
```

## Key takeaways

- DiD = (change for the treated) − (change for the comparison group); it identifies the ATT under parallel trends and no anticipation.
- The regression interaction, the first-difference regression and the four-means formula are the same estimate.
- Parallel trends is scale-dependent and fails with transitory pre-treatment shocks such as the Ashenfelter dip.
