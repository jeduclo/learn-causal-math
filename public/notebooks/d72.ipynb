Two extensions of DiD. Triple differences add a within-state comparison group that is not targeted by the policy, so state-specific shocks cancel. Changes-in-changes drops additivity altogether and maps ranks across distributions, giving effects on whole outcome distributions that do not depend on the outcome's scale.

## Theory

### Triple differences

DiD fails if the treated state has its own shock in the post period. If the policy targets a **subgroup** (married women of child-bearing age; low-income adults), we can use an untargeted subgroup in the same states to absorb state shocks:

$$
DDD = \underbrace{\big(\Delta\bar Y_{T, A} - \Delta\bar Y_{C, A}\big)}_{\text{DiD, affected group}} - \underbrace{\big(\Delta\bar Y_{T, B} - \Delta\bar Y_{C, B}\big)}_{\text{DiD, unaffected group}},
$$

where $T, C$ index treated and control states, $A, B$ the affected and unaffected subgroups, and $\Delta$ the before–after change.

**Assumption:** the bias of the DiD (non-parallel trends) is the same for the two subgroups. DDD does not need two separate parallel-trends assumptions; it needs only that the *difference* between subgroups would have evolved in parallel across states (Olden and Møen).

**Regression:** a saturated regression with state-group, period and subgroup indicators and all their interactions; the coefficient on the triple interaction $T \times \text{Post} \times A$ is the DDD.

### Changes-in-changes (Athey and Imbens)

DiD assumes additivity: the counterfactual shift is the same in levels for everyone. CiC instead assumes

- untreated outcomes are $Y(0) = h(U, t)$ with $h$ strictly increasing in an unobserved "rank" $U$;
- the distribution of $U$ within each group does not change over time (it may differ between groups).

Then a treated unit at rank $q$ in its group's pre-period distribution would have the outcome of the control-group rank that the same $U$ maps to. The counterfactual distribution of the treated in the post period is

$$
F_{Y(0), 11}(y) = F_{10}\Big(F_{00}^{-1}\big(F_{01}(y)\big)\Big),
$$

where $F_{gt}$ is the observed distribution in group $g$, period $t$. Equivalently, the counterfactual for a treated pre-period outcome $y$ is $F_{01}^{-1}(F_{00}(y))$: find $y$'s rank among control units before, then the control outcome at that rank after.

### Why CiC

CiC gives **quantile treatment effects** (effects at the median, at the 10th percentile), not just a mean, and is invariant to monotone transformations of the outcome: levels and logs give the same answer, resolving Day 67's scale problem. It requires continuous outcomes (or bounds for discrete ones) and support overlap.

### Exercise 8 · The CiC counterfactual

Assume $Y(0) = h(U, t)$, $h$ strictly increasing in $U$, and $U \mid G = g$ has the same distribution in both periods. Prove $F_{Y(0), 11}(y) = F_{10}(F_{00}^{-1}(F_{01}(y)))$ (continuous distributions).

<details><summary>Show solution</summary>

### Exercise 10 · When CiC equals DiD

Suppose $h(u, t) = u + \delta_t$ (an additive time shift). Show the CiC counterfactual mean equals the DiD counterfactual.

<details><summary>Show solution</summary>

**Step 1.** Controls' distribution shifts by $\delta_1 - \delta_0$: $F_{01}^{-1}(F_{00}(y)) = y + \delta_1 - \delta_0$.

**Step 2.** Each treated unit's counterfactual is its pre-period outcome plus $\delta_1 - \delta_0$; averaging gives $\bar Y_{10} + (\bar Y_{01} - \bar Y_{00})$.

**Step 3.** That is the DiD counterfactual. $\blacksquare$

</details>

## Verify in Python

*Your work:*

```python
import numpy as np

# Exercises 1 and 12
ddd = lambda tA, cA, tB, cB: (tA - cA) - (tB - cB)
print("maternity DDD:", ddd(-5, 1, -1, 2), " Medicaid DDD:", ddd(-9, -4, -2, -1))

# Exercise 4: one CiC counterfactual
c0 = np.array([5, 8, 12, 15.]); c1 = np.array([6, 10, 15, 20.])
q = (c0 <= 12).mean(); print("rank", q, "-> counterfactual", np.quantile(c1, q, method="inverted_cdf"), " DiD:", 12 + c1.mean() - c0.mean())

# Exercise 13: changes-in-changes on simulated earnings
rng = np.random.default_rng(72); n = 20_000
U0, U1 = rng.normal(0, 1, n), rng.normal(.5, 1, n)                 # controls, treated (different rank distributions)
h = lambda u, t: np.exp(2.5 + .4 * u + .15 * t * (1 + .3 * u))      # nonlinear, increasing in u
y00, y01 = h(U0, 0), h(rng.normal(0, 1, n), 1)                       # controls before / after (fresh draws, same distribution)
y10 = h(U1, 0); U1b = rng.normal(.5, 1, n)
y11_0 = h(U1b, 1); y11 = y11_0 * (1 + .05 + .1 * (U1b - .5))         # treated after, effect grows with u
cf = np.quantile(y01, np.searchsorted(np.sort(y00), y10) / n)        # CiC counterfactuals for treated pre-period units
did = y11.mean() - (y10.mean() + y01.mean() - y00.mean())
print(f"DiD {did:.2f}  CiC mean effect {y11.mean() - cf.mean():.2f}  true mean effect {(y11 - y11_0).mean():.2f}")
for p in (.1, .5, .9):
    print(f"q{int(p * 100)}: CiC {np.quantile(y11, p) - np.quantile(cf, p):.2f}  truth {np.quantile(y11, p) - np.quantile(y11_0, p):.2f}")
```

## Key takeaways

- DDD subtracts the DiD of an untargeted subgroup in the same states; it needs only that their difference would have trended in parallel.
- CiC replaces additive parallel trends with stable rank distributions, giving quantile effects and scale invariance.
- Neither can separate policies that coincide in every dimension used for comparison.
