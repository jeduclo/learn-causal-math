## Theory

### The propensity score as a balancing score

$e(x) = P(D = 1 \mid X = x)$. Rosenbaum and Rubin (1983) showed:

1. **Balancing:** $D \perp\!\!\!\perp X \mid e(X)$: within groups of equal propensity, covariates are distributed the same way among treated and controls.
2. **Sufficiency:** if $(Y(0), Y(1)) \perp\!\!\!\perp D \mid X$, then $(Y(0), Y(1)) \perp\!\!\!\perp D \mid e(X)$.

So adjusting for the one-dimensional $e(X)$, by stratification (Day 43), matching or weighting, is enough.

### Matching on the score

Match each treated unit to the control(s) with the closest $\hat e$ (often on the logit scale, with a caliper). Matching on an estimated score needs its own variance formula (Abadie and Imbens, 2016). King and Nielsen caution that pruning by the score can increase imbalance in individual covariates; always check balance.

### Weights and their targets

| Weights (treated, control) | Estimand |
|---|---|
| $1/e,\; 1/(1 - e)$ | ATE |
| $1,\; e/(1 - e)$ | ATT |
| $(1 - e)/e,\; 1$ | ATC |
| $1 - e,\; e$ (overlap weights) | ATO: effect in the population with most overlap |

Normalise weights within each arm (Hájek, Day 44). **Stabilised weights** multiply by the marginal probability of the arm, keeping the mean weight near 1. **Overlap weights** are bounded in $[0, 1]$, so they never explode; they target

$$
ATO = \frac{E[e(X)(1 - e(X))\tau(X)]}{E[e(X)(1 - e(X))]},
$$

which emphasises units that could plausibly be in either group.

### Exact balance

With a logistic propensity model including $X$, overlap-weighted means of every covariate in the model are **exactly equal** in the two arms (Li, Morgan and Zaslavsky, 2018), a direct consequence of the logistic score equations (Exercise 10).

### Estimated beats true

Using the estimated score in IPW gives a smaller asymptotic variance than using the true score (Hirano, Imbens and Ridder, 2003): estimating it removes chance imbalances in the sample.

### Exercise 6 · The balancing property

Prove $P(D = 1 \mid X, e(X)) = e(X)$, and hence $D \perp\!\!\!\perp X \mid e(X)$.

<details><summary>Show solution</summary>

**Step 1.** $e(X)$ is a function of $X$, so $P(D = 1 \mid X, e(X)) = P(D = 1 \mid X) = e(X)$.

**Step 2.** By the LIE, $P(D = 1 \mid e(X)) = E[P(D = 1 \mid X) \mid e(X)] = E[e(X) \mid e(X)] = e(X)$.

**Step 3.** So $P(D = 1 \mid X, e(X)) = P(D = 1 \mid e(X))$: given $e(X)$, $X$ carries no further information about $D$. $\blacksquare$

</details>

*Your work:*

### Exercise 7 · Sufficiency of the score

Assume $Y(d) \perp\!\!\!\perp D \mid X$. Prove $P(D = 1 \mid Y(d), e(X)) = e(X)$.

<details><summary>Show solution</summary>

**Step 1.** $P(D = 1 \mid Y(d), e(X)) = E[P(D = 1 \mid Y(d), X) \mid Y(d), e(X)]$ (LIE).

**Step 2.** By unconfoundedness, $P(D = 1 \mid Y(d), X) = e(X)$.

**Step 3.** So the expression is $E[e(X) \mid Y(d), e(X)] = e(X)$, which does not depend on $Y(d)$. $\blacksquare$

</details>

### Exercise 8 · ATT weights

Prove $\dfrac{E\big[(1 - D)\frac{e(X)}{1 - e(X)}Y\big]}{E\big[(1 - D)\frac{e(X)}{1 - e(X)}\big]} = E[Y(0) \mid D = 1]$.

<details><summary>Show solution</summary>

**Step 1.** Numerator, by the LIE and unconfoundedness: $E\big[\frac{e}{1 - e}E[(1 - D)Y(0) \mid X]\big] = E\big[\frac{e}{1 - e}(1 - e)\mu_0(X)\big] = E[e(X)\mu_0(X)]$.

**Step 2.** Denominator: $E[e(X)] = P(D = 1)$.

**Step 3.** $E[e(X)\mu_0(X)]/P(D = 1) = E[\mu_0(X) \mid D = 1]$ (since $e(X)/P(D = 1)$ is the density ratio of $X$ among the treated) $= E[Y(0) \mid D = 1]$. $\blacksquare$

</details>

### Exercise 9 · What overlap weights estimate

Prove that the overlap-weighted difference in means converges to the ATO.

<details><summary>Show solution</summary>

**Step 1.** Treated part: $E[D(1 - e)Y]/E[D(1 - e)] = E[e(1 - e)\mu_1(X)]/E[e(1 - e)]$ (condition on $X$; $E[D \mid X] = e$).

**Step 2.** Control part: $E[(1 - D)eY]/E[(1 - D)e] = E[e(1 - e)\mu_0(X)]/E[e(1 - e)]$.

**Step 3.** Subtract: $E[e(1 - e)\tau(X)]/E[e(1 - e)]$. $\blacksquare$

</details>

### Exercise 10 · Exact balance from logistic regression

Let $\hat e_i$ be logistic-regression fitted values with covariates $(1, X)$. The MLE satisfies $\sum_i(D_i - \hat e_i)(1, X_i) = \mathbf 0$. Show that overlap-weighted covariate means are identical in the two arms.

<details><summary>Show solution</summary>

**Step 1.** Split $\hat e_i = D_i\hat e_i + (1 - D_i)\hat e_i$: $\sum_i D_i(1 - \hat e_i)X_i = \sum_i(1 - D_i)\hat e_iX_i$.

**Step 2.** The same with the constant: $\sum_i D_i(1 - \hat e_i) = \sum_i(1 - D_i)\hat e_i$.

**Step 3.** Divide: the overlap-weighted mean of $X$ among treated equals that among controls. $\blacksquare$

</details>

```python
import numpy as np

# Exercises 4 and 11
print("PS matching ATT:", ((11 - 8) + (7 - 6)) / 2)
e_c = np.array([.8, .5, .2, .2]); y_c = np.array([7, 9, 5, 6.])
w = e_c / (1 - e_c)
print("ATT weights:", w, " weighted control mean:", round(w @ y_c / w.sum(), 3), " ATT:", round(11 - w @ y_c / w.sum(), 3))

rng = np.random.default_rng(82)
def logit_fit(X, d, iters=25):
    b = np.zeros(X.shape[1])
    for _ in range(iters):
        p = 1 / (1 + np.exp(-X @ b)); W = p * (1 - p)
        b += np.linalg.solve(X.T @ (W[:, None] * X), X.T @ (d - p))
    return 1 / (1 + np.exp(-X @ b))

# Exercise 10: exact balance with overlap weights
n = 2000; X = rng.normal(size=(n, 2)); d = (rng.random(n) < 1 / (1 + np.exp(-(X @ [1.5, -1])))).astype(float)
eh = logit_fit(np.column_stack([np.ones(n), X]), d)
bal = lambda wt, wc: ((d * wt)[:, None] * X).sum(0) / (d * wt).sum() - (((1 - d) * wc)[:, None] * X).sum(0) / ((1 - d) * wc).sum()
print("covariate mean gaps, overlap weights:", bal(1 - eh, eh).round(12), "  IPW:", bal(1 / eh, 1 / (1 - eh)).round(4))

# Exercises 13-14
res = {k: [] for k in ["PS matching", "IPW (estimated e)", "IPW (true e)", "overlap weights"]}
for _ in range(400):
    X = rng.normal(size=(n, 2)); lin = 2.2 * X[:, 0] - 1.2 * X[:, 1]; e = 1 / (1 + np.exp(-lin))
    d = (rng.random(n) < e).astype(float); y = 1 + X @ [1, 1] + 2 * d + rng.normal(size=n)
    eh = logit_fit(np.column_stack([np.ones(n), X]), d)
    hajek = lambda p: (d * y / p).sum() / (d / p).sum() - ((1 - d) * y / (1 - p)).sum() / ((1 - d) / (1 - p)).sum()
    res["IPW (estimated e)"].append(hajek(eh)); res["IPW (true e)"].append(hajek(e))
    res["overlap weights"].append((d * (1 - eh) * y).sum() / (d * (1 - eh)).sum() - ((1 - d) * eh * y).sum() / ((1 - d) * eh).sum())
    lt, lc = np.log(eh / (1 - eh))[d == 1], np.log(eh / (1 - eh))[d == 0]
    order = np.argsort(lc); pos = np.clip(np.searchsorted(lc[order], lt), 1, len(lc) - 1)
    near = np.where(np.abs(lc[order][pos] - lt) < np.abs(lc[order][pos - 1] - lt), order[pos], order[pos - 1])
    res["PS matching"].append((y[d == 1] - y[d == 0][near]).mean())
for k, v in res.items():
    print(f"{k:18s} mean {np.mean(v):.3f}  SD {np.std(v):.3f}")
```

## Key takeaways

- The propensity score balances covariates and is sufficient for adjustment under unconfoundedness.
- Different weights target different populations: ATE, ATT, ATC, and the overlap population (ATO), whose weights are bounded and give exact mean balance with a logistic score.
- Use the estimated score, check balance, and choose the estimand deliberately when overlap is poor.
