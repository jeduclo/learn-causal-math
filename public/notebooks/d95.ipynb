# Day 95: Sensitivity Analysis: How Strong Would Hidden Confounding Have to Be?

*Beginner to Ninja: Math for Causal Inference*

Every observational estimate rests on no unmeasured confounding, an assumption the data cannot test. Sensitivity analysis replaces a yes/no assumption with a quantitative question: how strong would an omitted confounder have to be to change our conclusion? We cover the omitted-variable-bias formula in partial R² form (Cinelli and Hazlett), the robustness value, the E-value for risk ratios, and Rosenbaum bounds for matched designs.

**Learning outcomes**

- Derive the omitted-variable bias of OLS in terms of partial R² values and compute the robustness value.
- Compute E-values for risk ratios and their confidence limits.
- Carry out a Rosenbaum sensitivity analysis for matched pairs with the sign test.
- Benchmark hypothetical confounders against observed covariates and report sensitivity transparently.

## Theory

### Omitted-variable bias, partial-R² form

Let the "long" regression be $Y = \tau D + X'\beta + \gamma U + \varepsilon$ and the "short" regression omit $U$. The bias of the short OLS coefficient is $\hat\tau_s - \hat\tau_l = \hat\gamma\hat\delta$, with $\hat\delta$ the coefficient of $D$ in a regression of $U$ on $D, X$. Cinelli and Hazlett (2020) rewrite it as

$$
|\widehat{\text{bias}}| = \sqrt{\frac{R^2_{Y\sim U\mid D,X}\;R^2_{D\sim U\mid X}}{1 - R^2_{D\sim U\mid X}}}\cdot\frac{\mathrm{sd}(Y^{\perp D,X})}{\mathrm{sd}(D^{\perp X})},
$$

where $R^2_{Y\sim U\mid D,X}$ is the share of residual outcome variation explained by $U$, and $R^2_{D\sim U\mid X}$ the share of residual treatment variation explained by $U$. These are interpretable, and can be **benchmarked** against observed covariates ("a confounder as strong as income").

### The robustness value

The **robustness value** $RV$ is the common strength $R^2_{Y\sim U\mid D,X} = R^2_{D\sim U\mid X} = RV$ that would bring the estimate to zero:

$$
RV = \tfrac12\Big(\sqrt{f^4 + 4f^2} - f^2\Big), \qquad f = \frac{|t_D|}{\sqrt{\mathrm{df}}},
$$

with $t_D$ the $t$-statistic of $D$ in the short regression. Confounders weaker than $RV$ on both sides cannot explain the effect away.

### The E-value

For an observed risk ratio $RR > 1$, the **E-value** (VanderWeele and Ding) is

$$
E = RR + \sqrt{RR(RR - 1)}:
$$

the minimum strength of association, on the risk-ratio scale, that a confounder would need with both treatment and outcome to explain the estimate away. For $RR < 1$ use $1/RR$; report it for the confidence limit closest to 1 as well.

### Rosenbaum bounds

In matched pairs, hidden bias may make the odds of treatment differ by up to a factor $\Gamma$ within a pair. Under the null of no effect, the number $T$ of discordant pairs in which the treated unit has the higher outcome is then bounded by $\text{Binomial}(S, \Gamma/(1 + \Gamma))$, giving an upper bound on the p-value:

$$
p^+(\Gamma) = P\Big(\text{Bin}\big(S, \tfrac{\Gamma}{1 + \Gamma}\big) \ge T\Big).
$$

The **sensitivity value** $\Gamma^*$ is the $\Gamma$ at which $p^+$ reaches 0.05.

### Exercise 6 · Omitted-variable bias

Using the Frisch–Waugh–Lovell theorem, show that the short coefficient equals $\hat\tau_l + \hat\gamma\hat\delta$.

<details><summary>Show solution</summary>

**Step 3.** So $\hat\tau_s = \hat\tau_l + \hat\gamma\hat\delta$. $\blacksquare$

### Exercise 7 · The partial-R² form

Show $|\hat\gamma\hat\delta| = \sqrt{R^2_{Y\sim U\mid D,X}R^2_{D\sim U\mid X}/(1 - R^2_{D\sim U\mid X})}\cdot\mathrm{sd}(Y^{\perp D,X})/\mathrm{sd}(D^{\perp X})$.

<details><summary>Show solution</summary>

**Step 1.** By FWL, $\hat\gamma = \mathrm{cov}(Y^{\perp}, U^{\perp})/\mathrm{var}(U^{\perp})$ with residuals on $(D, X)$, so $|\hat\gamma| = \sqrt{R^2_{Y\sim U\mid D,X}}\,\mathrm{sd}(Y^{\perp D,X})/\mathrm{sd}(U^{\perp D,X})$.

**Step 2.** Likewise $|\hat\delta| = |\mathrm{cov}(U^{\perp X}, D^{\perp X})|/\mathrm{var}(D^{\perp X}) = \sqrt{R^2_{D\sim U\mid X}}\,\mathrm{sd}(U^{\perp X})/\mathrm{sd}(D^{\perp X})$.

**Step 3.** $\mathrm{var}(U^{\perp D,X}) = \mathrm{var}(U^{\perp X})(1 - R^2_{D\sim U\mid X})$ (removing $D^{\perp X}$ from $U^{\perp X}$).

**Step 4.** Multiply: the $\mathrm{sd}(U^{\perp X})$ terms cancel, leaving the formula. $\blacksquare$

</details>

### Exercise 8 · The robustness value

Show that the partial $R^2$ of $D$ with $Y$ given $X$ is $R^2_{Y\sim D\mid X} = t^2/(t^2 + \mathrm{df})$, i.e. $f^2 = R^2/(1 - R^2)$, and that setting both confounder partial $R^2$'s equal to $RV$ and $|\text{bias}| = |\hat\tau_s|$ yields $RV^2/(1 - RV) = f^2$, whose root is the formula given.

<details><summary>Show solution</summary>

**Step 1.** $t^2 = \hat\tau^2/\widehat{\mathrm{se}}^2 = \mathrm{df}\cdot R^2/(1 - R^2)$, a standard identity for a single coefficient, so $f^2 = t^2/\mathrm{df} = R^2/(1 - R^2)$.

**Step 2.** $|\hat\tau_s| = \sqrt{R^2_{Y\sim D\mid X}/(1 - R^2_{Y\sim D\mid X})}\cdot\mathrm{sd}(Y^{\perp D,X})/\mathrm{sd}(D^{\perp X}) = f\cdot\mathrm{sd}(Y^{\perp D,X})/\mathrm{sd}(D^{\perp X})$.

**Step 3.** Setting bias equal to the estimate with both $R^2 = RV$: $\sqrt{RV^2/(1 - RV)} = f$, so $RV^2 + f^2RV - f^2 = 0$.

**Step 4.** The positive root is $RV = \tfrac12(\sqrt{f^4 + 4f^2} - f^2)$. $\blacksquare$

</details>

### Exercise 9 · The E-value

The bounding factor of Ding and VanderWeele is $B = \frac{RR_{EU}RR_{UD}}{RR_{EU} + RR_{UD} - 1}$. Show that with $RR_{EU} = RR_{UD} = E$, $B = RR$ is solved by $E = RR + \sqrt{RR(RR - 1)}$.

<details><summary>Show solution</summary>

**Step 1.** $E^2/(2E - 1) = RR$, so $E^2 - 2RR\,E + RR = 0$.

**Step 2.** $E = RR \pm\sqrt{RR^2 - RR}$; take the root above 1. $\blacksquare$

</details>

### Exercise 10 · Rosenbaum's bound

In a discordant pair, under the null and with hidden bias at most $\Gamma$, show the probability that the treated unit is the one with the higher outcome lies in $[1/(1 + \Gamma), \Gamma/(1 + \Gamma)]$, so the count $T$ is stochastically below $\text{Bin}(S, \Gamma/(1 + \Gamma))$.

<details><summary>Show solution</summary>

**Step 1.** Under the null, outcomes are fixed; which pair member is treated decides whether "treated is higher".

**Step 2.** The odds of treatment within the pair are between $1/\Gamma$ and $\Gamma$, so the probability that a given unit is the treated one lies in $[1/(1 + \Gamma), \Gamma/(1 + \Gamma)]$.

**Step 3.** Pairs are independent, so $T$ is a sum of independent Bernoullis with probabilities at most $\Gamma/(1 + \Gamma)$, stochastically dominated by the binomial. $\blacksquare$

</details>

## Verify in Python

*Your work:*

```python
import numpy as np
from scipy.stats import binom
from scipy.optimize import brentq

rng = np.random.default_rng(95)
n = 5000; X = rng.normal(size=n); U = rng.normal(size=n)              # U: unobserved deprivation
D = .5 * X + .4 * U + rng.normal(size=n)                              # PM2.5 exposure
Y = -.3 * D + .6 * X - .5 * U + rng.normal(size=n)                    # lung-function score
def fit(cols, y):
    Z = np.column_stack([np.ones(n)] + cols); b = np.linalg.lstsq(Z, y, rcond=None)[0]; return b, y - Z @ b

b_s, rY = fit([D, X], Y); b_l, rYu = fit([D, X, U], Y)
_, rD = fit([X], D); _, rDu = fit([X, U], D)
R2yu, R2du = 1 - rYu.var() / rY.var(), 1 - rDu.var() / rD.var()
print(f"short {b_s[1]:.3f}  long {b_l[1]:.3f}  bias {b_s[1] - b_l[1]:.4f}")
print(f"R2(Y~U|D,X) {R2yu:.3f}  R2(D~U|X) {R2du:.3f}  formula |bias| {np.sqrt(R2yu * R2du / (1 - R2du)) * rY.std() / rD.std():.4f}")

df = n - 3; t = b_s[1] / np.sqrt(rY @ rY / df / (rD @ rD)); f = abs(t) / np.sqrt(df)
RV = .5 * (np.sqrt(f ** 4 + 4 * f ** 2) - f ** 2)
print(f"t {t:.2f}  RV {RV:.3f}  bias of an RV-strength confounder {np.sqrt(RV ** 2 / (1 - RV)) * rY.std() / rD.std():.3f} vs |estimate| {abs(b_s[1]):.3f}")
f4 = .2; print("Ex 4 RV:", round(.5 * (np.sqrt(f4 ** 4 + 4 * f4 ** 2) - f4 ** 2), 3))

ev = lambda rr: rr + np.sqrt(rr * (rr - 1))
print("E-values:", round(ev(2.0), 3), round(ev(1.5), 3), round(ev(1 / .7), 3))

S, T = 100, 72                                                        # Rosenbaum sign test
for G in (1, 1.5, 2, 2.5): print(f"Gamma {G}: p+ = {binom.sf(T - 1, S, G / (1 + G)):.4f}")
print("Gamma* =", round(brentq(lambda G: binom.sf(T - 1, S, G / (1 + G)) - .05, 1, 3), 3))
```

## Key takeaways

- Unmeasured confounding cannot be tested, but its required strength can be quantified.
- The partial-R² bias formula and robustness value express sensitivity in terms that can be benchmarked against observed covariates.
- E-values (risk ratios) and Rosenbaum's Γ (matched designs) answer the same question on other scales; report them for the estimate and the confidence limit.
