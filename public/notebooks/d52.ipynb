# Day 52: The Back-Door Criterion

*Beginner to Ninja: Math for Causal Inference*

A graphical test for which covariates to adjust for. A set Z is admissible if it contains no descendant of the treatment and blocks every path that enters the treatment through an arrow pointing into it; then the adjustment formula identifies the causal effect.

**Learning outcomes**

- Identify back-door paths and test whether a set satisfies the back-door criterion.
- Find all minimal admissible adjustment sets in small DAGs.
- Compute causal effects with the back-door adjustment formula and its weighting form.
- Connect the criterion to Phase 3's unconfoundedness and to regression adjustment.

## Theory

### Back-door paths

A **back-door path** from $X$ to $Y$ is a path that starts with an arrow **into** $X$: $X \leftarrow \cdots Y$. Directed paths $X \to \cdots \to Y$ carry the effect; back-door paths carry confounding.

### The criterion

A set of variables $\mathbf Z$ satisfies the **back-door criterion** relative to $(X, Y)$ if

1. no node in $\mathbf Z$ is a descendant of $X$, and
2. $\mathbf Z$ blocks every back-door path from $X$ to $Y$ (in the sense of d-separation).

### The back-door adjustment formula

If $\mathbf Z$ satisfies the criterion, the causal effect is identified:

$$
P(y \mid \operatorname{do}(x)) = \sum_{\mathbf z} P(y \mid x, \mathbf z)\,P(\mathbf z).
$$

Equivalently, in weighting form,

$$
P(y \mid \operatorname{do}(x)) = \sum_{\mathbf z} \frac{P(x, y, \mathbf z)}{P(x \mid \mathbf z)},
$$

which is IPW (Day 44) with propensity $P(x \mid \mathbf z)$.

### Why the two conditions?

- Condition 2 removes all non-causal association.
- Condition 1 keeps the causal paths intact: conditioning on a mediator would block part of the effect; conditioning on a descendant of $X$ that is also affected by $Y$'s causes could open new paths.

### Link to potential outcomes

In SCM terms, $\mathbf Z$ satisfying the back-door criterion implies **conditional ignorability**, $Y(x) \perp\!\!\!\perp X \mid \mathbf Z$ (generalising Day 51, Exercise 9). The criterion is a way to *check* Phase 3's unconfoundedness assumption from a graph, and to choose which $X$ variables to include.

### Several valid sets

There may be many admissible sets, or none among the observed variables. The parents of $X$ always qualify if they are observed. Different admissible sets give the same estimand but different precision (Day 53).

### Linear models

If $E[Y \mid x, \mathbf z]$ is linear, $E[Y \mid x, \mathbf z] = a + bx + \mathbf c^\top\mathbf z$, then by the adjustment formula $E[Y \mid \operatorname{do}(x)] = a + bx + \mathbf c^\top E[\mathbf Z]$: the regression coefficient $b$ is the causal effect when $\mathbf Z$ is admissible.

### Exercise 6 · Deriving the adjustment formula

Assume $\mathbf Z$ satisfies the back-door criterion. Prove $P(y \mid \operatorname{do}(x)) = \sum_{\mathbf z} P(y \mid x, \mathbf z)P(\mathbf z)$, using two facts: (a) interventions do not change the distribution of non-descendants of $X$ (Day 51), and (b) if $\mathbf Z$ blocks all back-door paths, then $P(y \mid \operatorname{do}(x), \mathbf z) = P(y \mid x, \mathbf z)$ (proved formally on Day 55).

<details><summary>Show solution</summary>

**Step 1.** Law of total probability in the intervened model: $P(y \mid \operatorname{do}(x)) = \sum_{\mathbf z} P(y \mid \operatorname{do}(x), \mathbf z)\,P(\mathbf z \mid \operatorname{do}(x))$.

**Step 2.** $\mathbf Z$ contains no descendant of $X$ (condition 1), so by (a) $P(\mathbf z \mid \operatorname{do}(x)) = P(\mathbf z)$.

**Step 3.** By (b), $P(y \mid \operatorname{do}(x), \mathbf z) = P(y \mid x, \mathbf z)$: once $\mathbf Z$ blocks the back-door paths, the only remaining association between $X$ and $Y$ is causal, so observing $x$ and setting $x$ are the same.

**Step 4.** Substitute. $\blacksquare$

### Exercise 8 · The weighting form

Prove $\sum_{\mathbf z} P(y \mid x, \mathbf z)P(\mathbf z) = \sum_{\mathbf z} P(x, y, \mathbf z)/P(x \mid \mathbf z)$.

<details><summary>Show solution</summary>

**Step 3.** Sum over $\mathbf z$. $\blacksquare$

### Exercise 10 · Why descendants are excluded

In the chain $X \to M \to Y$ with $M := aX + U_M$ and $Y := bM + U_Y$, there are no back-door paths, so $\{M\}$ blocks all of them. Show that "adjusting" for $M$ gives the wrong answer.

<details><summary>Show solution</summary>

**Step 1.** True effect: $E[Y \mid \operatorname{do}(x)] = abx$, slope $ab$.

**Step 2.** $E[Y \mid x, m] = bm$ (given $M$, $X$ adds nothing: chain).

**Step 3.** "Adjusted" formula: $\sum_m E[Y \mid x, m]P(m) = bE[M]$, which does not depend on $x$: slope 0.

**Answer:** condition 1 is necessary in general; conditioning on a mediator removes the effect it carries. $\blacksquare$

### Exercise 13 · Breastfeeding and child health

A DAG: mother's education $E \to$ income $I$, $E \to$ breastfeeding $B$, $I \to B$, $I \to$ child health $H$, $B \to$ infections $N \to H$, and $B \to$ clinic visits $C \leftarrow H$. Which minimal set identifies the effect of breastfeeding on child health? Which candidate controls must be excluded, and why?

**Step 1.** Back-door paths from $B$: $B \leftarrow I \to H$ and $B \leftarrow E \to I \to H$.

**Step 2.** $\{I\}$ blocks both (fork on the first, chain on the second): minimal admissible set. $\{E, I\}$ is also admissible; $\{E\}$ alone leaves $B \leftarrow I \to H$ open.

**Answer:** adjust for income (optionally education); never for infections or clinic visits.

### Exercise 15 · Hospital report cards

A regulator compares 30-day mortality across hospitals. Candidate adjustment variables: patient severity at admission, age, insurance type, length of stay, and severity at discharge. Draw the relevant DAG fragment and choose the adjustment set for "the effect of being treated at hospital $H$".

<details><summary>Show solution</summary>

**Step 2.** Length of stay and discharge severity are affected by the hospital (descendants), often mediators of quality.

```python
import itertools
import numpy as np
import networkx as nx

dsep = getattr(nx, "is_d_separator", None) or nx.d_separated   # name changed in networkx 3.3

def backdoor_ok(G, x, y, Z):
    """Back-door criterion: no descendant of x, and Z d-separates x and y once x's outgoing edges are removed."""
    if Z & nx.descendants(G, x):
        return False
    H = G.copy(); H.remove_edges_from(list(G.out_edges(x)))
    return dsep(H, {x}, {y}, set(Z))

G = nx.DiGraph([("Z1", "Z3"), ("Z2", "Z3"), ("Z1", "X"), ("Z2", "Y"), ("Z3", "X"), ("Z3", "Y"), ("X", "W"), ("W", "Y")])
cands = ["Z1", "Z2", "Z3", "W"]
valid = [set(c) for k in range(5) for c in itertools.combinations(cands, k) if backdoor_ok(G, "X", "Y", set(c))]
print("admissible sets:", [sorted(v) for v in valid])

# Exercise 12: sign reversal
pT = {1: .3, 0: .7}; pE = {(1, 0): .60, (1, 1): .70, (0, 0): .20, (0, 1): .35}     # key (z, t)
naive = {t: sum(.5 * (pT[z] if t else 1 - pT[z]) * pE[z, t] for z in (0, 1)) / .5 for t in (0, 1)}
adj = {t: sum(.5 * pE[z, t] for z in (0, 1)) for t in (0, 1)}
print("naive:", round(naive[1] - naive[0], 4), " adjusted:", round(adj[1] - adj[0], 4))

# Exercise 14: over-control
rng = np.random.default_rng(52); n = 200_000
P = rng.normal(size=n)
T = (P + rng.normal(size=n) > 0).astype(float)
J = 2 * T + rng.normal(size=n)
Y = 0.6 * T + 0.2 * J + 3 * P + rng.normal(size=n)          # total effect 0.6 + 0.2*2 = 1.0
for name, cols in [("none", []), ("P", [P]), ("P, J", [P, J])]:
    X = np.column_stack([np.ones(n), T] + cols)
    print(f"controls {name:5s}: coefficient on T = {np.linalg.lstsq(X, Y, rcond=None)[0][1]:.3f}")
```

## Key takeaways

- Back-door paths start with an arrow into the treatment; they carry confounding.
- A set is admissible if it blocks them all and contains no descendant of the treatment; then $P(y \mid \operatorname{do}(x)) = \sum_{\mathbf z} P(y \mid x, \mathbf z)P(\mathbf z)$.
- A variable can be both a confounder and a collider; the criterion, not intuition, decides what to add.
