When the regressor is correlated with the error, OLS is biased whatever the sample size. Omitted variables, measurement error and simultaneity are the three classic causes. An instrument moves the regressor but touches the outcome only through it, and the ratio of two covariances recovers the causal effect.

- State the instrumental-variable conditions (relevance, independence, exclusion) and draw them as a DAG.

- Judge candidate instruments in labour, health and industrial-organisation settings.

## Theory

### Endogeneity

In $Y = \alpha + \beta X + \varepsilon$, OLS estimates $\beta$ consistently only if $\operatorname{Cov}(X, \varepsilon) = 0$. Otherwise $X$ is **endogenous** and

$$
\operatorname{plim}\hat\beta_{OLS} = \beta + \frac{\operatorname{Cov}(X, \varepsilon)}{\operatorname{Var}(X)}.
$$

More data do not help: the bias does not shrink with $n$.

### Three sources

1. **Omitted variables.** If $\varepsilon = \gamma A + u$ with $A$ unobserved, the bias is $\gamma\operatorname{Cov}(X, A)/\operatorname{Var}(X)$ (Day 50).
2. **Measurement error.** If we observe $X = X^* + e$ with classical error $e$, OLS is **attenuated**: $\operatorname{plim}\hat\beta = \lambda\beta$, with reliability $\lambda = \operatorname{Var}(X^*)/(\operatorname{Var}(X^*) + \operatorname{Var}(e))$.
3. **Simultaneity.** Price and quantity are set jointly by supply and demand; the OLS slope mixes the two curves.

### Instruments

A variable $Z$ is a valid **instrument** for $X$ if

1. **Relevance:** $\operatorname{Cov}(Z, X) \ne 0$;
2. **Independence:** $Z$ is as good as randomly assigned (no common cause with $Y$);
3. **Exclusion:** $Z$ affects $Y$ only through $X$.

Conditions 2 and 3 together give $\operatorname{Cov}(Z, \varepsilon) = 0$. As a DAG: $Z \to X \to Y$, $U \to X$, $U \to Y$, with no arrow from $Z$ to $Y$ or from $U$ to $Z$.

### The IV estimand

Take covariances of $Y = \alpha + \beta X + \varepsilon$ with $Z$: $\operatorname{Cov}(Z, Y) = \beta\operatorname{Cov}(Z, X)$. So

$$
\beta_{IV} = \frac{\operatorname{Cov}(Z, Y)}{\operatorname{Cov}(Z, X)} = \frac{\text{reduced form}}{\text{first stage}},
$$

where the **first stage** is the slope of $X$ on $Z$ and the **reduced form** the slope of $Y$ on $Z$. Intuition: the instrument shifts $X$ by the first stage and, through $X$ alone, shifts $Y$ by the reduced form; the ratio is the effect per unit of $X$.

### What linearity buys

Day 56 showed that the instrument DAG does not identify $P(y \mid \operatorname{do}(x))$ nonparametrically. With a constant linear effect it identifies $\beta$. With heterogeneous effects and a binary treatment it identifies a local effect for compliers (Day 63).

### Classic instruments

Quarter of birth for schooling (compulsory-schooling laws); distance to a college; draft lottery numbers for military service; weather for agricultural supply; judge leniency for incarceration; lotteries for school places or health insurance.

### Exercise 6 · Omitted-variable bias

Let $Y = \beta X + \gamma A + u$ with $u$ uncorrelated with $(X, A)$. Prove $\operatorname{plim}\hat\beta_{OLS} = \beta + \gamma\operatorname{Cov}(X, A)/\operatorname{Var}(X)$ for the short regression.

### Exercise 7 · Attenuation bias

Let $Y = \beta X^* + u$ and $X = X^* + e$, with $e$ independent of $(X^*, u)$ and mean 0. Prove $\operatorname{plim}\hat\beta = \lambda\beta$.

<details><summary>Show solution</summary>

**Step 1.** $\operatorname{Cov}(X, Y) = \operatorname{Cov}(X^* + e, \beta X^* + u) = \beta\operatorname{Var}(X^*)$.

**Step 2.** $\operatorname{Var}(X) = \operatorname{Var}(X^*) + \operatorname{Var}(e)$.

**Step 3.** Ratio: $\beta\operatorname{Var}(X^*)/(\operatorname{Var}(X^*) + \operatorname{Var}(e)) = \lambda\beta$. $\blacksquare$

</details>

### Exercise 8 · Simultaneity bias

Demand $Q = -bP + u_d$ and supply $Q = sP + u_s$ ($b, s > 0$), with independent shocks of variances $\sigma_d^2, \sigma_s^2$. Prove the OLS slope of $Q$ on $P$ converges to $(s\sigma_d^2 - b\sigma_s^2)/(\sigma_d^2 + \sigma_s^2)$.

<details><summary>Show solution</summary>

**Step 1.** Equilibrium: $-bP + u_d = sP + u_s$, so $P = (u_d - u_s)/(b + s)$.

**Step 2.** $Q = sP + u_s = (s u_d + b u_s)/(b + s)$.

**Step 3.** $\operatorname{Cov}(P, Q) = (s\sigma_d^2 - b\sigma_s^2)/(b + s)^2$ and $\operatorname{Var}(P) = (\sigma_d^2 + \sigma_s^2)/(b + s)^2$.

**Step 4.** Ratio as stated. $\blacksquare$ OLS is a variance-weighted mix of the supply slope $s$ and the demand slope $-b$.

</details>

### Exercise 9 · IV is consistent

With $Y = \alpha + \beta X + \varepsilon$, $\operatorname{Cov}(Z, \varepsilon) = 0$ and $\operatorname{Cov}(Z, X) \ne 0$, prove $\hat\beta_{IV} = \widehat{\operatorname{Cov}}(Z, Y)/\widehat{\operatorname{Cov}}(Z, X) \to \beta$.

<details><summary>Show solution</summary>

**Step 1.** Sample covariances converge to population ones (LLN).

**Step 2.** $\operatorname{Cov}(Z, Y) = \beta\operatorname{Cov}(Z, X) + \operatorname{Cov}(Z, \varepsilon) = \beta\operatorname{Cov}(Z, X)$.

**Step 3.** By the continuous mapping theorem (denominator non-zero), the ratio converges to $\beta$. $\blacksquare$

</details>

### Exercise 10 · A second measurement as an instrument

Two independent noisy measurements $X_1 = X^* + e_1$ and $X_2 = X^* + e_2$ are available. Prove that IV with $X_1$ as the regressor and $X_2$ as the instrument is consistent for $\beta$ in $Y = \beta X^* + u$.

<details><summary>Show solution</summary>

**Step 4.** By Exercise 9, IV converges to $\beta$. $\blacksquare$

## Verify in Python

*Your work:*

```python
import numpy as np

rng = np.random.default_rng(61); n = 1_000_000
slope = lambda y, x: np.cov(x, y)[0, 1] / np.var(x, ddof=1)
iv = lambda y, x, z: np.cov(z, y)[0, 1] / np.cov(z, x)[0, 1]

# Omitted ability, with an instrument (true effect 0.08)
A = rng.normal(size=n); Z = rng.normal(size=n)
S = 0.3 * Z + 2 * A + rng.normal(0, 2, n)
W = 0.08 * S + 0.10 * A + rng.normal(0, .3, n)
print(f"schooling: OLS {slope(W, S):.4f}  IV {iv(W, S, Z):.4f}  (truth 0.08)")

# Measurement error: Exercise 2 and a second measurement as instrument
Xs = rng.normal(0, 2, n); Y = .5 * Xs + rng.normal(size=n)
X1 = Xs + rng.normal(size=n); X2 = Xs + rng.normal(size=n)
print(f"attenuation: OLS {slope(Y, X1):.3f} (0.4)  IV with second measure {iv(Y, X1, X2):.3f} (0.5)")

# Simultaneity: Exercise 8 with b = 1.2, s = 0.8, sd_d = sd_s = 1, and a supply shifter (weather)
b, s = 1.2, .8
ud = rng.normal(size=n); storm = rng.normal(size=n); us = rng.normal(size=n) - 0.5 * storm
P = (ud - us) / (b + s); Q = -b * P + ud
print(f"market: OLS {slope(Q, P):.3f} (formula {(s * 1 - b * 1.25) / (1 + 1.25):.3f})  IV demand slope {iv(Q, P, storm):.3f} (truth -1.2)")

# Exercises 11-14
print("college IV:", round(.042 / .32, 3), " fish elasticity:", -.36 / .30, " calories:", .64 * .002, " rainfall bias:", round(.02 / .1, 3))
```

## Key takeaways

- Endogeneity ($\operatorname{Cov}(X, \varepsilon) \ne 0$) comes from omitted variables, measurement error or simultaneity, and does not vanish with more data.
- An instrument must move the treatment (relevance), be as good as random (independence) and act only through the treatment (exclusion).
- $\beta_{IV} = \operatorname{Cov}(Z, Y)/\operatorname{Cov}(Z, X)$ = reduced form ÷ first stage; small exclusion violations are magnified by weak first stages.
