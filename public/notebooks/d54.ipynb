## Theory

### When back-door fails

Suppose $X \leftarrow U \to Y$ with $U$ unobserved. No back-door set exists. But if $X$ affects $Y$ **only** through a mediator $M$ that is not itself confounded with $X$, we can still identify the effect:

$$
U \to X, \quad U \to Y, \quad X \to M \to Y.
$$

Classic example: smoking $X$, tar deposits $M$, lung cancer $Y$, unobserved genotype $U$.

### The criterion

A set $\mathbf M$ satisfies the **front-door criterion** relative to $(X, Y)$ if

1. $\mathbf M$ intercepts **all directed paths** from $X$ to $Y$;
2. there is **no unblocked back-door path** from $X$ to $\mathbf M$;
3. **all back-door paths** from $\mathbf M$ to $Y$ are blocked by $X$.

### The formula

If $\mathbf M$ satisfies the criterion and $P(x, \mathbf m) > 0$,

$$
P(y \mid \operatorname{do}(x)) = \sum_{\mathbf m} P(\mathbf m \mid x)\sum_{x'} P(y \mid \mathbf m, x')\,P(x').
$$

### The logic in two steps

1. **Effect of $X$ on $M$.** By condition 2, $P(m \mid \operatorname{do}(x)) = P(m \mid x)$.
2. **Effect of $M$ on $Y$.** By condition 3, $X$ is a back-door set for $M \to Y$: $P(y \mid \operatorname{do}(m)) = \sum_{x'} P(y \mid m, x')P(x')$.
3. **Chain them.** By condition 1, $X$ affects $Y$ only via $M$: $P(y \mid \operatorname{do}(x)) = \sum_m P(m \mid \operatorname{do}(x))\,P(y \mid \operatorname{do}(m))$.

### The linear case

With $M = aX + e_M$ and $Y = bM + cU + e_Y$, $X$ correlated with $U$: estimate $a$ by regressing $M$ on $X$, and $b$ by regressing $Y$ on $M$ **and** $X$. The effect is $\hat a\hat b$.

### Plausibility

The conditions are strong: no direct effect of $X$ on $Y$ outside $M$, and no unobserved common cause of $M$ and $Y$. They hold most credibly when $M$ is a mechanical step: signing up versus attending, being offered versus receiving, prescribed versus dispensed.

### Exercise 5 · Why average over $x'$?

In the second step, why do we average $P(y \mid m, x')$ over $P(x')$ instead of using $P(y \mid m, x)$ at the treatment value $x$?

<details><summary>Show solution</summary>

**Step 1.** We need the effect of $M$ on $Y$ in the *whole population*, $P(y \mid \operatorname{do}(m))$.

**Step 2.** $X$ is used only as a back-door adjustment variable for $M \to Y$ (it blocks $M \leftarrow X \leftarrow U \to Y$), so it must be averaged over its marginal distribution, as in any back-door formula.

**Answer:** using $P(y \mid m, x)$ alone would describe the $U$-mix of people with $X = x$, which is confounded.

### Exercise 7 · Step 2 in general

Prove $P(y \mid \operatorname{do}(m)) = \sum_{x} P(y \mid m, x)P(x)$ under the front-door conditions.

<details><summary>Show solution</summary>

**Step 1.** Condition 3: $\{X\}$ blocks all back-door paths from $M$ to $Y$.

**Step 2.** $X$ is not a descendant of $M$: $M$ lies on the directed paths from $X$, and acyclicity forbids $M \to \cdots \to X$.

**Step 3.** So $\{X\}$ satisfies the back-door criterion for $(M, Y)$ and the adjustment formula gives the result. $\blacksquare$

### Exercise 8 · Chaining the steps

Using the SCM $U \to X$, $U \to Y$, $X \to M \to Y$, prove $P(y \mid \operatorname{do}(x)) = \sum_m P(m \mid \operatorname{do}(x))P(y \mid \operatorname{do}(m))$.

<details><summary>Show solution</summary>

**Step 1.** Truncated factorisation for $\operatorname{do}(x)$: $P(u, m, y \mid \operatorname{do}(x)) = P(u)P(m \mid x)P(y \mid m, u)$.

**Step 2.** Sum over $u$ and $m$: $P(y \mid \operatorname{do}(x)) = \sum_m P(m \mid x)\sum_u P(y \mid m, u)P(u)$.

**Step 3.** Truncated factorisation for $\operatorname{do}(m)$: $P(u, x, y \mid \operatorname{do}(m)) = P(u)P(x \mid u)P(y \mid m, u)$; summing over $u, x$ gives $P(y \mid \operatorname{do}(m)) = \sum_u P(y \mid m, u)P(u)$.

**Step 4.** By Exercise 6, $P(m \mid x) = P(m \mid \operatorname{do}(x))$. Substitute both into Step 2. $\blacksquare$ Combined with Exercise 7, this proves the front-door formula.

### Exercise 9 · The linear product

With $X = gU + e_X$, $M = aX + e_M$, $Y = bM + cU + e_Y$ (all disturbances independent), prove the population coefficient on $M$ in the regression of $Y$ on $(M, X)$ is $b$, so $\hat a\hat b \to ab$.

<details><summary>Show solution</summary>

**Step 4.** The regression of $M$ on $X$ recovers $a$ because $e_M$ is independent of $X$. $\blacksquare$

```python
import itertools
import numpy as np

# Exercise 11: the SCM behind the table (U = genotype, unobserved)
pU = .5; pX = {1: .8, 0: .2}; pM = {1: .9, 0: .1}; pY = {(0, 0): .1, (1, 0): .2, (0, 1): .4, (1, 1): .5}   # key (m, u)
truth = {x: sum((pM[x] if m else 1 - pM[x]) * sum((pU if u else 1 - pU) * pY[m, u] for u in (0, 1)) for m in (0, 1)) for x in (0, 1)}
J = {(u, x, m): (pU if u else 1 - pU) * (pX[u] if x else 1 - pX[u]) * (pM[x] if m else 1 - pM[x])
     for u, x, m in itertools.product((0, 1), repeat=3)}
Px = {x: sum(v for k, v in J.items() if k[1] == x) for x in (0, 1)}
Pm_x = {(m, x): sum(v for k, v in J.items() if k[1] == x and k[2] == m) / Px[x] for m in (0, 1) for x in (0, 1)}
Py_mx = {(m, x): sum(v * pY[m, k[0]] for k, v in J.items() if k[1] == x and k[2] == m) / (Px[x] * Pm_x[m, x]) for m in (0, 1) for x in (0, 1)}
print("observed P(Y=1|M,X):", {k: round(v, 3) for k, v in Py_mx.items()})
naive = {x: sum(Pm_x[m, x] * Py_mx[m, x] for m in (0, 1)) for x in (0, 1)}
front = {x: sum(Pm_x[m, x] * sum(Py_mx[m, xp] * Px[xp] for xp in (0, 1)) for m in (0, 1)) for x in (0, 1)}
print(f"naive {naive[1] - naive[0]:.3f}  front-door {front[1] - front[0]:.3f}  truth {truth[1] - truth[0]:.3f}")

# Exercises 13-14: linear front door with and without a direct effect
rng = np.random.default_rng(54); n = 500_000
L = rng.normal(size=n)                                 # financial literacy (unobserved)
X = (L + rng.normal(size=n) > 0).astype(float)
M = 0.6 * X + rng.normal(size=n)
for d in (0, 10):
    Y = 50 * M + d * X + 40 * L + rng.normal(0, 20, n)
    a = np.polyfit(X, M, 1)[0]
    b = np.linalg.lstsq(np.column_stack([np.ones(n), M, X]), Y, rcond=None)[0][1]
    naive = Y[X == 1].mean() - Y[X == 0].mean()
    print(f"direct effect {d}: naive {naive:.1f}, front-door {a * b:.1f}, truth {30 + d}")
```

## Key takeaways

- The front door identifies an effect despite unobserved treatment–outcome confounding, if a mediator captures the whole effect, is unconfounded with the treatment, and is unconfounded with the outcome given the treatment.
- Formula: $P(y \mid \operatorname{do}(x)) = \sum_m P(m \mid x)\sum_{x'} P(y \mid m, x')P(x')$; in linear models, the product of two regression coefficients.
- Its assumptions are strong; they are most credible when the mediator is a mechanical step of the treatment.
