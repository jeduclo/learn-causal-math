## Theory

### Conditions for an optimum

For a smooth $f: \mathbb{R}^k \to \mathbb{R}$:

- **First-order condition (FOC):** at an interior optimum, $\nabla f(x^*) = 0$.
- **Second-order condition (SOC):** if also $H(x^*)$ is positive definite, $x^*$ is a strict local minimum; negative definite, a strict local maximum; indefinite, a **saddle point**.

### Convexity

A function is **convex** if the chord lies above the graph:

$$
f(\lambda x + (1 - \lambda)y) \le \lambda f(x) + (1 - \lambda)f(y), \qquad \lambda \in [0, 1].
$$

For smooth functions, convex $\iff$ the tangent plane lies below: $f(y) \ge f(x) + \nabla f(x)^{\top}(y - x)$ $\iff$ $H(x)$ is PSD everywhere. The payoff: **for a convex function, any point with zero gradient is a global minimum**, and strict convexity makes it unique. Least squares, ridge, lasso and the (negative) logit log-likelihood are all convex.

### Algorithms

When the FOC cannot be solved by hand:

- **Gradient descent:** $x_{t+1} = x_t - \eta\nabla f(x_t)$. Cheap per step; needs a good step size $\eta$. On a quadratic with Hessian eigenvalues $\lambda_j$ it converges iff $\eta < 2/\lambda_{\max}$, and slowly when the condition number is large.
- **Newton's method:** $x_{t+1} = x_t - H(x_t)^{-1}\nabla f(x_t)$. It minimises the local second-order Taylor approximation (Day 23) at each step; it solves a quadratic in one step and converges very fast near the optimum, but each step needs the Hessian.

For maximum likelihood, Newton's method applied to $\ell$ is the classic **Newton–Raphson** (for the logit it is called iteratively reweighted least squares).

### Exercise 10 · When gradient descent converges

For $f(x) = \frac12x^{\top}Ax$ with $A$ PD, show gradient descent converges to 0 iff $\eta < 2/\lambda_{\max}$.

<details><summary>Show solution</summary>

**Step 1.** $\nabla f = Ax$, so $x_{t+1} = x_t - \eta Ax_t = (I - \eta A)x_t$, and by induction $x_t = (I - \eta A)^tx_0$.

**Step 2.** If $Aq_j = \lambda_jq_j$, then $(I - \eta A)q_j = (1 - \eta\lambda_j)q_j$: same eigenvectors, eigenvalues $1 - \eta\lambda_j$.

**Step 3.** Write $x_0 = \sum_jc_jq_j$. Then $x_t = \sum_jc_j(1 - \eta\lambda_j)^tq_j$.

**Step 4.** $x_t \to 0$ for every start iff $|1 - \eta\lambda_j| < 1$ for all $j$, i.e. $0 < \eta < 2/\lambda_j$ for all $j$, i.e. $\eta < 2/\lambda_{\max}$.

**Step 5.** Speed: the slowest direction shrinks by $|1 - \eta\lambda_{\min}|$ per step. With $\eta$ near $2/\lambda_{\max}$ and $\lambda_{\min} \ll \lambda_{\max}$ (ill-conditioning), this is close to 1, so progress is slow.

</details>

### Exercise 11 · A competitive firm

A firm sells at $p = 10$, produces $q = 2L^{0.5}$ and pays wage $w = 5$. Find the profit-maximising labour, output and profit.

<details><summary>Show solution</summary>

**Step 1.** Profit: $\pi(L) = p\,q - wL = 10(2L^{0.5}) - 5L = 20L^{0.5} - 5L$.

**Step 2.** FOC: $\pi'(L) = 10L^{-0.5} - 5 = 0 \Rightarrow L^{-0.5} = 0.5 \Rightarrow L = 4$.

**Step 3.** SOC: $\pi''(L) = -5L^{-1.5} < 0$: a maximum.

**Step 4.** Output $q = 2\sqrt4 = 4$; profit $= 40 - 20 = 20$.

**Answer:** hire 4 units of labour, produce 4, earn profit 20 (the value of the marginal product equals the wage).

</details>

### Exercise 12 · Monopoly pricing

Inverse demand $P = 100 - 2Q$, cost $C = 20Q$. Find the monopoly quantity, price and profit.

<details><summary>Show solution</summary>

**Step 1.** Profit: $\pi(Q) = (100 - 2Q)Q - 20Q = 80Q - 2Q^2$.

**Step 2.** FOC (marginal revenue = marginal cost): $80 - 4Q = 0 \Rightarrow Q = 20$. SOC: $\pi'' = -4 < 0$.

**Step 3.** Price: $P = 100 - 2(20) = 60$.

**Step 4.** Profit: $(60 - 20) \times 20 = 800$.

**Answer:** $Q = 20$, $P = 60$, profit 800.

### Exercise 13 · Newton–Raphson for a Poisson MLE

Ten counts have $\sum x_i = 30$. Starting from $\lambda_0 = 1$, perform two Newton–Raphson steps on $\ell(\lambda) = \sum x_i\ln\lambda - n\lambda$.

<details><summary>Show solution</summary>

**Step 1.** Derivatives: $\ell'(\lambda) = 30/\lambda - 10$ and $\ell''(\lambda) = -30/\lambda^2$. Newton update: $\lambda \leftarrow \lambda - \ell'(\lambda)/\ell''(\lambda)$.

**Step 2.** From $\lambda_0 = 1$: $\ell' = 20$, $\ell'' = -30$, so $\lambda_1 = 1 + 20/30 \approx 1.667$.

**Step 3.** From $\lambda_1 = 5/3$: $\ell' = 18 - 10 = 8$, $\ell'' = -30/(25/9) = -10.8$, so $\lambda_2 = 1.667 + 8/10.8 \approx 2.407$.

**Step 4.** Further steps give 2.883, 2.995, 3.000 (code below).

**Answer:** the iterates converge to $\hat\lambda = \bar x = 3$, slowly at first and then very fast near the optimum.

### Exercise 14 · Estimating a CES production function

An economist fits a CES production function $\ln Y_i = \ln A + \frac{\nu}{\rho}\ln\big(\delta K_i^{\rho} + (1 - \delta)L_i^{\rho}\big) + \varepsilon_i$ by non-linear least squares. The objective can have several local minima and flat regions (e.g. near $\rho = 0$, the Cobb–Douglas limit). What should she do?

<details><summary>Show solution</summary>

**Step 1.** Recognise the risk: a gradient method stops at *a* point with zero gradient, which need not be the global minimum.

**Step 2.** Start the optimiser from several points (or a grid over $\rho$ and $\delta$) and compare the objective values reached.

**Step 3.** Reparametrise where possible (e.g. profile out $\ln A$, which enters linearly) and check the curvature (Hessian) at the solution.

**Answer:** use multiple starts, report sensitivity, and prefer formulations that are convex. This is one reason linear and logit models, whose objectives are convex, dominate applied work.

</details>

### Exercise 15 · When the MLE does not exist

In a logit of treatment on one covariate, every unit with $x > 5$ is treated and every unit with $x \le 5$ is not. What happens to Newton–Raphson, and what does it mean for propensity-score methods?

<details><summary>Show solution</summary>

**Step 1.** Perfect separation: every unit with $x > 5$ is treated and every unit with $x \le 5$ is not.

**Step 2.** A steeper slope always fits better: the log-likelihood keeps increasing as $\beta_1 \to \infty$, so there is no finite maximum.

**Step 3.** Newton–Raphson therefore diverges, with fitted probabilities heading to exactly 0 and 1 (the code shows the slope growing every iteration).

**Answer:** no MLE exists. For causal inference this is a failure of **overlap**: some covariate values have no comparison units, so effects there cannot be estimated without extrapolation (Day 42).

</details>

## Verify in Python

```python
import numpy as np

# Gradient descent on a quadratic with condition number 100 vs Newton
A = np.diag([1.0, 100.0]); grad = lambda x: A @ x
x = np.array([1.0, 1.0]); eta = 1.9 / 100
for t in range(200):
    x = x - eta * grad(x)
print("GD after 200 steps:", x.round(4), "  (slow along the small-eigenvalue direction)")
x0 = np.array([1.0, 1.0]); print("Newton after 1 step:", x0 - np.linalg.solve(A, grad(x0)))

# Newton-Raphson for the Poisson MLE (Exercise 13)
lam, path = 1.0, []
for _ in range(6):
    lam = lam - (30 / lam - 10) / (-30 / lam**2); path.append(round(lam, 4))
print("Newton path:", path)

# Perfect separation: the logit slope runs away
x = np.arange(1, 11.); T = (x > 5).astype(float); X = np.column_stack([np.ones(10), x]); b = np.zeros(2)
for it in range(8):
    p = 1 / (1 + np.exp(-X @ b)); W = p * (1 - p)
    b = b + np.linalg.solve(X.T @ (X * W[:, None]) + 1e-12 * np.eye(2), X.T @ (T - p))
    print(f"iteration {it + 1}: slope = {b[1]:.2f}")
```

## Key takeaways

- FOC finds candidates; the Hessian classifies them; convexity turns local into global.
- Gradient descent is cheap but slowed by ill-conditioning; Newton uses curvature and converges fast near the optimum.
- Non-existence of an MLE (separation) is a data problem, often an overlap problem, not a software problem.
