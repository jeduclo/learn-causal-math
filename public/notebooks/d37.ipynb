## Theory

### The sharp null

Fisher (1935) tested the **sharp null**

$$
H_0: Y_i(1) = Y_i(0) \quad\text{for every unit } i.
$$

Under $H_0$, each unit's outcome would have been the same whatever its assignment, so the observed $Y_i$ fills in *both* potential outcomes. The whole science table is known.

### The randomisation distribution

Pick a test statistic $S(\mathbf T, \mathbf Y)$, e.g. the difference in means. Because the science table is known under $H_0$, we can compute $S$ for *every* assignment the design could have produced. The exact p-value is

$$
p = \frac{\#\{\text{assignments } \mathbf t: |S(\mathbf t, \mathbf Y)| \ge |S^{obs}|\}}{\#\{\text{assignments}\}}.
$$

No normality, no large sample, no model: the only randomness is the one the researcher created. When there are too many assignments to list, draw $R$ at random (Monte Carlo); the p-value then has standard error $\sqrt{p(1 - p)/R}$.

### Choices and extensions

- **Statistic:** any function of the data. Rank statistics (Wilcoxon) resist outliers.
- **Other sharp nulls:** $H_0: Y_i(1) = Y_i(0) + \tau_0$ for all $i$. Subtract $\tau_0$ from treated outcomes and test again. Collecting all $\tau_0$ not rejected gives a **confidence interval** (test inversion).
- **Design:** permute exactly as the design randomised: within blocks for blocked designs, whole clusters for cluster designs.
- **Sharp versus weak null:** the sharp null (no effect for anyone) is stronger than Neyman's weak null (zero *average* effect). Randomisation tests are exact for the former and only approximately valid for the latter.

### Exercise 6 · The test is exact

Under the sharp null and the actual design, prove that $P(p \le \alpha) \le \alpha$.

<details><summary>Show solution</summary>

**Step 1.** Under $H_0$ the outcome vector $\mathbf Y$ is fixed whatever the assignment, so the statistic $S(\mathbf t)$ is a fixed number for each of the $K$ possible assignments.

**Step 2.** The observed assignment is one of them, chosen uniformly at random by the design. So $|S^{obs}|$ is a uniformly random draw from the $K$ values $|S(\mathbf t)|$.

**Step 3.** The p-value is the share of values at least as large as the draw. A uniformly random draw lands in the top $100\alpha\%$ with probability at most $\alpha$ (ties only make the p-value larger).

**Answer:** $P(p \le \alpha) \le \alpha$: exact Type I error control in any sample size. $\blacksquare$

</details>

### Exercise 7 · Confidence intervals by inversion

For $H_0(\tau_0): Y_i(1) = Y_i(0) + \tau_0$, show how to impute the science table, and explain why $\{\tau_0: p(\tau_0) > \alpha\}$ is a $1 - \alpha$ confidence set.

<details><summary>Show solution</summary>

**Step 1.** Treated units: $Y_i(0) = Y_i - \tau_0$. Controls: $Y_i(1) = Y_i + \tau_0$.

**Step 2.** Equivalently, subtract $\tau_0$ from treated outcomes; under $H_0(\tau_0)$ the adjusted outcomes satisfy the sharp null of no effect, so Exercise 6 applies.

**Step 3.** If the true effect is $\tau^*$, the test of $H_0(\tau^*)$ rejects with probability at most $\alpha$, so $\tau^*$ is in the set with probability at least $1 - \alpha$. $\blacksquare$

</details>

Show that, under the sharp null, the randomisation distribution of the Wilcoxon rank-sum statistic (the sum of treated units' ranks) depends only on $n$ and $n_1$ (with no ties), not on the outcome values.

<details><summary>Show solution</summary>

### Exercise 12 · A small hospital trial

10 patients are randomised, 5 per arm; all 5 treated recover versus 1 of 5 controls. Compute Fisher's exact p-values.

<details><summary>Show solution</summary>

**Step 1.** Under $H_0$, the 6 recoverers would recover whatever the arm. The number of recoverers landing in the treated arm is hypergeometric: choose 5 treated from 10 patients, 6 of whom recover.

**Step 2.** All 5 treated recover: $\binom65\binom40/\binom{10}{5} = 6/252 \approx 0.024$. That is the one-sided p-value (no table is more extreme).

**Step 3.** The opposite extreme (only 1 treated recoverer) also has probability $\binom61\binom44/252 = 6/252$.

**Step 4.** Two-sided $p = 12/252 \approx 0.048$.

**Answer:** one-sided 0.024, two-sided 0.048: significant at 5% with just 10 patients, and no large-sample approximation needed.

</details>

A study randomises 10 of 20 schools; outcomes are measured for 3,000 pupils. How should the randomisation test permute the data, and why?

## Verify in Python

*Your work:*

```python
import itertools
import numpy as np
from scipy import stats

y = np.array([12, 15, 11, 14, 9, 10, 13, 8.]); treated = np.array([1, 1, 1, 1, 0, 0, 0, 0], bool)

def rand_test(y, treated, tau0=0.0):
    z = y - tau0 * treated                                   # impute under H0(tau0)
    obs = z[treated].mean() - z[~treated].mean()
    dist = np.array([z[list(c)].mean() - np.delete(z, list(c)).mean()
                     for c in itertools.combinations(range(len(z)), treated.sum())])
    return obs, (dist >= obs - 1e-9).mean(), (np.abs(dist) >= abs(obs) - 1e-9).mean()

obs, p1, p2 = rand_test(y, treated)
print(f"observed {obs}, one-sided p {p1:.3f}, two-sided p {p2:.3f}")
print("Welch t-test p:", round(stats.ttest_ind(y[treated], y[~treated], equal_var=False).pvalue, 3))
print("test of tau0 = 2: p =", round(rand_test(y, treated, 2)[2], 3))
ci = [t for t in np.arange(-4, 10, 0.1) if rand_test(y, treated, t)[2] > 0.10]
print(f"90% interval by inversion: [{min(ci):.1f}, {max(ci):.1f}]")

# Fisher's exact test for the hospital trial
print("Fisher exact (two-sided):", round(stats.fisher_exact([[5, 0], [1, 4]])[1], 4),
      " one-sided:", round(stats.fisher_exact([[5, 0], [1, 4]], alternative="greater")[1], 4))
```

## Key takeaways

- Under the sharp null, every potential outcome is known, so the randomisation itself gives an exact null distribution.
- Permute exactly as the design randomised (within blocks, by cluster).
- Inverting sharp-null tests yields confidence intervals; randomisation tests concern "no effect for anyone".
