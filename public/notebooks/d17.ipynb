## Theory

### The product

If $A$ is $m \times n$ and $B$ is $n \times p$, the product $C = AB$ is $m \times p$ with

$$
c_{ij} = \sum_{\ell=1}^n a_{i\ell}b_{\ell j} = (\text{row } i \text{ of } A) \cdot (\text{column } j \text{ of } B).
$$

The inner dimensions must match ("conformable"). Matrix multiplication is **associative** and **distributive** but **not commutative**: in general $AB \ne BA$, and one of them may not even exist.

### Two ways to read Ax

$$
Ax = \sum_{j} x_j\,a_{\cdot j}
$$

is a **linear combination of the columns** of $A$ with weights $x_j$. So $X\beta$, the vector of fitted values, is "$\beta_0$ times the intercept column plus $\beta_1$ times the education column plus …". Alternatively, entry $i$ of $Ax$ is row $i$ of $A$ dotted with $x$: unit $i$'s prediction.

A matrix also acts as a **linear transformation**: it maps vectors to vectors, preserving sums and scalar multiples. In 2D you can watch it stretch, rotate, shear or flip the plane (Simulation Lab).

### Transpose and trace

$$
(AB)^{\top} = B^{\top}A^{\top}, \qquad \operatorname{tr}(A) = \sum_i a_{ii}, \qquad \operatorname{tr}(AB) = \operatorname{tr}(BA).
$$

### The cross-product matrices

For a design matrix $X$ ($n \times k$) and outcome $y$ ($n \times 1$):

$$
X^{\top}X = \Big[\sum_i x_{ij}x_{i\ell}\Big]_{j,\ell}, \qquad X^{\top}y = \Big[\sum_i x_{ij}y_i\Big]_{j}.
$$

With an intercept and one regressor,

$$
X^{\top}X = \begin{pmatrix} n & \sum x_i \\ \sum x_i & \sum x_i^2 \end{pmatrix}, \qquad
X^{\top}y = \begin{pmatrix} \sum y_i \\ \sum x_iy_i \end{pmatrix}.
$$

On Day 27 the OLS estimator will be $\hat\beta = (X^{\top}X)^{-1}X^{\top}y$: everything it needs is in these two objects.

### The centring matrix

$$
M_{\mathbf 1} = I_n - \tfrac1n\mathbf{1}\mathbf{1}^{\top}, \qquad M_{\mathbf 1}y = y - \bar y\mathbf{1}.
$$

Multiplying by $M_{\mathbf 1}$ subtracts the mean, the matrix version of "demeaning", which becomes fixed-effects estimation on Day 28.

Show that $X^{\top}X$ is symmetric, that its diagonal holds the sums of squares of the columns, and derive its form with an intercept and one regressor.

### Exercise 14 · Demeaning household incomes

Three households have monthly incomes $y = (1, 2, 6)$ (thousand dollars). Write the centring matrix $M_{\mathbf 1}$ for $n = 3$ and compute $M_{\mathbf 1}y$.

<details><summary>Show solution</summary>

**Step 1.** $M_{\mathbf 1} = I - \frac13\mathbf 1\mathbf 1^{\top}$: diagonal entries $1 - \frac13 = \frac23$, off-diagonal $-\frac13$:

$$
M_{\mathbf 1} = \frac13\begin{pmatrix}2 & -1 & -1\\-1 & 2 & -1\\-1 & -1 & 2\end{pmatrix}.
$$

**Step 2.** First entry: $\frac13(2 \cdot 1 - 2 - 6) = -2$; similarly the others.

**Step 3.** Equivalently, subtract the mean $\bar y = 3$: $(1 - 3, 2 - 3, 6 - 3)$.

**Answer:** $M_{\mathbf 1}y = (-2, -1, 3)$: each household's income relative to the average.

</details>

### Exercise 15 · Computing X′y for a large survey

A survey has $n = 1000$ households and $k = 3$ regressors, so $X^{\top}$ is $3 \times 1000$, $X$ is $1000 \times 3$ and $y$ is $1000 \times 1$. Software must compute $X(X^{\top}y)$ (needed for fitted values). Compare the multiplications needed for $(XX^{\top})y$ and $X(X^{\top}y)$.

<details><summary>Show solution</summary>

**Step 1.** $(XX^{\top})y$: forming $XX^{\top}$ ($1000 \times 1000$) costs $1000 \times 1000 \times 3 = 3{,}000{,}000$ multiplications.

**Step 2.** Multiplying that by $y$ costs another $1000 \times 1000 = 1{,}000{,}000$: total about **4 million**.

**Step 3.** $X(X^{\top}y)$: $X^{\top}y$ ($3 \times 1$) costs $3 \times 1000 = 3{,}000$.

**Step 4.** Then $X$ times a $3 \times 1$ vector costs $1000 \times 3 = 3{,}000$: total **6,000**.

**Answer:** same result, about 700 times cheaper, which is why software never builds an $n \times n$ matrix when it can avoid it.

</details>

## Verify in Python

*Your work:*

```python
import numpy as np

A = np.array([[1, 2], [3, 4]]); B = np.array([[0, 1], [1, 0]])
print("AB =\n", A @ B, "\nBA =\n", B @ A)
print("(AB)' == B'A':", np.array_equal((A @ B).T, B.T @ A.T), " tr(AB), tr(BA):", np.trace(A @ B), np.trace(B @ A))

x = np.array([1, 2, 3, 4.]); y = np.array([2, 3, 5, 6.])
X = np.column_stack([np.ones(4), x])
print("X'X =\n", X.T @ X, "\nX'y =", X.T @ y)
beta = np.array([0.5, 1.4]); e = y - X @ beta
print("residuals:", e.round(2), " X'e =", (X.T @ e).round(10))

n = 3; M = np.eye(n) - np.ones((n, n)) / n
print("M y =", M @ np.array([1, 2, 6.]))
```

## Key takeaways

- $Ax$ is a weighted combination of the columns of $A$; $X\beta$ is the vector of fitted values.
- $(AB)^{\top} = B^{\top}A^{\top}$ and $\operatorname{tr}(AB) = \operatorname{tr}(BA)$ are used constantly in proofs.
- $X^{\top}X$ and $X^{\top}y$ hold all the sums of squares and cross-products that regression needs.
