## Theory

### The finite-population variance

Neyman (1923) treated potential outcomes as fixed and asked how the difference in means $\hat\tau = \bar Y_1 - \bar Y_0$ varies across all possible completely randomised assignments. With finite-population variances (divisor $n - 1$)

$$
S_1^2 = \tfrac{1}{n-1}\sum_i(Y_i(1) - \bar Y(1))^2, \quad S_0^2 = \tfrac{1}{n-1}\sum_i(Y_i(0) - \bar Y(0))^2, \quad S_\tau^2 = \tfrac{1}{n-1}\sum_i(\tau_i - \tau_{fp})^2,
$$

the exact variance is

$$
\operatorname{Var}(\hat\tau) = \frac{S_1^2}{n_1} + \frac{S_0^2}{n_0} - \frac{S_\tau^2}{n}.
$$

### Estimating it

$S_1^2$ and $S_0^2$ are estimated without bias by the sample variances $s_1^2$ (treated) and $s_0^2$ (control). But $S_\tau^2$ involves both potential outcomes of each unit and cannot be estimated. Neyman's estimator drops it:

$$
\hat V = \frac{s_1^2}{n_1} + \frac{s_0^2}{n_0}, \qquad E[\hat V] - \operatorname{Var}(\hat\tau) = \frac{S_\tau^2}{n} \ge 0.
$$

So $\hat V$ is **conservative**: unbiased when effects are constant ($S_\tau^2 = 0$), too large otherwise. Confidence intervals $\hat\tau \pm 1.96\sqrt{\hat V}$ therefore cover *at least* 95% in large samples.

A sharper (still valid) bound uses $S_\tau^2 \ge (S_1 - S_0)^2$.

### Finite versus super population

If units are a random sample from a large population, the variance of $\hat\tau$ around the *super-population* ATE is $\sigma_1^2/n_1 + \sigma_0^2/n_0$, with no negative term, and $\hat V$ is unbiased for it (Imbens & Rubin 2015). The same formula therefore serves both views: exact for one, conservative for the other.

### Exercise 7 · Variance of the treated mean

Prove $\operatorname{Var}(\bar Y_1) = S_1^2\big(\frac{1}{n_1} - \frac1n\big)$.

<details><summary>Show solution</summary>

Write $a_i = Y_i(1)$ and use Day 35's moments: $\operatorname{Var}(T_i) = \frac{n_1n_0}{n^2}$, $\operatorname{Cov}(T_i, T_j) = -\frac{n_1n_0}{n^2(n - 1)}$.

**Step 1.** $\operatorname{Var}\Big(\sum_iT_ia_i\Big) = \frac{n_1n_0}{n^2}\Big[\sum_ia_i^2 - \frac{1}{n - 1}\sum_{i \ne j}a_ia_j\Big]$.

**Step 2.** Use $\sum_{i \ne j}a_ia_j = (\sum_ia_i)^2 - \sum_ia_i^2$. The bracket becomes

$$
\frac{n\sum_ia_i^2 - (\sum_ia_i)^2}{n - 1} = \frac{n\sum_i(a_i - \bar a)^2}{n - 1} = nS_1^2.
$$

**Step 3.** So $\operatorname{Var}(\sum_iT_ia_i) = \frac{n_1n_0}{n}S_1^2$.

**Step 4.** Divide by $n_1^2$: $\operatorname{Var}(\bar Y_1) = \frac{n_0}{nn_1}S_1^2 = S_1^2\Big(\frac{1}{n_1} - \frac1n\Big)$. $\blacksquare$

The $-1/n$ is a finite-population correction: treating everyone ($n_1 = n$) gives zero variance.

</details>

### Exercise 8 · The full Neyman formula (sketch)

Using Exercise 7 and $\operatorname{Cov}(\bar Y_1, \bar Y_0) = -S_{10}/n$, where $S_{10}$ is the finite-population covariance of $Y(1)$ and $Y(0)$, derive $\operatorname{Var}(\hat\tau)$.

<details><summary>Show solution</summary>

**Step 1.** $\operatorname{Var}(\hat\tau) = \operatorname{Var}(\bar Y_1) + \operatorname{Var}(\bar Y_0) - 2\operatorname{Cov}(\bar Y_1, \bar Y_0)$.

**Step 2.** Substitute: $S_1^2\big(\frac{1}{n_1} - \frac1n\big) + S_0^2\big(\frac{1}{n_0} - \frac1n\big) + \frac{2S_{10}}{n}$.

**Step 3.** Since $\tau_i = Y_i(1) - Y_i(0)$, $S_\tau^2 = S_1^2 + S_0^2 - 2S_{10}$.

**Step 4.** So $-\frac{S_1^2 + S_0^2 - 2S_{10}}{n} = -\frac{S_\tau^2}{n}$, giving $\operatorname{Var}(\hat\tau) = \frac{S_1^2}{n_1} + \frac{S_0^2}{n_0} - \frac{S_\tau^2}{n}$. $\blacksquare$

</details>

### Exercise 9 · Neyman's estimator is conservative

Given $E[s_1^2] = S_1^2$ and $E[s_0^2] = S_0^2$ under complete randomisation, prove $E[\hat V] \ge \operatorname{Var}(\hat\tau)$, with equality iff effects are constant.

<details><summary>Show solution</summary>

**Step 1.** $E[\hat V] = S_1^2/n_1 + S_0^2/n_0$.

**Step 2.** Subtract the true variance: $E[\hat V] - \operatorname{Var}(\hat\tau) = S_\tau^2/n \ge 0$.

**Step 3.** $S_\tau^2 = 0$ iff all $\tau_i$ equal $\tau_{fp}$, i.e. constant effects. $\blacksquare$

</details>

*Your work:*

### Exercise 10 · A sharper bound

Prove $S_\tau^2 \ge (S_1 - S_0)^2$ and deduce a smaller conservative variance estimator.

<details><summary>Show solution</summary>

**Step 1.** $S_\tau^2 = S_1^2 + S_0^2 - 2S_{10}$.

**Step 2.** Cauchy–Schwarz (Day 16): $S_{10} \le S_1S_0$.

**Step 3.** So $S_\tau^2 \ge S_1^2 + S_0^2 - 2S_1S_0 = (S_1 - S_0)^2$.

**Step 4.** Hence $\operatorname{Var}(\hat\tau) \le \frac{S_1^2}{n_1} + \frac{S_0^2}{n_0} - \frac{(S_1 - S_0)^2}{n}$, estimated by plugging in $s_1, s_0$. $\blacksquare$

The bound is attained when $Y(1)$ and $Y(0)$ are perfectly positively correlated.

</details>

## Verify in Python

*Your work:*

```python
import itertools
import numpy as np

y1 = np.array([14, 12, 13, 15.]); y0 = np.array([10, 12, 8, 15.]); n, n1 = 4, 2
S1, S0, St = y1.var(ddof=1), y0.var(ddof=1), (y1 - y0).var(ddof=1)
print(f"S1^2 {S1:.3f}, S0^2 {S0:.3f}, Stau^2 {St:.3f}; Neyman variance {S1 / n1 + S0 / (n - n1) - St / n:.4f}")
ests = []
for c in itertools.combinations(range(n), n1):
    T = np.zeros(n, bool); T[list(c)] = True
    ests.append(y1[T].mean() - y0[~T].mean())
print("variance over all assignments:", np.var(ests))

# Coverage of Neyman intervals with heterogeneous effects
rng = np.random.default_rng(36)
N, N1 = 200, 100
Y0 = rng.normal(50, 10, N); Y1 = Y0 + rng.normal(5, 6, N)   # heterogeneous effects
tau = (Y1 - Y0).mean(); cover = []; vhat = []
for _ in range(20_000):
    T = np.zeros(N, bool); T[rng.choice(N, N1, replace=False)] = True
    est = Y1[T].mean() - Y0[~T].mean()
    v = Y1[T].var(ddof=1) / N1 + Y0[~T].var(ddof=1) / (N - N1)
    vhat.append(v); cover.append(abs(est - tau) <= 1.96 * np.sqrt(v))
true_v = Y1.var(ddof=1) / N1 + Y0.var(ddof=1) / (N - N1) - (Y1 - Y0).var(ddof=1) / N
print(f"true variance {true_v:.3f}, mean Neyman estimate {np.mean(vhat):.3f}, coverage {np.mean(cover):.3f}")
```

## Key takeaways

- Over all randomisations, $\operatorname{Var}(\hat\tau) = S_1^2/n_1 + S_0^2/n_0 - S_\tau^2/n$.
- The last term needs both potential outcomes per unit, so it is dropped: Neyman's estimator is conservative.
- It is exact under constant effects and unbiased for the super-population variance.
