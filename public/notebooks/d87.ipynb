# Day 87: Influence Functions

*Beginner to Ninja: Math for Causal Inference*

The influence function measures how much an estimate moves when one observation is added at a point. It gives the estimator's asymptotic variance, a recipe for standard errors, a measure of robustness to outliers, and, for causal parameters, the doubly robust correction that modern methods are built on.

**Learning outcomes**

- Define the influence function as a Gâteaux derivative and compute it for means, variances, ratios, medians and regression slopes.
- Use asymptotic linearity to derive variances and standard errors from influence functions.
- Relate bounded influence functions to robustness against outliers.
- Write the influence function of the ATE and see the AIPW estimator as plug-in plus mean influence.

## Theory

### Definition

Think of an estimator as a functional $T(F)$ of the data distribution $F$ (the mean $T(F) = \int x\,dF$, for instance). The **influence function** at a point $x$ is

$$
IF(x) = \lim_{\varepsilon \to 0}\frac{T\big((1 - \varepsilon)F + \varepsilon\delta_x\big) - T(F)}{\varepsilon},
$$

the effect of moving a little probability mass to $x$. It satisfies $E_F[IF(X)] = 0$.

### Asymptotic linearity

Most estimators behave like averages of their influence function:

$$
\sqrt n(\hat\theta - \theta) = \frac{1}{\sqrt n}\sum_{i=1}^nIF(X_i) + o_p(1) \;\Rightarrow\; \hat\theta \approx N\Big(\theta, \frac{E[IF(X)^2]}{n}\Big).
$$

So a standard error is $\widehat{\operatorname{SD}}(\widehat{IF})/\sqrt n$. This unifies the delta method (Day 12) and sandwich variances (Day 29).

### A small catalogue

| Estimand | Influence function |
|---|---|
| Mean $\mu$ | $x - \mu$ |
| Variance $\sigma^2$ | $(x - \mu)^2 - \sigma^2$ |
| Ratio $R = \mu_Y/\mu_X$ | $(y - Rx)/\mu_X$ |
| Median $m$ | $\operatorname{sign}(x - m)/(2f(m))$ |
| OLS slope (one regressor) | $(x - \mu_x)(y - \alpha - \beta x)/\operatorname{Var}(X)$ |

### Robustness

An unbounded influence function (mean, OLS) means one extreme observation can move the estimate arbitrarily; a bounded one (median) limits any single point's impact.

### Causal parameters

For the ATE under unconfoundedness, with $\mu_d(x) = E[Y \mid D = d, X = x]$ and $e(x)$ the propensity, the (efficient) influence function is

$$
\varphi(W) = \mu_1(X) - \mu_0(X) + \frac{D(Y - \mu_1(X))}{e(X)} - \frac{(1 - D)(Y - \mu_0(X))}{1 - e(X)} - \tau.
$$

The AIPW estimator (Day 45) is the plug-in estimate $\frac1n\sum(\hat\mu_1 - \hat\mu_0)$ **plus** the average of the correction terms: a "one-step" correction of a plug-in estimator by its estimated influence function. Days 88–91 build on this.

Derive the influence function of $\sigma^2 = E[X^2] - \mu^2$.

### Exercise 8 · A ratio by the chain rule

Derive the influence function of $R = \mu_Y/\mu_X$.

<details><summary>Show solution</summary>

**Step 1.** Influence functions of the two means: $y - \mu_Y$ and $x - \mu_X$.

**Step 2.** Chain rule: $IF_R = \frac{1}{\mu_X}(y - \mu_Y) - \frac{\mu_Y}{\mu_X^2}(x - \mu_X)$.

**Step 3.** $= \frac{y - Rx}{\mu_X}$ (the $R\mu_X$ and $\mu_Y$ terms cancel). $\blacksquare$

</details>

### Exercise 9 · The OLS slope

For $\beta = \operatorname{Cov}(X, Y)/\operatorname{Var}(X)$, show $IF(x, y) = (x - \mu_x)(y - \alpha - \beta x)/\operatorname{Var}(X)$.

<details><summary>Show solution</summary>

**Step 1.** Influence of $\operatorname{Cov}(X, Y)$: $(x - \mu_x)(y - \mu_y) - \operatorname{Cov}(X, Y)$; of $\operatorname{Var}(X)$: $(x - \mu_x)^2 - \operatorname{Var}(X)$.

**Step 2.** Ratio rule: $IF = \frac{(x - \mu_x)(y - \mu_y) - \operatorname{Cov}}{\operatorname{Var}} - \beta\frac{(x - \mu_x)^2 - \operatorname{Var}}{\operatorname{Var}}$.

**Step 3.** The constants cancel ($\operatorname{Cov}/\operatorname{Var} = \beta$); collect: $\frac{(x - \mu_x)\big((y - \mu_y) - \beta(x - \mu_x)\big)}{\operatorname{Var}(X)}$, and $(y - \mu_y) - \beta(x - \mu_x) = y - \alpha - \beta x$. $\blacksquare$

</details>

### Exercise 10 · The ATE influence function has mean zero

Show $E[\varphi(W)] = 0$ when $\mu_0, \mu_1, e$ are the true functions.

<details><summary>Show solution</summary>

**Step 1.** $E[\mu_1(X) - \mu_0(X)] = \tau$ under unconfoundedness.

**Step 2.** $E\big[D(Y - \mu_1(X))/e(X)\big] = E\big[E[D(Y - \mu_1(X)) \mid X]/e(X)\big] = E\big[e(X)\cdot 0/e(X)\big] = 0$, since $E[Y \mid D = 1, X] = \mu_1(X)$.

**Step 3.** Likewise for the control term. So $E[\varphi] = \tau + 0 - 0 - \tau = 0$. $\blacksquare$

</details>

```python
import numpy as np

print("Ex 3:", (2000 - 40 * 60) / 50, " Ex 4:", 12 / 20, " Ex 12 mean:", round((100 * 20000 + 5e6) / 101))
rng = np.random.default_rng(87)

# Exercise 11: ratio estimator, IF vs bootstrap
x = rng.poisson(50, 200).astype(float); y = 40 * x + rng.normal(0, 300, 200) + rng.gamma(2, 50, 200)
R = y.mean() / x.mean(); ifv = (y - R * x) / x.mean()
boot = [(lambda i: y[i].mean() / x[i].mean())(rng.integers(0, 200, 200)) for _ in range(2000)]
print(f"ratio {R:.3f}  IF SE {ifv.std(ddof=1) / np.sqrt(200):.4f}  bootstrap SE {np.std(boot):.4f}")

# Exercise 9: OLS slope IF reproduces the sandwich SE
n = 1000; X = rng.normal(size=n); Yv = 1 + 2 * X + rng.normal(0, 1 + .5 * np.abs(X), n)
b = np.polyfit(X, Yv, 1); res = Yv - np.polyval(b, X)
ifb = (X - X.mean()) * res / X.var()
print(f"slope {b[0]:.3f}  IF SE {ifb.std() / np.sqrt(n):.4f}  HC0 sandwich {np.sqrt(((X - X.mean()) ** 2 * res ** 2).sum()) / ((X - X.mean()) ** 2).sum():.4f}")

# Exercise 13: AIPW with its influence function
def logit_fit(Z, d, iters=25):
    bb = np.zeros(Z.shape[1])
    for _ in range(iters):
        p = 1 / (1 + np.exp(-Z @ bb)); bb += np.linalg.solve(Z.T @ ((p * (1 - p))[:, None] * Z), Z.T @ (d - p))
    return 1 / (1 + np.exp(-Z @ bb))
def aipw(W, D, Yo):
    Z = np.column_stack([np.ones(len(D)), W]); e = logit_fit(Z, D)
    m1 = Z @ np.linalg.lstsq(Z[D == 1], Yo[D == 1], rcond=None)[0]; m0 = Z @ np.linalg.lstsq(Z[D == 0], Yo[D == 0], rcond=None)[0]
    psi = m1 - m0 + D * (Yo - m1) / e - (1 - D) * (Yo - m0) / (1 - e)
    return psi.mean(), psi.std(ddof=1) / np.sqrt(len(D))
n = 2000; W = rng.normal(size=(n, 2)); D = (rng.random(n) < 1 / (1 + np.exp(-(W @ [.8, -.5])))).astype(float)
Yo = -1 + W @ [.5, .3] + .3 * D + rng.normal(0, 1, n)
est, se = aipw(W, D, Yo)
bs = [aipw(W[i], D[i], Yo[i])[0] for i in (rng.integers(0, n, n) for _ in range(300))]
print(f"AIPW {est:.3f}  IF SE {se:.4f}  bootstrap SE {np.std(bs):.4f}  (truth 0.3)")
```

## Key takeaways

- The influence function is the derivative of an estimator with respect to the data distribution; estimators are approximately averages of it.
- Its variance gives the estimator's variance and a standard error without resampling; its boundedness measures robustness.
- The ATE's influence function contains the inverse-weighted residual terms; adding their average to a plug-in estimate gives AIPW.
