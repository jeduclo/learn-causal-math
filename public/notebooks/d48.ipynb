# Day 48: Chains, Forks & Colliders

*Beginner to Ninja: Math for Causal Inference*

Every path in a DAG is built from three junctions. Chains and forks transmit association unless you condition on their middle node; colliders block it unless you do. This asymmetry drives everything in Phase 4.

**Learning outcomes**

- Classify each three-node junction on a path as a chain, fork or collider.
- Compute correlations in linear models by multiplying path coefficients.
- Prove that conditioning on a chain or fork middle removes association while conditioning on a collider (or its descendant) creates it.
- Recognise collider bias in hiring, hospital and birth-weight data.

## Theory

### Three junctions

Any path through a middle node $M$ looks like one of these:

| Name | Shape | Unconditionally | Conditioning on $M$ |
|---|---|---|---|
| **Chain** (mediator) | $A \to M \to B$ | $A, B$ associated | association **blocked** |
| **Fork** (confounder) | $A \leftarrow M \to B$ | $A, B$ associated | association **blocked** |
| **Collider** | $A \to M \leftarrow B$ | $A, B$ **independent** | association **created** |

Conditioning on a **descendant** of a collider also opens it, partially.

### Intuition

- **Chain:** job training raises skills, skills raise wages. Among workers with the same skills, training tells you nothing more about wages.
- **Fork:** income raises both gym membership and health. Among people with the same income, gym membership and health are unrelated (in this story).
- **Collider:** ability and connections both help land a job at a top firm. Among employees, a worker with weak connections must have had high ability to get in: the two become *negatively* related.

### Linear path analysis

Take standardised variables (mean 0, variance 1) in a linear model where each arrow has a coefficient. Sewall Wright's **path-tracing rule**: the correlation between two variables is the sum, over all paths **without colliders**, of the products of the coefficients along the path.

Chain $Z \xrightarrow{a} X \xrightarrow{b} Y$: $\operatorname{corr}(Z, Y) = ab$. Fork $Z \xleftarrow{a} X \xrightarrow{b} Y$: $\operatorname{corr}(Z, Y) = ab$. Collider $Z \to X \leftarrow Y$: $\operatorname{corr}(Z, Y) = 0$.

### Conditioning in the Gaussian case

For jointly normal variables, conditioning on $X$ leaves the covariance

$$
\operatorname{Cov}(Z, Y \mid X) = \operatorname{Cov}(Z, Y) - \frac{\operatorname{Cov}(Z, X)\operatorname{Cov}(X, Y)}{\operatorname{Var}(X)},
$$

which is what a regression of $Y$ on $Z$ and $X$ estimates (Frisch–Waugh–Lovell, Day 28). For the collider $X = Z + Y + e$ the first term is 0 and the second is non-zero, so conditioning *creates* a covariance.

### Ways of "conditioning"

Conditioning happens whenever you stratify, add a variable to a regression, match on it, or **select the sample** on it: studying only hospital patients, only employed workers, only firms that survived, only people who answered a survey. Selection is conditioning you may not notice (Day 57).

### Exercise 6 · The linear chain

Let $X = aZ + e_X$ and $Y = bX + e_Y$, with $Z, e_X, e_Y$ independent, mean 0. Prove $\operatorname{Cov}(Z, Y) = ab\operatorname{Var}(Z)$ and $\operatorname{Cov}(Z, Y \mid X) = 0$ (in the partial-covariance sense).

<details><summary>Show solution</summary>

**Step 1.** Substitute: $Y = abZ + be_X + e_Y$.

**Step 2.** $\operatorname{Cov}(Z, Y) = ab\operatorname{Var}(Z)$, since $Z$ is independent of $e_X, e_Y$.

**Step 3.** $\operatorname{Cov}(Z, X) = a\operatorname{Var}(Z)$ and $\operatorname{Cov}(X, Y) = b\operatorname{Var}(X)$.

**Step 4.** Partial covariance:

$$
\operatorname{Cov}(Z, Y) - \frac{\operatorname{Cov}(Z, X)\operatorname{Cov}(X, Y)}{\operatorname{Var}(X)} = ab\operatorname{Var}(Z) - \frac{a\operatorname{Var}(Z)\,b\operatorname{Var}(X)}{\operatorname{Var}(X)} = 0. \quad \blacksquare
$$

</details>

### Exercise 7 · The linear fork

Let $Z = aX + e_Z$ and $Y = bX + e_Y$ with $X, e_Z, e_Y$ independent. Prove $\operatorname{Cov}(Z, Y) = ab\operatorname{Var}(X)$ and the partial covariance given $X$ is 0.

<details><summary>Show solution</summary>

**Step 1.** $\operatorname{Cov}(Z, Y) = \operatorname{Cov}(aX + e_Z, bX + e_Y) = ab\operatorname{Var}(X)$.

**Step 2.** $\operatorname{Cov}(Z, X) = a\operatorname{Var}(X)$ and $\operatorname{Cov}(X, Y) = b\operatorname{Var}(X)$.

**Step 3.** Partial covariance: $ab\operatorname{Var}(X) - \dfrac{a\operatorname{Var}(X)\,b\operatorname{Var}(X)}{\operatorname{Var}(X)} = 0$. $\blacksquare$

</details>

### Exercise 8 · Conditioning on a collider

Let $Z, Y$ be independent $N(0, 1)$ and $X = Z + Y + e$ with $e \sim N(0, \sigma^2)$ independent. Prove that the partial correlation of $Z$ and $Y$ given $X$ is $-1/(1 + \sigma^2)$.

<details><summary>Show solution</summary>

**Step 1.** $\operatorname{Var}(X) = 2 + \sigma^2$, $\operatorname{Cov}(Z, X) = \operatorname{Cov}(Y, X) = 1$, $\operatorname{Cov}(Z, Y) = 0$.

**Step 2.** Partial covariance: $0 - \dfrac{1 \cdot 1}{2 + \sigma^2} = -\dfrac{1}{2 + \sigma^2}$.

**Step 3.** Partial variances: $\operatorname{Var}(Z \mid X) = \operatorname{Var}(Y \mid X) = 1 - \dfrac{1}{2 + \sigma^2} = \dfrac{1 + \sigma^2}{2 + \sigma^2}$.

**Step 4.** Partial correlation:

$$
\frac{-1/(2 + \sigma^2)}{(1 + \sigma^2)/(2 + \sigma^2)} = -\frac{1}{1 + \sigma^2}. \quad \blacksquare
$$

With $\sigma^2 = 1$ it is −0.5; with no noise it is −1 (Exercise 4).

</details>

### Exercise 9 · Conditioning on a collider's descendant

In Exercise 8 add $W = X + u$ with $u \sim N(0, \tau^2)$ independent. Prove the partial correlation of $Z$ and $Y$ given $W$ is $-1/(1 + \sigma^2 + \tau^2)$.

<details><summary>Show solution</summary>

**Step 1.** $W = Z + Y + e + u$, so $\operatorname{Var}(W) = 2 + \sigma^2 + \tau^2$ and $\operatorname{Cov}(Z, W) = \operatorname{Cov}(Y, W) = 1$.

**Step 2.** This has exactly the form of Exercise 8 with noise variance $\sigma^2 + \tau^2$ in place of $\sigma^2$.

**Step 3.** So the partial correlation is $-1/(1 + \sigma^2 + \tau^2)$. $\blacksquare$

A noisy descendant opens the collider partially: weaker, but not zero.

</details>

### Exercise 10 · Regression in a chain

In the chain of Exercise 6, regress $Y$ on both $Z$ and $X$. Prove that the population coefficient on $Z$ is 0.

<details><summary>Show solution</summary>

**Step 1.** By Frisch–Waugh–Lovell, the coefficient on $Z$ equals $\operatorname{Cov}(\tilde Z, Y)/\operatorname{Var}(\tilde Z)$, where $\tilde Z$ is $Z$ residualised on $X$.

**Step 2.** $\operatorname{Cov}(\tilde Z, Y)$ is the partial covariance of $Z$ and $Y$ given $X$.

**Step 3.** By Exercise 6 it is 0, so the coefficient is 0. $\blacksquare$ Controlling for a mediator removes the whole effect that passes through it.

### Exercise 12 · The birth-weight paradox

Maternal smoking $S$ (30% of mothers) and a serious birth defect $D$ (5%) are independent. Both cause low birth weight $L$: $P(L = 1 \mid S, D)$ is 0.05 (neither), 0.15 (smoking only), 0.6 (defect only), 0.7 (both). Infant mortality is $P(M = 1 \mid L, D) = 0.005 + 0.02L + 0.10D$; smoking affects mortality only through birth weight. Compare mortality for babies of smokers and non-smokers (a) overall and (b) among low-birth-weight babies.

<details><summary>Show solution</summary>

**Step 1.** Overall, $P(L = 1 \mid S = 1) = 0.95(0.15) + 0.05(0.7) = 0.1775$ and $P(L = 1 \mid S = 0) = 0.95(0.05) + 0.05(0.6) = 0.0775$.

**Step 2.** Overall mortality: $S = 1$: $0.005 + 0.02(0.1775) + 0.10(0.05) = 0.01355$; $S = 0$: $0.005 + 0.02(0.0775) + 0.005 = 0.01155$. Smoking raises mortality.

**Step 3.** Among low-birth-weight babies, use Bayes: $P(D = 1 \mid L = 1, S = 1) = \dfrac{0.05(0.7)}{0.05(0.7) + 0.95(0.15)} = \dfrac{0.035}{0.1775} \approx 0.197$.

**Step 4.** $P(D = 1 \mid L = 1, S = 0) = \dfrac{0.05(0.6)}{0.05(0.6) + 0.95(0.05)} = \dfrac{0.03}{0.0775} \approx 0.387$.

**Step 5.** Mortality among low-birth-weight babies: $S = 1$: $0.025 + 0.10(0.197) \approx 0.0447$; $S = 0$: $0.025 + 0.10(0.387) \approx 0.0637$.

**Answer:** among low-birth-weight babies, those of smokers die *less* often. Low birth weight is a collider; a small baby of a non-smoker is more likely to have a defect. Smoking is harmful overall.

</details>

## Verify in Python

*Your work:*

```python
import numpy as np

def pcorr(a, b, c):
    """Partial correlation of a and b given c (residualise both on c)."""
    C = np.column_stack([np.ones_like(c), c])
    ra = a - C @ np.linalg.lstsq(C, a, rcond=None)[0]
    rb = b - C @ np.linalg.lstsq(C, b, rcond=None)[0]
    return np.corrcoef(ra, rb)[0, 1]

rng = np.random.default_rng(48)
n = 1_000_000
# chain, fork and collider with standardised coefficients
Z = rng.normal(size=n); X = .5 * Z + np.sqrt(1 - .25) * rng.normal(size=n); Y = .6 * X + np.sqrt(1 - .36) * rng.normal(size=n)
print(f"chain:    corr {np.corrcoef(Z, Y)[0, 1]:.3f} (0.30)  given X {pcorr(Z, Y, X):.3f}")
X = rng.normal(size=n); Z = .7 * X + np.sqrt(1 - .49) * rng.normal(size=n); Y = -.4 * X + np.sqrt(1 - .16) * rng.normal(size=n)
print(f"fork:     corr {np.corrcoef(Z, Y)[0, 1]:.3f} (-0.28) given X {pcorr(Z, Y, X):.3f}")
Z = rng.normal(size=n); Y = rng.normal(size=n); X = Z + Y + rng.normal(size=n); W = X + rng.normal(size=n)
print(f"collider: corr {np.corrcoef(Z, Y)[0, 1]:.3f}  given X {pcorr(Z, Y, X):.3f} (-1/2)  given W {pcorr(Z, Y, W):.3f} (-1/3)")

# Exercise 11: hiring
A = rng.normal(size=n); C = rng.normal(size=n); hired = A + C > 1.5
print(f"hired {hired.mean():.3f}; corr(A, C) all {np.corrcoef(A, C)[0, 1]:.3f}, hired {np.corrcoef(A[hired], C[hired])[0, 1]:.3f}")

# Exercise 12: birth-weight paradox (exact)
pL = {(0, 0): .05, (1, 0): .15, (0, 1): .6, (1, 1): .7}          # key (s, d)
mort = lambda l, d: .005 + .02 * l + .10 * d
for s in (0, 1):
    overall = sum((.05 if d else .95) * (pL[s, d] * mort(1, d) + (1 - pL[s, d]) * mort(0, d)) for d in (0, 1))
    pd = .05 * pL[s, 1] / (.05 * pL[s, 1] + .95 * pL[s, 0])
    print(f"smoker={s}: mortality overall {overall:.5f}; P(defect|LBW) {pd:.3f}; mortality | LBW {mort(1, 0) + .10 * pd:.4f}")
```

## Key takeaways

- Chains and forks transmit association; conditioning on their middle node blocks it.
- Colliders block association; conditioning on them, or on their descendants, or selecting the sample on them, creates it.
- In linear models, correlations are sums of products of path coefficients over collider-free paths.
